```python

import pandas as pd
import numpy as np

recovery_path = (
    DATA_DIR / "Delta_2023_2025_Recovery_Model.csv"
)

timing_cols = [
    "YEAR",
    "FL_DATE",
    "CRS_DEP_MINUTES",
    "PREV_CRS_ARR_MINUTES",
    "PREV_ARR_DELAY",
    "SCHED_GROUND_MIN",
    "ARR_DELAY_NEW"
]

timing = pd.read_csv(
    recovery_path,
    usecols=timing_cols,
    low_memory=False
)

# Estimate incoming aircraft's actual arrival
# relative to its scheduled arrival.
timing["EST_PREV_ACTUAL_ARR"] = (
    timing["PREV_CRS_ARR_MINUTES"]
    + timing["PREV_ARR_DELAY"]
)

# Compare with the current flight's scheduled departure.
timing["TIME_BEFORE_SCHED_DEP"] = (
    timing["CRS_DEP_MINUTES"]
    - timing["EST_PREV_ACTUAL_ARR"]
)

# Check whether scheduled ground time agrees with
# the difference between scheduled flight times.
timing["SCHED_TIME_DIFFERENCE"] = (
    timing["CRS_DEP_MINUTES"]
    - timing["PREV_CRS_ARR_MINUTES"]
)

timing["GROUND_TIME_DISCREPANCY"] = (
    timing["SCHED_TIME_DIFFERENCE"]
    - timing["SCHED_GROUND_MIN"]
)

print("=" * 65)
print("RECOVERY DECISION-TIME AUDIT")
print("=" * 65)

print("\nTotal recovery flights:", len(timing))

print("\nIncoming arrival relative to scheduled departure:")
print(
    "Arrived by scheduled departure:",
    (timing["TIME_BEFORE_SCHED_DEP"] >= 0).sum()
)

print(
    "Arrived after scheduled departure:",
    (timing["TIME_BEFORE_SCHED_DEP"] < 0).sum()
)

print("\nTiming distribution:")
print(
    timing["TIME_BEFORE_SCHED_DEP"].describe(
        percentiles=[.01, .05, .25, .5, .75, .95, .99]
    )
)

print("\nScheduled ground-time consistency:")
print(
    timing["GROUND_TIME_DISCREPANCY"]
    .value_counts()
    .head(10)
)

print("\nAnnual timing feasibility:")
print(
    timing.groupby("YEAR").agg(
        flights=("YEAR", "size"),
        arrived_by_schedule=(
            "TIME_BEFORE_SCHED_DEP",
            lambda x: (x >= 0).sum()
        ),
        arrived_after_schedule=(
            "TIME_BEFORE_SCHED_DEP",
            lambda x: (x < 0).sum()
        )
    )
)

print("\nDECISION-TIME AUDIT COMPLETE")
```

```python

import numpy as np
import pandas as pd

print("=" * 65)
print("RECOVERY TIMING — CONSISTENCY VALIDATION")
print("=" * 65)

timing["REMAINING_GROUND_MIN"] = (
    timing["SCHED_GROUND_MIN"]
    - timing["PREV_ARR_DELAY"]
)

timing["TIMING_DIFFERENCE"] = (
    timing["REMAINING_GROUND_MIN"]
    - timing["TIME_BEFORE_SCHED_DEP"]
)

print("\nMaximum absolute calculation difference:")
print(timing["TIMING_DIFFERENCE"].abs().max())

print("\nNon-matching rows:")
print((timing["TIMING_DIFFERENCE"].abs() > 0.001).sum())

print("\nRemaining ground-time distribution:")
print(
    timing["REMAINING_GROUND_MIN"].describe(
        percentiles=[.01, .05, .25, .5, .75, .95, .99]
    )
)

print("\nRecovery timing categories:")

conditions = [
    timing["REMAINING_GROUND_MIN"] < 0,
    timing["REMAINING_GROUND_MIN"].between(0, 29),
    timing["REMAINING_GROUND_MIN"].between(30, 59),
    timing["REMAINING_GROUND_MIN"] >= 60
]

labels = [
    "Incoming after scheduled departure",
    "0–29 minutes remaining",
    "30–59 minutes remaining",
    "60+ minutes remaining"
]

timing["RECOVERY_TIME_CATEGORY"] = np.select(
    conditions,
    labels,
    default="Unknown"
)

summary = (
    timing.groupby("RECOVERY_TIME_CATEGORY", observed=True)
    .agg(
        flights=("ARR_DELAY_NEW", "size"),
        mean_arr_delay=("ARR_DELAY_NEW", "mean"),
        median_arr_delay=("ARR_DELAY_NEW", "median"),
        delay_15_rate=(
            "ARR_DELAY_NEW",
            lambda x: (x >= 15).mean() * 100
        )
    )
)

summary["share_pct"] = (
    summary["flights"] / len(timing) * 100
)

display(summary.round(2))

print("\nTIMING VALIDATION COMPLETE")
```

```python

# ================================================================
# BUSI6029 — ASSESSMENT 2
# STEP 2F: EXTREME DELAY AND TARGET DISTRIBUTION AUDIT
# ================================================================
# Purpose:
# 1. Examine the distribution of downstream arrival delays.
# 2. Identify extreme incoming-aircraft delay observations.
# 3. Assess whether extreme cases warrant sensitivity analysis.
#
# Dataset: Delta_2023_2025_Recovery_Model.csv
# Target: ARR_DELAY_NEW
# Note: This step is descriptive; no rows are removed.
# ================================================================

import pandas as pd

# ------------------------------------------------
# 2F.1 — Downstream arrival-delay distribution
# ------------------------------------------------

print("=" * 65)
print("STEP 2F — EXTREME DELAY AND TARGET DISTRIBUTION AUDIT")
print("=" * 65)

print("\nDownstream arrival-delay distribution:")
print(
    timing["ARR_DELAY_NEW"].describe(
        percentiles=[.01, .05, .25, .5, .75, .90, .95, .99, .999]
    )
)

# ------------------------------------------------
# 2F.2 — Extreme incoming-aircraft delays
# ------------------------------------------------

thresholds = [60, 120, 180, 300, 600]

extreme_summary = []

for threshold in thresholds:
    mask = timing["PREV_ARR_DELAY"] >= threshold

    extreme_summary.append({
        "Incoming delay threshold": f"{threshold}+ min",
        "Flights": int(mask.sum()),
        "Share (%)": round(mask.mean() * 100, 3),
        "Mean target delay": round(
            timing.loc[mask, "ARR_DELAY_NEW"].mean(), 2
        )
    })

print("\nExtreme incoming-aircraft delays:")
display(pd.DataFrame(extreme_summary))

# ------------------------------------------------
# 2F.3 — Extreme downstream arrival delays
# ------------------------------------------------

target_thresholds = [15, 60, 120, 180, 300, 600]

target_summary = []

for threshold in target_thresholds:
    mask = timing["ARR_DELAY_NEW"] >= threshold

    target_summary.append({
        "Target delay threshold": f"{threshold}+ min",
        "Flights": int(mask.sum()),
        "Share (%)": round(mask.mean() * 100, 3)
    })

print("\nExtreme downstream arrival delays:")
display(pd.DataFrame(target_summary))

# ------------------------------------------------
# 2F.4 — Completion
# ------------------------------------------------

print("\nSTEP 2F AUDIT COMPLETE")
print("No records were removed or modified.")
```

```python

# ================================================================
# BUSI6029 — ASSESSMENT 2
# STEP 2G: FEATURE ELIGIBILITY AND LEAKAGE AUDIT
# ================================================================
# Purpose:
# 1. Define predictors available at the recovery decision point.
# 2. Explicitly identify prohibited outcome/leakage variables.
# 3. Check missing values in candidate predictors.
# 4. Confirm the temporal modelling population.
#
# No model is trained and no CSV is modified.
# ================================================================

import pandas as pd

# ------------------------------------------------
# 2G.1 — Candidate prediction-time features
# ------------------------------------------------

schedule_features = [
    "MONTH",
    "DAY_OF_WEEK",
    "IS_WEEKEND",
    "CRS_DEP_HOUR",
    "CRS_ARR_HOUR",
    "CRS_ELAPSED_TIME",
    "DISTANCE",
    "ORIGIN",
    "DEST"
]

recovery_features = [
    "PREV_ARR_DELAY",
    "SCHED_GROUND_MIN"
]

candidate_features = schedule_features + recovery_features

# ------------------------------------------------
# 2G.2 — Explicit leakage exclusion list
# ------------------------------------------------

leakage_features = [
    "DEP_TIME",
    "DEP_DELAY",
    "DEP_DELAY_NEW",
    "DEP_DEL15",
    "TAXI_OUT",
    "WHEELS_OFF",
    "WHEELS_ON",
    "TAXI_IN",
    "ARR_TIME",
    "ARR_DELAY",
    "ARR_DEL15",
    "ACTUAL_ELAPSED_TIME",
    "AIR_TIME",
    "CARRIER_DELAY",
    "WEATHER_DELAY",
    "NAS_DELAY",
    "SECURITY_DELAY",
    "LATE_AIRCRAFT_DELAY",
    "SIGNIFICANT_ARR_DELAY",
    "SEVERE_ARR_DELAY",
    "ARR_DELAY_SEVERITY",
    "DISRUPTED"
]

target = "ARR_DELAY_NEW"

print("=" * 65)
print("STEP 2G — FEATURE ELIGIBILITY AND LEAKAGE AUDIT")
print("=" * 65)

print("\nSchedule-only predictors:", len(schedule_features))
print("Additional recovery predictors:", len(recovery_features))
print("Total candidate predictors:", len(candidate_features))

print("\nCandidate/leakage overlap:")
print(set(candidate_features) & set(leakage_features))

print("\nTarget accidentally included:")
print(target in candidate_features)

# ------------------------------------------------
# 2G.3 — Verify features exist in recovery CSV
# ------------------------------------------------

recovery_path = (
    DATA_DIR / "Delta_2023_2025_Recovery_Model.csv"
)

header = pd.read_csv(recovery_path, nrows=0)

missing_columns = [
    col for col in candidate_features
    if col not in header.columns
]

print("\nMissing candidate columns:")
print(missing_columns)

# ------------------------------------------------
# 2G.4 — Read selected columns only
# ------------------------------------------------

audit_cols = [
    "YEAR",
    target
] + candidate_features

feature_audit = pd.read_csv(
    recovery_path,
    usecols=audit_cols,
    low_memory=False
)

print("\nMissing values by candidate feature:")

missing_summary = pd.DataFrame({
    "Missing": feature_audit[candidate_features].isna().sum(),
    "Missing (%)": (
        feature_audit[candidate_features].isna().mean() * 100
    ).round(3)
})

display(missing_summary)

# ------------------------------------------------
# 2G.5 — Confirm chronological split counts
# ------------------------------------------------

print("\nTemporal modelling split:")

print(
    "Training period (2023–2024):",
    (feature_audit["YEAR"] <= 2024).sum()
)

print(
    "Holdout test period (2025):",
    (feature_audit["YEAR"] == 2025).sum()
)

print("\nSTEP 2G AUDIT COMPLETE")
```

```python

# ================================================================
# BUSI6029 — ASSESSMENT 2
# STEP 2H: CHRONOLOGICAL DATA PARTITIONING
# ================================================================
# Purpose:
# 1. Separate training (2023), validation (2024), test (2025).
# 2. Preserve the same population for fair model comparisons.
# 3. Confirm no year overlaps or missing targets.
#
# No model is trained and no CSV is modified.
# ================================================================

# 2H.1 — Define chronological partitions

train_mask = feature_audit["YEAR"] == 2023
val_mask   = feature_audit["YEAR"] == 2024
test_mask  = feature_audit["YEAR"] == 2025

# 2H.2 — Summarise partition sizes and target behaviour

split_summary = []

for name, mask in [
    ("Training", train_mask),
    ("Validation", val_mask),
    ("Holdout test", test_mask)
]:
    y = feature_audit.loc[mask, target]

    split_summary.append({
        "Partition": name,
        "Year": int(feature_audit.loc[mask, "YEAR"].iloc[0]),
        "Flights": len(y),
        "Mean delay (min)": round(y.mean(), 2),
        "Median delay (min)": round(y.median(), 2),
        "15+ min delay (%)": round((y >= 15).mean() * 100, 2),
        "Missing target": int(y.isna().sum())
    })

print("=" * 65)
print("STEP 2H — CHRONOLOGICAL DATA PARTITIONING")
print("=" * 65)

display(pd.DataFrame(split_summary))

# 2H.3 — Validate partition completeness

total_partitioned = (
    train_mask.sum() + val_mask.sum() + test_mask.sum()
)

print("\nTotal partitioned:", total_partitioned)
print("Total available:", len(feature_audit))

print(
    "All rows assigned:",
    total_partitioned == len(feature_audit)
)

print(
    "No overlapping partitions:",
    not (
        (train_mask & val_mask).any()
        or (train_mask & test_mask).any()
        or (val_mask & test_mask).any()
    )
)

print("\nSTEP 2H COMPLETE")
```

```python

# ================================================================
# BUSI6029 — ASSESSMENT 2
# STEP 2I: MODELLING FEATURE MATRIX PREPARATION
# ================================================================
# Purpose:
# 1. Prepare schedule-only and recovery-aware feature sets.
# 2. Separate categorical and numerical predictors.
# 3. Create chronological training, validation and test arrays.
# 4. Confirm feature alignment and target integrity.
#
# No model is trained and no source CSV is modified.
# ================================================================

# 2I.1 — Define categorical and numerical predictors

categorical_features = ["ORIGIN", "DEST"]

schedule_numeric_features = [
    col for col in schedule_features
    if col not in categorical_features
]

recovery_numeric_features = (
    schedule_numeric_features + recovery_features
)

# 2I.2 — Build feature matrices

X_train_schedule = feature_audit.loc[
    train_mask, schedule_features
].copy()

X_val_schedule = feature_audit.loc[
    val_mask, schedule_features
].copy()

X_test_schedule = feature_audit.loc[
    test_mask, schedule_features
].copy()

X_train_recovery = feature_audit.loc[
    train_mask, candidate_features
].copy()

X_val_recovery = feature_audit.loc[
    val_mask, candidate_features
].copy()

X_test_recovery = feature_audit.loc[
    test_mask, candidate_features
].copy()

# 2I.3 — Build target arrays

y_train = feature_audit.loc[train_mask, target].copy()
y_val = feature_audit.loc[val_mask, target].copy()
y_test = feature_audit.loc[test_mask, target].copy()

# 2I.4 — Verify dimensions and alignment

print("=" * 65)
print("STEP 2I — MODELLING FEATURE MATRIX PREPARATION")
print("=" * 65)

print("\nSchedule-only feature dimensions:")
print("Training:", X_train_schedule.shape)
print("Validation:", X_val_schedule.shape)
print("Test:", X_test_schedule.shape)

print("\nRecovery-aware feature dimensions:")
print("Training:", X_train_recovery.shape)
print("Validation:", X_val_recovery.shape)
print("Test:", X_test_recovery.shape)

print("\nTarget dimensions:")
print("Training:", y_train.shape)
print("Validation:", y_val.shape)
print("Test:", y_test.shape)

print("\nCategorical features:", categorical_features)
print("Schedule numerical features:", schedule_numeric_features)
print("Recovery numerical features:", recovery_numeric_features)

print("\nTarget excluded from predictors:",
      target not in X_train_recovery.columns)

print("Training index aligned:",
      X_train_recovery.index.equals(y_train.index))

print("Validation index aligned:",
      X_val_recovery.index.equals(y_val.index))

print("Test index aligned:",
      X_test_recovery.index.equals(y_test.index))

print("\nSTEP 2I COMPLETE")
```

```python

# ================================================================
# BUSI6029 — ASSESSMENT 2
# STEP 2J: NAÏVE PREDICTION BASELINES
# ================================================================
# Purpose:
# 1. Establish simple reference models.
# 2. Evaluate performance on 2024 validation data.
# 3. Quantify errors during significant disruptions.
# 4. Create a reusable results table for later models.
#
# No 2025 holdout data is used for model selection.
# ================================================================

from sklearn.metrics import mean_absolute_error
from sklearn.metrics import mean_squared_error, r2_score
import numpy as np

# ------------------------------------------------
# 2J.1 — Define baseline predictions
# ------------------------------------------------

train_mean_delay = y_train.mean()

baseline_predictions = {
    "Zero-delay baseline": np.zeros(len(y_val)),
    "Training-mean baseline": np.full(
        len(y_val), train_mean_delay
    )
}

# ------------------------------------------------
# 2J.2 — Evaluate baseline performance
# ------------------------------------------------

baseline_results = []

for model_name, predictions in baseline_predictions.items():

    mae = mean_absolute_error(y_val, predictions)

    rmse = np.sqrt(
        mean_squared_error(y_val, predictions)
    )

    r2 = r2_score(y_val, predictions)

    severe_mask = y_val >= 15

    severe_mae = mean_absolute_error(
        y_val[severe_mask],
        predictions[severe_mask.to_numpy()]
    )

    baseline_results.append({
        "Model": model_name,
        "MAE": round(mae, 3),
        "RMSE": round(rmse, 3),
        "R2": round(r2, 4),
        "MAE on 15+ min delays": round(severe_mae, 3)
    })

# ------------------------------------------------
# 2J.3 — Display comparison
# ------------------------------------------------

print("=" * 65)
print("STEP 2J — NAÏVE PREDICTION BASELINES")
print("=" * 65)

print(f"\n2023 training mean delay: {train_mean_delay:.3f} min")
print(f"2024 validation mean delay: {y_val.mean():.3f} min")

baseline_results_df = pd.DataFrame(baseline_results)

display(baseline_results_df)

print("\nSTEP 2J COMPLETE")
```

```python

# ================================================================
# BUSI6029 — ASSESSMENT 2
# STEP 2K: LEAKAGE-SAFE FEATURE PREPROCESSING
# ================================================================
# Purpose:
# 1. One-hot encode airport categories.
# 2. Scale numerical features for linear regression.
# 3. Fit transformations using 2023 training data only.
# 4. Verify 2024 validation data can be transformed.
#
# No predictive model is trained in this step.
# ================================================================

from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder, StandardScaler
from sklearn.pipeline import Pipeline

# ------------------------------------------------
# 2K.1 — Define preprocessing configuration
# ------------------------------------------------

linear_preprocessor = ColumnTransformer(
    transformers=[
        (
            "numeric",
            StandardScaler(),
            recovery_numeric_features
        ),
        (
            "categorical",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=True
            ),
            categorical_features
        )
    ],
    remainder="drop",
    sparse_threshold=1.0
)

# ------------------------------------------------
# 2K.2 — Fit preprocessing on training data only
# ------------------------------------------------

print("=" * 65)
print("STEP 2K — LEAKAGE-SAFE FEATURE PREPROCESSING")
print("=" * 65)

X_train_transformed = linear_preprocessor.fit_transform(
    X_train_recovery
)

# ------------------------------------------------
# 2K.3 — Transform validation data
# ------------------------------------------------

X_val_transformed = linear_preprocessor.transform(
    X_val_recovery
)

# ------------------------------------------------
# 2K.4 — Validate transformed dimensions
# ------------------------------------------------

print("\nOriginal training shape:", X_train_recovery.shape)
print("Transformed training shape:", X_train_transformed.shape)

print("\nOriginal validation shape:", X_val_recovery.shape)
print("Transformed validation shape:", X_val_transformed.shape)

print("\nNumber of transformed features:")
print(len(linear_preprocessor.get_feature_names_out()))

print("\nTraining and validation feature counts match:")
print(
    X_train_transformed.shape[1]
    == X_val_transformed.shape[1]
)

print("\nTransformed data types:")
print("Training:", type(X_train_transformed).__name__)
print("Validation:", type(X_val_transformed).__name__)

print("\nSTEP 2K COMPLETE")
```

```python

# ================================================================
## Step 2L — Model 1: Ridge Regression (Recovery-Aware)
# ================================================================
# Purpose:
# 1. Train the first recovery-aware predictive model.
# 2. Evaluate predictions on 2024 validation data.
# 3. Compare performance against naive baselines.
# 4. Assess prediction errors for significant disruptions.
#
# Training: 2023
# Validation: 2024
# Final holdout: 2025 (not accessed)
# ================================================================

from sklearn.linear_model import Ridge

# ------------------------------------------------
# 2L.1 — Initialise and train Ridge Regression
# ------------------------------------------------

ridge_model = Ridge(
    alpha=10.0,
    solver="lsqr",
    max_iter=2000,
    tol=1e-4
)

print("=" * 65)
print("STEP 2L — MODEL 1: RIDGE REGRESSION")
print("=" * 65)

print("\nTraining Ridge Regression...")

ridge_model.fit(
    X_train_transformed,
    y_train
)

print("Model training complete.")

# ------------------------------------------------
# 2L.2 — Predict validation outcomes
# ------------------------------------------------

ridge_predictions_raw = ridge_model.predict(
    X_val_transformed
)

# Arrival-delay target is non-negative.
# Clip negative predictions to maintain target support.

ridge_predictions = np.maximum(
    ridge_predictions_raw,
    0
)

# ------------------------------------------------
# 2L.3 — Calculate validation metrics
# ------------------------------------------------

ridge_mae = mean_absolute_error(
    y_val,
    ridge_predictions
)

ridge_rmse = np.sqrt(
    mean_squared_error(y_val, ridge_predictions)
)

ridge_r2 = r2_score(
    y_val,
    ridge_predictions
)

severe_mask = y_val >= 15

ridge_severe_mae = mean_absolute_error(
    y_val[severe_mask],
    ridge_predictions[severe_mask.to_numpy()]
)

# ------------------------------------------------
# 2L.4 — Store results for model comparison
# ------------------------------------------------

ridge_results = {
    "Model": "Ridge Regression (Recovery)",
    "MAE": round(ridge_mae, 3),
    "RMSE": round(ridge_rmse, 3),
    "R2": round(ridge_r2, 4),
    "MAE on 15+ min delays": round(
        ridge_severe_mae, 3
    )
}

print("\n2024 validation performance:")
display(pd.DataFrame([ridge_results]))

# ------------------------------------------------
# 2L.5 — Compare against baseline models
# ------------------------------------------------

comparison_df = pd.concat(
    [
        baseline_results_df,
        pd.DataFrame([ridge_results])
    ],
    ignore_index=True
)

print("\nBaseline versus Ridge Regression:")
display(comparison_df)

# ------------------------------------------------
# 2L.6 — Prediction sanity checks
# ------------------------------------------------

print("\nPrediction diagnostics:")

print(
    "Raw negative predictions:",
    int((ridge_predictions_raw < 0).sum())
)

print(
    "Mean predicted delay:",
    round(ridge_predictions.mean(), 3)
)

print(
    "Mean actual delay:",
    round(y_val.mean(), 3)
)

print(
    "Maximum predicted delay:",
    round(ridge_predictions.max(), 3)
)

print("\nSTEP 2L COMPLETE")
```

```python

# ================================================================
# BUSI6029 — ASSESSMENT 2
# STEP 2N: MODEL 2 — RANDOM FOREST REGRESSION
# ================================================================
# Purpose:
# 1. Train a nonlinear recovery-aware predictive model.
# 2. Capture interactions and nonlinear delay propagation.
# 3. Evaluate performance on 2024 validation data.
# 4. Compare Random Forest with Ridge and naive baselines.
#
# Training: 2023
# Validation: 2024
# Final holdout: 2025 (not accessed)
# ================================================================

from sklearn.ensemble import RandomForestRegressor

# ------------------------------------------------
# 2N.1 — Configure tree-based preprocessing
# ------------------------------------------------

tree_preprocessor = ColumnTransformer(
    transformers=[
        (
            "numeric",
            "passthrough",
            recovery_numeric_features
        ),
        (
            "categorical",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=True
            ),
            categorical_features
        )
    ],
    remainder="drop",
    sparse_threshold=1.0
)

# ------------------------------------------------
# 2N.2 — Create Random Forest pipeline
# ------------------------------------------------

rf_pipeline = Pipeline(
    steps=[
        ("preprocessor", tree_preprocessor),
        (
            "model",
            RandomForestRegressor(
                n_estimators=60,
                max_depth=16,
                min_samples_leaf=20,
                max_features=0.8,
                n_jobs=-1,
                random_state=42
            )
        )
    ]
)

print("=" * 65)
print("STEP 2N — MODEL 2: RANDOM FOREST REGRESSION")
print("=" * 65)

# ------------------------------------------------
# 2N.3 — Train model
# ------------------------------------------------

print("\nTraining Random Forest...")
print("This may take several minutes.")

rf_pipeline.fit(
    X_train_recovery,
    y_train
)

print("Random Forest training complete.")

# ------------------------------------------------
# 2N.4 — Predict validation outcomes
# ------------------------------------------------

rf_predictions_raw = rf_pipeline.predict(
    X_val_recovery
)

rf_predictions = np.maximum(
    rf_predictions_raw,
    0
)

# ------------------------------------------------
# 2N.5 — Calculate validation metrics
# ------------------------------------------------

rf_results = {
    "Model": "Random Forest (Recovery)",
    "MAE": round(
        mean_absolute_error(y_val, rf_predictions), 3
    ),
    "RMSE": round(
        np.sqrt(mean_squared_error(y_val, rf_predictions)), 3
    ),
    "R2": round(
        r2_score(y_val, rf_predictions), 4
    ),
    "MAE on 15+ min delays": round(
        mean_absolute_error(
            y_val[severe_mask],
            rf_predictions[severe_mask.to_numpy()]
        ), 3
    )
}

# ------------------------------------------------
# 2N.6 — Compare models
# ------------------------------------------------

print("\n2024 validation performance:")
display(pd.DataFrame([rf_results]))

model_comparison = pd.concat(
    [
        baseline_results_df,
        pd.DataFrame([schedule_results]),
        pd.DataFrame([ridge_results]),
        pd.DataFrame([rf_results])
    ],
    ignore_index=True
)

print("\nModel comparison:")
display(model_comparison)

# ------------------------------------------------
# 2N.7 — Prediction diagnostics
# ------------------------------------------------

print("\nPrediction diagnostics:")

print(
    "Mean predicted delay:",
    round(rf_predictions.mean(), 3)
)

print(
    "Maximum predicted delay:",
    round(rf_predictions.max(), 3)
)

print(
    "Negative raw predictions:",
    int((rf_predictions_raw < 0).sum())
)

print("\nSTEP 2N COMPLETE")
```

```python

# ================================================================
# BUSI6029 — ASSESSMENT 2
# STEP 2O: MODEL 3 — HISTOGRAM GRADIENT BOOSTING
# ================================================================
# Purpose:
# 1. Train a third predictive modelling approach.
# 2. Capture nonlinear delay patterns using boosting.
# 3. Compare against Ridge and Random Forest.
# 4. Preserve the untouched 2025 holdout.
#
# Training: 2023
# Validation: 2024
# ================================================================

from sklearn.ensemble import HistGradientBoostingRegressor
from sklearn.preprocessing import OrdinalEncoder

# ------------------------------------------------
# 2O.1 — Configure preprocessing
# ------------------------------------------------

boost_preprocessor = ColumnTransformer(
    transformers=[
        (
            "numeric",
            "passthrough",
            recovery_numeric_features
        ),
        (
            "categorical",
            OrdinalEncoder(
                handle_unknown="use_encoded_value",
                unknown_value=np.nan
            ),
            categorical_features
        )
    ],
    remainder="drop"
)

# ------------------------------------------------
# 2O.2 — Create boosting pipeline
# ------------------------------------------------

# Numerical features appear first, followed by the
# two encoded categorical airport features.

boost_categorical_mask = (
    [False] * len(recovery_numeric_features)
    + [True] * len(categorical_features)
)

boost_pipeline = Pipeline(
    steps=[
        ("preprocessor", boost_preprocessor),
        (
            "model",
            HistGradientBoostingRegressor(
                max_iter=150,
                max_leaf_nodes=31,
                min_samples_leaf=40,
                learning_rate=0.08,
                l2_regularization=1.0,
                categorical_features=boost_categorical_mask,
                early_stopping=False,
                random_state=42
            )
        )
    ]
)

# ------------------------------------------------
# 2O.3 — Train model
# ------------------------------------------------

print("=" * 65)
print("STEP 2O — MODEL 3: HISTOGRAM GRADIENT BOOSTING")
print("=" * 65)

print("\nTraining Histogram Gradient Boosting...")

boost_pipeline.fit(
    X_train_recovery,
    y_train
)

print("Boosting model training complete.")

# ------------------------------------------------
# 2O.4 — Predict 2024 validation outcomes
# ------------------------------------------------

boost_predictions_raw = boost_pipeline.predict(
    X_val_recovery
)

boost_predictions = np.maximum(
    boost_predictions_raw,
    0
)

# ------------------------------------------------
# 2O.5 — Calculate validation metrics
# ------------------------------------------------

boost_results = {
    "Model": "Gradient Boosting (Recovery)",
    "MAE": round(
        mean_absolute_error(y_val, boost_predictions), 3
    ),
    "RMSE": round(
        np.sqrt(mean_squared_error(y_val, boost_predictions)), 3
    ),
    "R2": round(
        r2_score(y_val, boost_predictions), 4
    ),
    "MAE on 15+ min delays": round(
        mean_absolute_error(
            y_val[severe_mask],
            boost_predictions[severe_mask.to_numpy()]
        ), 3
    )
}

# ------------------------------------------------
# 2O.6 — Compare all models
# ------------------------------------------------

model_comparison = pd.concat(
    [
        baseline_results_df,
        pd.DataFrame([schedule_results]),
        pd.DataFrame([ridge_results]),
        pd.DataFrame([rf_results]),
        pd.DataFrame([boost_results])
    ],
    ignore_index=True
)

print("\n2024 validation model comparison:")
display(model_comparison)

# ------------------------------------------------
# 2O.7 — Prediction diagnostics
# ------------------------------------------------

print("\nPrediction diagnostics:")

print(
    "Mean predicted delay:",
    round(boost_predictions.mean(), 3)
)

print(
    "Maximum predicted delay:",
    round(boost_predictions.max(), 3)
)

print(
    "Negative raw predictions:",
    int((boost_predictions_raw < 0).sum())
)

print("\nSTEP 2O COMPLETE")
```

```python

# ================================================================
# BUSI6029 — ASSESSMENT 2
# STEP 2P: MODEL PERFORMANCE BY DELAY SEVERITY
# ================================================================
# Purpose:
# 1. Compare prediction errors across delay categories.
# 2. Identify weaknesses during operational disruptions.
# 3. Examine systematic overprediction or underprediction.
# 4. Support operational model selection.
#
# Dataset: 2024 validation only.
# Categories are defined by ACTUAL arrival delay.
# This is retrospective evaluation, not real-time grouping.
# ================================================================

# ------------------------------------------------
# 2P.1 — Define actual delay categories
# ------------------------------------------------

delay_categories = {
    "No positive delay (0 min)": y_val == 0,
    "Minor delay (1–14 min)": (y_val > 0) & (y_val < 15),
    "Moderate delay (15–59 min)": (y_val >= 15) & (y_val < 60),
    "Major delay (60–119 min)": (y_val >= 60) & (y_val < 120),
    "Severe delay (120+ min)": y_val >= 120
}

# ------------------------------------------------
# 2P.2 — Collect model predictions
# ------------------------------------------------

prediction_sets = {
    "Ridge Regression": ridge_predictions,
    "Random Forest": rf_predictions,
    "Gradient Boosting": boost_predictions
}

# ------------------------------------------------
# 2P.3 — Calculate category-level metrics
# ------------------------------------------------

severity_results = []

for category, mask in delay_categories.items():

    actual = y_val[mask].to_numpy()
    category_count = len(actual)

    for model_name, predictions in prediction_sets.items():

        predicted = predictions[mask.to_numpy()]

        category_mae = mean_absolute_error(
            actual,
            predicted
        )

        category_bias = np.mean(
            predicted - actual
        )

        severity_results.append({
            "Delay Category": category,
            "Model": model_name,
            "Flights": category_count,
            "MAE": round(category_mae, 3),
            "Bias": round(category_bias, 3)
        })

severity_results_df = pd.DataFrame(
    severity_results
)

# ------------------------------------------------
# 2P.4 — Display MAE comparison
# ------------------------------------------------

print("=" * 65)
print("STEP 2P — OPERATIONAL DELAY-SEVERITY EVALUATION")
print("=" * 65)

print("\nMAE by actual delay category:")

severity_mae_table = severity_results_df.pivot(
    index="Delay Category",
    columns="Model",
    values="MAE"
)

display(severity_mae_table)

# ------------------------------------------------
# 2P.5 — Display prediction bias
# ------------------------------------------------

print("\nPrediction bias by actual delay category:")
print("Positive = overprediction; Negative = underprediction")

severity_bias_table = severity_results_df.pivot(
    index="Delay Category",
    columns="Model",
    values="Bias"
)

display(severity_bias_table)

# ------------------------------------------------
# 2P.6 — Show category sample sizes
# ------------------------------------------------

print("\nNumber of flights per delay category:")

category_counts = severity_results_df[
    ["Delay Category", "Flights"]
].drop_duplicates()

display(category_counts)

print("\nSTEP 2P COMPLETE")
```

```python

# ================================================================
# BUSI6029 — ASSESSMENT 2
# STEP 2Q: OPERATIONAL DELAY DETECTION
# ================================================================
# Purpose:
# 1. Convert regression predictions into delay alerts.
# 2. Evaluate 15+ and 60+ minute delay thresholds.
# 3. Measure precision, recall and F1-score.
# 4. Compare operational usefulness across models.
#
# Validation: 2024 only.
# No 2025 holdout data is accessed.
# ================================================================

from sklearn.metrics import (
    precision_score,
    recall_score,
    f1_score,
    confusion_matrix
)

# ------------------------------------------------
# 2Q.1 — Define operational delay thresholds
# ------------------------------------------------

delay_thresholds = [15, 60]

# Reuse the existing prediction_sets dictionary
# created in Step 2P.

# ------------------------------------------------
# 2Q.2 — Evaluate detection performance
# ------------------------------------------------

detection_results = []

for threshold in delay_thresholds:

    actual_flag = (
        y_val.to_numpy() >= threshold
    )

    for model_name, predictions in prediction_sets.items():

        predicted_flag = (
            predictions >= threshold
        )

        tn, fp, fn, tp = confusion_matrix(
            actual_flag,
            predicted_flag,
            labels=[False, True]
        ).ravel()

        detection_results.append({
            "Threshold": f"{threshold}+ minutes",
            "Model": model_name,
            "Precision": round(
                precision_score(
                    actual_flag,
                    predicted_flag,
                    zero_division=0
                ), 4
            ),
            "Recall": round(
                recall_score(
                    actual_flag,
                    predicted_flag,
                    zero_division=0
                ), 4
            ),
            "F1": round(
                f1_score(
                    actual_flag,
                    predicted_flag,
                    zero_division=0
                ), 4
            ),
            "True Positives": int(tp),
            "False Positives": int(fp),
            "False Negatives": int(fn),
            "True Negatives": int(tn)
        })

# ------------------------------------------------
# 2Q.3 — Display detection comparison
# ------------------------------------------------

detection_results_df = pd.DataFrame(
    detection_results
)

print("=" * 65)
print("STEP 2Q — OPERATIONAL DELAY DETECTION")
print("=" * 65)

print("\nPrecision, recall and F1:")
display(
    detection_results_df[
        ["Threshold", "Model", "Precision", "Recall", "F1"]
    ]
)

# ------------------------------------------------
# 2Q.4 — Display operational alert counts
# ------------------------------------------------

print("\nAlert classification counts:")
display(
    detection_results_df[
        [
            "Threshold",
            "Model",
            "True Positives",
            "False Positives",
            "False Negatives",
            "True Negatives"
        ]
    ]
)

# ------------------------------------------------
# 2Q.5 — Report actual disruption prevalence
# ------------------------------------------------

print("\nActual disruption prevalence:")

for threshold in delay_thresholds:
    actual_rate = (
        y_val.to_numpy() >= threshold
    ).mean() * 100

    print(
        f"{threshold}+ minute delays: "
        f"{actual_rate:.2f}%"
    )

print("\nSTEP 2Q COMPLETE")
```

```python

# ================================================================
# BUSI6029 — ASSESSMENT 2
# STEP 2R: OPERATIONAL RISK RANKING
# ================================================================
# Purpose:
# 1. Rank flights by predicted arrival delay.
# 2. Simulate limited operational review capacity.
# 3. Evaluate how many actual disruptions are captured.
# 4. Compare Ridge, Random Forest and Gradient Boosting.
#
# Dataset: 2024 validation only.
# This evaluates screening, NOT aircraft reassignment.
# ================================================================

# ------------------------------------------------
# 2R.1 — Define hypothetical review capacities
# ------------------------------------------------

review_capacities = [0.01, 0.05, 0.10, 0.20]

actual_delays = y_val.to_numpy()
n_flights = len(actual_delays)

# ------------------------------------------------
# 2R.2 — Rank predictions and calculate outcomes
# ------------------------------------------------

ranking_results = []

for model_name, predictions in prediction_sets.items():

    ranked_indices = np.argsort(-predictions)

    for capacity in review_capacities:

        n_reviewed = int(np.ceil(n_flights * capacity))

        selected_indices = ranked_indices[:n_reviewed]

        selected_actual = actual_delays[selected_indices]

        for threshold in [15, 60]:

            actual_disruptions = actual_delays >= threshold

            selected_disruptions = (
                selected_actual >= threshold
            )

            precision_at_capacity = (
                selected_disruptions.mean()
            )

            recall_at_capacity = (
                selected_disruptions.sum()
                / actual_disruptions.sum()
            )

            ranking_results.append({
                "Model": model_name,
                "Review Capacity": f"{capacity:.0%}",
                "Delay Threshold": f"{threshold}+ min",
                "Flights Reviewed": n_reviewed,
                "Precision": round(
                    precision_at_capacity, 4
                ),
                "Recall": round(
                    recall_at_capacity, 4
                ),
                "Disruptions Captured": int(
                    selected_disruptions.sum()
                )
            })

ranking_results_df = pd.DataFrame(ranking_results)

# ------------------------------------------------
# 2R.3 — Display 60+ minute disruption ranking
# ------------------------------------------------

print("=" * 65)
print("STEP 2R — OPERATIONAL RISK RANKING")
print("=" * 65)

print("\n60+ minute delay detection by review capacity:")

display(
    ranking_results_df[
        ranking_results_df["Delay Threshold"] == "60+ min"
    ][
        [
            "Model",
            "Review Capacity",
            "Flights Reviewed",
            "Precision",
            "Recall",
            "Disruptions Captured"
        ]
    ]
)

# ------------------------------------------------
# 2R.4 — Display 15+ minute disruption ranking
# ------------------------------------------------

print("\n15+ minute delay detection by review capacity:")

display(
    ranking_results_df[
        ranking_results_df["Delay Threshold"] == "15+ min"
    ][
        [
            "Model",
            "Review Capacity",
            "Precision",
            "Recall"
        ]
    ]
)

print("\nSTEP 2R COMPLETE")
```

```python

# ================================================================
# BUSI6029 — ASSESSMENT 2
# STEP 2S: RANDOM FOREST PERMUTATION IMPORTANCE
# ================================================================
# Purpose:
# 1. Assess feature importance on 2024 validation data.
# 2. Identify influential predictors of arrival delay.
# 3. Examine incoming-aircraft and turnaround features.
# 4. Support interpretation of the recovery-aware model.
#
# This is predictive importance, NOT causal importance.
# The 2025 holdout remains untouched.
# ================================================================

from sklearn.inspection import permutation_importance

# ------------------------------------------------
# 2S.1 — Select reproducible validation sample
# ------------------------------------------------

sample_size = min(20000, len(X_val_recovery))

sample_indices = np.random.default_rng(42).choice(
    len(X_val_recovery),
    size=sample_size,
    replace=False
)

X_importance = X_val_recovery.iloc[sample_indices]
y_importance = y_val.iloc[sample_indices]

# ------------------------------------------------
# 2S.2 — Calculate permutation importance
# ------------------------------------------------

print("=" * 65)
print("STEP 2S — RECOVERY-AWARE FEATURE IMPORTANCE")
print("=" * 65)

print(f"\nValidation sample size: {sample_size:,}")
print("Calculating permutation importance...")

importance_output = permutation_importance(
    rf_pipeline,
    X_importance,
    y_importance,
    scoring="neg_mean_squared_error",
    n_repeats=3,
    random_state=42,
    n_jobs=1
)

# ------------------------------------------------
# 2S.3 — Create feature importance table
# ------------------------------------------------

importance_df = pd.DataFrame({
    "Feature": X_importance.columns,
    "MSE Increase": importance_output.importances_mean,
    "Std Dev": importance_output.importances_std
}).sort_values(
    "MSE Increase",
    ascending=False
).reset_index(drop=True)

print("\nPermutation importance ranking:")
display(importance_df.round(3))

# ------------------------------------------------
# 2S.4 — Examine the recovery-specific predictors
# ------------------------------------------------

print("\nRecovery-specific feature importance:")

display(
    importance_df[
        importance_df["Feature"].isin(
            ["PREV_ARR_DELAY", "SCHED_GROUND_MIN"]
        )
    ].round(3)
)

print("\nSTEP 2S COMPLETE")
```

```python

# ================================================================
# BUSI6029 — ASSESSMENT 2
# STEP 2T: RECOVERY FEATURE ABLATION STUDY
# ================================================================
# Purpose:
# 1. Isolate the predictive contribution of each recovery feature.
# 2. Compare four feature configurations using Ridge.
# 3. Evaluate on the same 2024 validation flights.
# 4. Avoid causal interpretations of predictive improvements.
#
# Training: 2023
# Validation: 2024
# Holdout: 2025 untouched
# ================================================================

# ------------------------------------------------
# 2T.1 — Define feature configurations
# ------------------------------------------------

ablation_configs = {
    "Schedule only": schedule_features,
    "Schedule + previous delay": (
        schedule_features + ["PREV_ARR_DELAY"]
    ),
    "Schedule + ground time": (
        schedule_features + ["SCHED_GROUND_MIN"]
    ),
    "Full recovery-aware": candidate_features
}

ablation_results = []

print("=" * 65)
print("STEP 2T — RECOVERY FEATURE ABLATION STUDY")
print("=" * 65)

# ------------------------------------------------
# 2T.2 — Fit and evaluate each configuration
# ------------------------------------------------

for config_name, features in ablation_configs.items():

    numeric_cols = [
        col for col in features
        if col not in categorical_features
    ]

    preprocessor = ColumnTransformer(
        transformers=[
            (
                "numeric",
                StandardScaler(),
                numeric_cols
            ),
            (
                "categorical",
                OneHotEncoder(
                    handle_unknown="ignore",
                    sparse_output=True
                ),
                categorical_features
            )
        ],
        sparse_threshold=1.0
    )

    pipeline = Pipeline(
        steps=[
            ("preprocessor", preprocessor),
            (
                "model",
                Ridge(
                    alpha=10.0,
                    solver="lsqr",
                    max_iter=2000,
                    tol=1e-4
                )
            )
        ]
    )

    pipeline.fit(
        X_train_recovery[features],
        y_train
    )

    predictions = np.maximum(
        pipeline.predict(X_val_recovery[features]),
        0
    )

    ablation_results.append({
        "Configuration": config_name,
        "Features": len(features),
        "MAE": round(
            mean_absolute_error(y_val, predictions), 3
        ),
        "RMSE": round(
            np.sqrt(mean_squared_error(y_val, predictions)), 3
        ),
        "R2": round(
            r2_score(y_val, predictions), 4
        )
    })

# ------------------------------------------------
# 2T.3 — Display comparison
# ------------------------------------------------

ablation_results_df = pd.DataFrame(
    ablation_results
)

print("\nRidge ablation results:")
display(ablation_results_df)

# ------------------------------------------------
# 2T.4 — Calculate full-versus-schedule improvement
# ------------------------------------------------

schedule_rmse = ablation_results_df.loc[
    ablation_results_df["Configuration"] == "Schedule only",
    "RMSE"
].iloc[0]

full_rmse = ablation_results_df.loc[
    ablation_results_df["Configuration"] == "Full recovery-aware",
    "RMSE"
].iloc[0]

improvement_pct = (
    (schedule_rmse - full_rmse)
    / schedule_rmse * 100
)

print(
    "\nFull recovery RMSE improvement over schedule only:",
    f"{improvement_pct:.2f}%"
)

print("\nSTEP 2T COMPLETE")
```

```python

# ================================================================
# BUSI6029 — ASSESSMENT 2
# STEP 2U: INCOMING DELAY × GROUND TIME INTERACTION
# ================================================================
# Purpose:
# 1. Test whether a delay-ground-time interaction improves Ridge.
# 2. Compare with the existing recovery-aware Ridge.
# 3. Evaluate on 2024 validation data.
# 4. Keep the 2025 holdout untouched.
# ================================================================

# ------------------------------------------------
# 2U.1 — Create interaction feature
# ------------------------------------------------

interaction_feature = "DELAY_GROUND_INTERACTION"

X_train_interaction = X_train_recovery.copy()
X_val_interaction = X_val_recovery.copy()

for df in [X_train_interaction, X_val_interaction]:
    df[interaction_feature] = (
        df["PREV_ARR_DELAY"]
        * df["SCHED_GROUND_MIN"]
    )

interaction_numeric_features = (
    recovery_numeric_features + [interaction_feature]
)

# ------------------------------------------------
# 2U.2 — Build preprocessing and model pipeline
# ------------------------------------------------

interaction_preprocessor = ColumnTransformer(
    transformers=[
        (
            "numeric",
            StandardScaler(),
            interaction_numeric_features
        ),
        (
            "categorical",
            OneHotEncoder(
                handle_unknown="ignore",
                sparse_output=True
            ),
            categorical_features
        )
    ],
    sparse_threshold=1.0
)

interaction_pipeline = Pipeline(
    steps=[
        ("preprocessor", interaction_preprocessor),
        (
            "model",
            Ridge(
                alpha=10.0,
                solver="lsqr",
                max_iter=2000,
                tol=1e-4
            )
        )
    ]
)

# ------------------------------------------------
# 2U.3 — Train and predict
# ------------------------------------------------

print("=" * 65)
print("STEP 2U — DELAY × GROUND TIME INTERACTION")
print("=" * 65)

print("\nTraining interaction-enhanced Ridge...")

interaction_pipeline.fit(
    X_train_interaction,
    y_train
)

interaction_predictions = np.maximum(
    interaction_pipeline.predict(X_val_interaction),
    0
)

# ------------------------------------------------
# 2U.4 — Evaluate interaction model
# ------------------------------------------------

interaction_results = {
    "Model": "Ridge + Interaction",
    "MAE": round(
        mean_absolute_error(y_val, interaction_predictions), 3
    ),
    "RMSE": round(
        np.sqrt(
            mean_squared_error(y_val, interaction_predictions)
        ), 3
    ),
    "R2": round(
        r2_score(y_val, interaction_predictions), 4
    ),
    "MAE on 15+ min delays": round(
        mean_absolute_error(
            y_val[severe_mask],
            interaction_predictions[severe_mask.to_numpy()]
        ), 3
    )
}

# ------------------------------------------------
# 2U.5 — Compare with existing Ridge model
# ------------------------------------------------

interaction_comparison = pd.DataFrame([
    ridge_results,
    interaction_results
])

print("\nRidge interaction comparison:")
display(interaction_comparison)

print("\nSTEP 2U COMPLETE")
```

```python

# ================================================================
# BUSI6029 — ASSESSMENT 2
# STEP 2V: RANDOM FOREST HYPERPARAMETER SCREENING
# ================================================================
# Purpose:
# 1. Compare a small set of Random Forest configurations.
# 2. Use a reproducible training subsample for efficiency.
# 3. Evaluate each model on 2024 validation data.
# 4. Identify promising settings for full-data retraining.
#
# Training: 2023 sample
# Validation: 2024
# Holdout: 2025 untouched
# ================================================================

# ------------------------------------------------
# 2V.1 — Prepare reproducible training sample
# ------------------------------------------------

sample_n = min(150000, len(X_train_recovery))

rng = np.random.default_rng(42)

train_positions = rng.choice(
    len(X_train_recovery),
    size=sample_n,
    replace=False
)

X_train_tuning = X_train_recovery.iloc[train_positions]
y_train_tuning = y_train.iloc[train_positions]

# ------------------------------------------------
# 2V.2 — Reuse tree preprocessing
# ------------------------------------------------

# Fit preprocessing only on the training sample.
# This preserves the same feature treatment as Step 2N.

X_tuning_encoded = tree_preprocessor.fit_transform(
    X_train_tuning
)

X_val_tuning_encoded = tree_preprocessor.transform(
    X_val_recovery
)

# ------------------------------------------------
# 2V.3 — Define controlled configurations
# ------------------------------------------------

rf_configs = [
    {
        "name": "RF_A",
        "max_depth": 12,
        "min_samples_leaf": 20
    },
    {
        "name": "RF_B",
        "max_depth": 16,
        "min_samples_leaf": 20
    },
    {
        "name": "RF_C",
        "max_depth": 20,
        "min_samples_leaf": 20
    },
    {
        "name": "RF_D",
        "max_depth": 16,
        "min_samples_leaf": 10
    }
]

# ------------------------------------------------
# 2V.4 — Train and evaluate configurations
# ------------------------------------------------

tuning_results = []

print("=" * 65)
print("STEP 2V — RANDOM FOREST HYPERPARAMETER SCREENING")
print("=" * 65)

print(f"\nTraining sample: {sample_n:,} flights")
print(f"Validation flights: {len(y_val):,}")

for config in rf_configs:

    print(f"\nTraining {config['name']}...")

    model = RandomForestRegressor(
        n_estimators=60,
        max_depth=config["max_depth"],
        min_samples_leaf=config["min_samples_leaf"],
        max_features=0.8,
        n_jobs=-1,
        random_state=42
    )

    model.fit(
        X_tuning_encoded,
        y_train_tuning
    )

    predictions = np.maximum(
        model.predict(X_val_tuning_encoded),
        0
    )

    tuning_results.append({
        "Configuration": config["name"],
        "Max Depth": config["max_depth"],
        "Min Leaf": config["min_samples_leaf"],
        "MAE": round(
            mean_absolute_error(y_val, predictions), 3
        ),
        "RMSE": round(
            np.sqrt(mean_squared_error(y_val, predictions)), 3
        ),
        "R2": round(
            r2_score(y_val, predictions), 4
        ),
        "Severe MAE": round(
            mean_absolute_error(
                y_val[severe_mask],
                predictions[severe_mask.to_numpy()]
            ), 3
        )
    })

    del model, predictions

# ------------------------------------------------
# 2V.5 — Display results
# ------------------------------------------------

tuning_results_df = pd.DataFrame(
    tuning_results
).sort_values("RMSE")

print("\nRandom Forest tuning results:")
display(tuning_results_df)

print("\nOriginal full-data Random Forest benchmark:")
display(pd.DataFrame([rf_results]))

print("\nSTEP 2V COMPLETE")
```

```python

# ================================================================
# BUSI6029 — ASSESSMENT 2
# STEP 2W: FULL-DATA RANDOM FOREST CONFIRMATION
# ================================================================
# Purpose:
# 1. Retrain promising RF configurations on all 2023 data.
# 2. Compare against the original full-data RF benchmark.
# 3. Select a final configuration using 2024 validation.
# 4. Keep the 2025 holdout completely untouched.
# ================================================================

# ------------------------------------------------
# 2W.1 — Define shortlisted configurations
# ------------------------------------------------

full_rf_configs = {
    "RF_A_Full": {
        "max_depth": 12,
        "min_samples_leaf": 20
    },
    "RF_C_Full": {
        "max_depth": 20,
        "min_samples_leaf": 20
    }
}

full_rf_models = {}

full_rf_results = [
    {
        "Model": "Original RF",
        "MAE": rf_results["MAE"],
        "RMSE": rf_results["RMSE"],
        "R2": rf_results["R2"],
        "Severe MAE": rf_results["MAE on 15+ min delays"]
    }
]

# ------------------------------------------------
# 2W.2 — Fit each model on full 2023 training data
# ------------------------------------------------

print("=" * 65)
print("STEP 2W — FULL-DATA RANDOM FOREST CONFIRMATION")
print("=" * 65)

for model_name, params in full_rf_configs.items():

    print(f"\nTraining {model_name}...")

    model_pipeline = Pipeline([
        (
            "preprocessor",
            ColumnTransformer(
                transformers=[
                    (
                        "numeric",
                        "passthrough",
                        recovery_numeric_features
                    ),
                    (
                        "categorical",
                        OneHotEncoder(
                            handle_unknown="ignore",
                            sparse_output=True
                        ),
                        categorical_features
                    )
                ],
                sparse_threshold=1.0
            )
        ),
        (
            "model",
            RandomForestRegressor(
                n_estimators=60,
                max_depth=params["max_depth"],
                min_samples_leaf=params["min_samples_leaf"],
                max_features=0.8,
                n_jobs=-1,
                random_state=42
            )
        )
    ])

    model_pipeline.fit(
        X_train_recovery,
        y_train
    )

    predictions = np.maximum(
        model_pipeline.predict(X_val_recovery),
        0
    )

    full_rf_models[model_name] = model_pipeline

    full_rf_results.append({
        "Model": model_name,
        "MAE": round(
            mean_absolute_error(y_val, predictions), 3
        ),
        "RMSE": round(
            np.sqrt(mean_squared_error(y_val, predictions)), 3
        ),
        "R2": round(
            r2_score(y_val, predictions), 4
        ),
        "Severe MAE": round(
            mean_absolute_error(
                y_val[severe_mask],
                predictions[severe_mask.to_numpy()]
            ), 3
        )
    })

    del predictions

# ------------------------------------------------
# 2W.3 — Compare full-data models
# ------------------------------------------------

full_rf_results_df = pd.DataFrame(
    full_rf_results
).sort_values("RMSE").reset_index(drop=True)

print("\nFull-data Random Forest comparison:")
display(full_rf_results_df)

# ------------------------------------------------
# 2W.4 — Identify provisional RMSE winner
# ------------------------------------------------

best_rf_name = full_rf_results_df.loc[0, "Model"]

print(f"\nLowest validation RMSE: {best_rf_name}")

print("\nSTEP 2W COMPLETE")
```

```python

# ================================================================
# BUSI6029 — ASSESSMENT 2
# STEP 2X: FINAL 2025 HOLDOUT EVALUATION
# ================================================================
# Purpose:
# 1. Freeze the selected 2024-validation winner.
# 2. Evaluate its generalisation to unseen 2025 flights.
# 3. Measure regression accuracy and disruption detection.
# 4. Compare performance with 2024 validation.
#
# Training: 2023
# Validation and model selection: 2024
# Final holdout: 2025
#
# IMPORTANT:
# Do not retrain or modify model settings using 2025 data.
# ================================================================

# ------------------------------------------------
# 2X.1 — Freeze selected Random Forest model
# ------------------------------------------------

final_rf_model = full_rf_models["RF_A_Full"]

print("=" * 65)
print("STEP 2X — FINAL 2025 HOLDOUT EVALUATION")
print("=" * 65)

print("\nSelected model: RF_A_Full")
print(f"Holdout flights: {len(y_test):,}")

# ------------------------------------------------
# 2X.2 — Predict unseen 2025 arrival delays
# ------------------------------------------------

final_test_predictions_raw = final_rf_model.predict(
    X_test_recovery
)

final_test_predictions = np.maximum(
    final_test_predictions_raw,
    0
)

# Recalculate 2024 predictions for consistent comparison.
final_val_predictions = np.maximum(
    final_rf_model.predict(X_val_recovery),
    0
)

# ------------------------------------------------
# 2X.3 — Evaluate regression performance
# ------------------------------------------------

holdout_results = []

for period, actual, predictions in [
    ("2024 Validation", y_val, final_val_predictions),
    ("2025 Holdout", y_test, final_test_predictions)
]:

    actual_array = actual.to_numpy()
    severe_actual = actual_array >= 15

    holdout_results.append({
        "Period": period,
        "Flights": len(actual_array),
        "Actual Mean Delay": round(
            actual_array.mean(), 3
        ),
        "Predicted Mean Delay": round(
            predictions.mean(), 3
        ),
        "MAE": round(
            mean_absolute_error(actual_array, predictions), 3
        ),
        "RMSE": round(
            np.sqrt(
                mean_squared_error(actual_array, predictions)
            ), 3
        ),
        "R2": round(
            r2_score(actual_array, predictions), 4
        ),
        "Severe MAE": round(
            mean_absolute_error(
                actual_array[severe_actual],
                predictions[severe_actual]
            ), 3
        ),
        "Mean Prediction Bias": round(
            np.mean(predictions - actual_array), 3
        )
    })

holdout_results_df = pd.DataFrame(
    holdout_results
)

print("\nRegression performance comparison:")
display(holdout_results_df)

# ------------------------------------------------
# 2X.4 — Evaluate disruption detection
# ------------------------------------------------

holdout_detection = []

for period, actual, predictions in [
    ("2024 Validation", y_val, final_val_predictions),
    ("2025 Holdout", y_test, final_test_predictions)
]:

    actual_array = actual.to_numpy()

    for threshold in [15, 60]:

        actual_flag = actual_array >= threshold
        predicted_flag = predictions >= threshold

        holdout_detection.append({
            "Period": period,
            "Threshold": f"{threshold}+ min",
            "Actual Prevalence": round(
                actual_flag.mean(), 4
            ),
            "Precision": round(
                precision_score(
                    actual_flag,
                    predicted_flag,
                    zero_division=0
                ), 4
            ),
            "Recall": round(
                recall_score(
                    actual_flag,
                    predicted_flag,
                    zero_division=0
                ), 4
            ),
            "F1": round(
                f1_score(
                    actual_flag,
                    predicted_flag,
                    zero_division=0
                ), 4
            )
        })

holdout_detection_df = pd.DataFrame(
    holdout_detection
)

print("\nDisruption detection comparison:")
display(holdout_detection_df)

print("\nSTEP 2X COMPLETE")
```

```python

# ================================================================
# BUSI6029 — ASSESSMENT 2
# STEP 2Z: ENHANCED FEATURE FEASIBILITY AND REDUNDANCY
# ================================================================
# Purpose:
# 1. Inspect candidate features using 2023–2024 only.
# 2. Check completeness and data types.
# 3. Verify derived time and turnaround variables.
# 4. Identify features that need further investigation.
#
# No model retraining.
# The 2025 holdout is not accessed.
# ================================================================

# ------------------------------------------------
# 2Z.1 — Select required source columns
# ------------------------------------------------

enhancement_columns = [
    "YEAR",
    "DAY_OF_MONTH",
    "MONTH",
    "CRS_DEP_HOUR",
    "CRS_DEP_MINUTE",
    "CRS_DEP_MINUTES",
    "CRS_ARR_HOUR",
    "CRS_ARR_MINUTE",
    "CRS_ARR_MINUTES",
    "PREV_CRS_ARR_MINUTES",
    "PREV_ARR_DELAY",
    "SCHED_GROUND_MIN",
    "GROUND_TIME_BAND",
    "TURNAROUND_CLASS",
    "PREV_DELAY_BAND",
    "SEASON",
    "ROUTE",
    "PREV_CANCELLED",
    "PREV_DIVERTED"
]

# ------------------------------------------------
# 2Z.2 — Read limited development-period sample
# ------------------------------------------------

sample_parts = []
sample_limit = 50000
collected = 0

for chunk in pd.read_csv(
    recovery_file,
    usecols=enhancement_columns,
    chunksize=100000,
    low_memory=False
):
    chunk = chunk[chunk["YEAR"].isin([2023, 2024])]

    remaining = sample_limit - collected
    selected = chunk.head(remaining)

    if not selected.empty:
        sample_parts.append(selected.copy())
        collected += len(selected)

    if collected >= sample_limit:
        break

enhancement_sample = pd.concat(
    sample_parts,
    ignore_index=True
)

print("=" * 65)
print("STEP 2Z — FEATURE FEASIBILITY AND REDUNDANCY")
print("=" * 65)

print("\nSample shape:", enhancement_sample.shape)
print("Years represented:")
print(enhancement_sample["YEAR"].value_counts())

# ------------------------------------------------
# 2Z.3 — Examine completeness and variability
# ------------------------------------------------

enhancement_profile = pd.DataFrame({
    "Column": enhancement_sample.columns,
    "Data Type": [
        str(enhancement_sample[c].dtype)
        for c in enhancement_sample.columns
    ],
    "Missing %": [
        round(enhancement_sample[c].isna().mean() * 100, 2)
        for c in enhancement_sample.columns
    ],
    "Unique Values": [
        enhancement_sample[c].nunique(dropna=True)
        for c in enhancement_sample.columns
    ]
}).sort_values("Column")

print("\nCandidate feature profile:")
display(enhancement_profile)

# ------------------------------------------------
# 2Z.4 — Check scheduled-time redundancy
# ------------------------------------------------

dep_minutes_reconstructed = (
    enhancement_sample["CRS_DEP_HOUR"] * 60
    + enhancement_sample["CRS_DEP_MINUTE"]
)

arr_minutes_reconstructed = (
    enhancement_sample["CRS_ARR_HOUR"] * 60
    + enhancement_sample["CRS_ARR_MINUTE"]
)

dep_match = np.isclose(
    dep_minutes_reconstructed,
    enhancement_sample["CRS_DEP_MINUTES"],
    equal_nan=True
)

arr_match = np.isclose(
    arr_minutes_reconstructed,
    enhancement_sample["CRS_ARR_MINUTES"],
    equal_nan=True
)

print("\nScheduled-time consistency:")

print(
    "Departure minutes match reconstructed values:",
    f"{dep_match.mean():.2%}"
)

print(
    "Arrival minutes match reconstructed values:",
    f"{arr_match.mean():.2%}"
)

# ------------------------------------------------
# 2Z.5 — Check turnaround feature derivation
# ------------------------------------------------

ground_reconstructed = (
    enhancement_sample["CRS_DEP_MINUTES"]
    - enhancement_sample["PREV_CRS_ARR_MINUTES"]
)

ground_match = np.isclose(
    ground_reconstructed,
    enhancement_sample["SCHED_GROUND_MIN"],
    equal_nan=True
)

print("\nScheduled ground-time consistency:")
print(
    "Ground-time values match source time difference:",
    f"{ground_match.mean():.2%}"
)

# ------------------------------------------------
# 2Z.6 — Inspect proposed operational feature
# ------------------------------------------------

enhancement_sample["REMAINING_GROUND_MIN"] = (
    enhancement_sample["SCHED_GROUND_MIN"]
    - enhancement_sample["PREV_ARR_DELAY"]
)

print("\nRemaining ground-time distribution:")

display(
    enhancement_sample["REMAINING_GROUND_MIN"]
    .describe(percentiles=[0.05, 0.25, 0.50, 0.75, 0.95])
    .to_frame("Minutes")
)

print("\nSTEP 2Z COMPLETE")
```

```python

# ================================================================
# BUSI6029 — ASSESSMENT 2
# STEP 2AA: TURNAROUND PRESSURE AND EXTREME VALUES
# ================================================================
# Purpose:
# 1. Investigate negative remaining ground times.
# 2. Identify records with extreme incoming delays.
# 3. Test illustrative turnaround-buffer thresholds.
# 4. Assess feature-engineering suitability.
#
# Development-period sample: 2023–2024 only.
# No model training or 2025 holdout access.
# ================================================================

# ------------------------------------------------
# 2AA.1 — Calculate operational pressure measures
# ------------------------------------------------

turnaround_check = enhancement_sample.copy()

turnaround_check["REMAINING_GROUND_MIN"] = (
    turnaround_check["SCHED_GROUND_MIN"]
    - turnaround_check["PREV_ARR_DELAY"]
)

turnaround_check["NEGATIVE_GROUND"] = (
    turnaround_check["REMAINING_GROUND_MIN"] < 0
)

# ------------------------------------------------
# 2AA.2 — Summarise turnaround conditions
# ------------------------------------------------

conditions = {
    "Negative remaining ground": (
        turnaround_check["REMAINING_GROUND_MIN"] < 0
    ),
    "Remaining ground below 30 min": (
        turnaround_check["REMAINING_GROUND_MIN"] < 30
    ),
    "Remaining ground below 45 min": (
        turnaround_check["REMAINING_GROUND_MIN"] < 45
    ),
    "Remaining ground below 60 min": (
        turnaround_check["REMAINING_GROUND_MIN"] < 60
    ),
    "Incoming delay of 120+ min": (
        turnaround_check["PREV_ARR_DELAY"] >= 120
    ),
    "Incoming delay of 240+ min": (
        turnaround_check["PREV_ARR_DELAY"] >= 240
    )
}

condition_summary = pd.DataFrame([
    {
        "Condition": name,
        "Flights": int(mask.sum()),
        "Share %": round(mask.mean() * 100, 2)
    }
    for name, mask in conditions.items()
])

print("=" * 65)
print("STEP 2AA — TURNAROUND PRESSURE VALIDATION")
print("=" * 65)

print("\nTurnaround pressure summary:")
display(condition_summary)

# ------------------------------------------------
# 2AA.3 — Inspect the most extreme records
# ------------------------------------------------

print("\nTen most negative remaining-ground records:")

display(
    turnaround_check[
        [
            "YEAR",
            "PREV_ARR_DELAY",
            "SCHED_GROUND_MIN",
            "REMAINING_GROUND_MIN",
            "CRS_DEP_MINUTES",
            "PREV_CRS_ARR_MINUTES"
        ]
    ]
    .nsmallest(10, "REMAINING_GROUND_MIN")
)

# ------------------------------------------------
# 2AA.4 — Check buffer sensitivity
# ------------------------------------------------

buffer_results = []

for minimum_turnaround in [30, 45, 60]:

    shortfall = np.maximum(
        minimum_turnaround
        - turnaround_check["REMAINING_GROUND_MIN"],
        0
    )

    buffer_results.append({
        "Assumed Minimum Turnaround": minimum_turnaround,
        "Flights with Shortfall": int((shortfall > 0).sum()),
        "Shortfall Share %": round(
            (shortfall > 0).mean() * 100, 2
        ),
        "Mean Shortfall, Affected Flights": round(
            shortfall[shortfall > 0].mean(), 2
        ) if (shortfall > 0).any() else 0
    })

buffer_sensitivity_df = pd.DataFrame(buffer_results)

print("\nIllustrative turnaround-buffer sensitivity:")
display(buffer_sensitivity_df)

print("\nSTEP 2AA COMPLETE")
```

```python

# ================================================================
# BUSI6029 — ASSESSMENT 2
# STEP 2AB: TURNAROUND PRESSURE VS DOWNSTREAM DELAYS
# ================================================================
# Purpose:
# 1. Examine observed delays across remaining-ground bands.
# 2. Quantify 15+ and 60+ minute disruption rates.
# 3. Compare illustrative turnaround shortfall scenarios.
# 4. Assess whether engineered turnaround measures are useful.
#
# Development data only: 2023–2024.
# No model training or 2025 evaluation.
# ================================================================

# ------------------------------------------------
# 2AB.1 — Recreate aligned development sample
# ------------------------------------------------

analysis_columns = [
    "YEAR",
    "PREV_ARR_DELAY",
    "SCHED_GROUND_MIN",
    "ARR_DELAY_NEW"
]

analysis_parts = []
collected = 0

for chunk in pd.read_csv(
    recovery_file,
    usecols=analysis_columns,
    chunksize=100000,
    low_memory=False
):
    chunk = chunk[chunk["YEAR"].isin([2023, 2024])]

    remaining = 50000 - collected
    selected = chunk.head(remaining)

    if not selected.empty:
        analysis_parts.append(selected.copy())
        collected += len(selected)

    if collected >= 50000:
        break

pressure_analysis = pd.concat(
    analysis_parts,
    ignore_index=True
)

# Verify alignment with the previous sample.
assert np.allclose(
    pressure_analysis["PREV_ARR_DELAY"],
    enhancement_sample["PREV_ARR_DELAY"]
)

assert np.allclose(
    pressure_analysis["SCHED_GROUND_MIN"],
    enhancement_sample["SCHED_GROUND_MIN"]
)

# ------------------------------------------------
# 2AB.2 — Engineer remaining-ground measure
# ------------------------------------------------

pressure_analysis["REMAINING_GROUND_MIN"] = (
    pressure_analysis["SCHED_GROUND_MIN"]
    - pressure_analysis["PREV_ARR_DELAY"]
)

pressure_analysis["REMAINING_GROUND_BAND"] = pd.cut(
    pressure_analysis["REMAINING_GROUND_MIN"],
    bins=[
        -np.inf,
        0,
        30,
        60,
        90,
        np.inf
    ],
    labels=[
        "Below 0",
        "0 to <30",
        "30 to <60",
        "60 to <90",
        "90+"
    ],
    right=False
)

# ------------------------------------------------
# 2AB.3 — Calculate downstream delay outcomes
# ------------------------------------------------

pressure_analysis["DELAY_15_PLUS"] = (
    pressure_analysis["ARR_DELAY_NEW"] >= 15
)

pressure_analysis["DELAY_60_PLUS"] = (
    pressure_analysis["ARR_DELAY_NEW"] >= 60
)

ground_band_results = (
    pressure_analysis
    .groupby(
        "REMAINING_GROUND_BAND",
        observed=True
    )
    .agg(
        Flights=("ARR_DELAY_NEW", "size"),
        Mean_Delay=("ARR_DELAY_NEW", "mean"),
        Median_Delay=("ARR_DELAY_NEW", "median"),
        Delay_15_Rate=("DELAY_15_PLUS", "mean"),
        Delay_60_Rate=("DELAY_60_PLUS", "mean")
    )
    .reset_index()
)

ground_band_results["Delay_15_Rate"] *= 100
ground_band_results["Delay_60_Rate"] *= 100

print("=" * 65)
print("STEP 2AB — TURNAROUND PRESSURE AND DELAY")
print("=" * 65)

print("\nDelay outcomes by remaining-ground band:")
display(ground_band_results.round(2))

# ------------------------------------------------
# 2AB.4 — Compare assumed turnaround buffers
# ------------------------------------------------

buffer_comparison = []

for minimum_turnaround in [30, 45, 60]:

    shortfall_flag = (
        pressure_analysis["REMAINING_GROUND_MIN"]
        < minimum_turnaround
    )

    for group_name, mask in [
        ("With shortfall", shortfall_flag),
        ("Without shortfall", ~shortfall_flag)
    ]:

        subset = pressure_analysis.loc[mask]

        buffer_comparison.append({
            "Assumed Minimum": minimum_turnaround,
            "Group": group_name,
            "Flights": len(subset),
            "Mean Delay": round(
                subset["ARR_DELAY_NEW"].mean(), 2
            ),
            "15+ Delay Rate %": round(
                subset["DELAY_15_PLUS"].mean() * 100, 2
            ),
            "60+ Delay Rate %": round(
                subset["DELAY_60_PLUS"].mean() * 100, 2
            )
        })

buffer_comparison_df = pd.DataFrame(
    buffer_comparison
)

print("\nDelay outcomes by hypothetical buffer:")
display(buffer_comparison_df)

print("\nSTEP 2AB COMPLETE")
```

```python

# ================================================================
# BUSI6029 — ASSESSMENT 2
# STEP 3C: AIRCRAFT ROTATION AND FEASIBILITY AUDIT
# ================================================================
# Purpose:
# 1. Examine historical aircraft rotations.
# 2. Assess daily aircraft and flight coverage.
# 3. Check flight identifiers and location continuity.
# 4. Identify scheduling limitations before building an ILP.
#
# Source: Original 82-column recovery CSV.
# No optimisation decisions are made in this step.
# ================================================================

# ------------------------------------------------
# 3C.1 — Select operational fields
# ------------------------------------------------

rotation_columns = [
    "FL_DATE",
    "TAIL_NUM",
    "OP_CARRIER_FL_NUM",
    "ORIGIN",
    "DEST",
    "CRS_DEP_TIME",
    "CRS_ARR_TIME",
    "CRS_DEP_MINUTES",
    "CRS_ARR_MINUTES",
    "CRS_ELAPSED_TIME",
    "PREV_DATE",
    "PREV_DEST",
    "PREV_ARR_DELAY",
    "SCHED_GROUND_MIN",
    "SAME_DAY_SEQUENCE",
    "LOCATION_CONTINUOUS",
    "ARR_DELAY_NEW"
]

audit_dates = [
    "2024-01-15",
    "2024-01-16",
    "2024-01-17"
]

# ------------------------------------------------
# 3C.2 — Load only selected operating dates
# ------------------------------------------------

rotation_parts = []

for chunk in pd.read_csv(
    recovery_file,
    usecols=rotation_columns,
    chunksize=100000,
    low_memory=False
):
    selected = chunk.loc[
        chunk["FL_DATE"].isin(audit_dates)
    ]

    if not selected.empty:
        rotation_parts.append(selected.copy())

if not rotation_parts:
    raise ValueError(
        "No records found for the selected audit dates."
    )

rotation_audit = pd.concat(
    rotation_parts,
    ignore_index=True
)

rotation_audit["FL_DATE"] = pd.to_datetime(
    rotation_audit["FL_DATE"]
)

print("=" * 65)
print("STEP 3C — AIRCRAFT ROTATION FEASIBILITY AUDIT")
print("=" * 65)

print("\nExtracted records:", len(rotation_audit))

# ------------------------------------------------
# 3C.3 — Daily operational summary
# ------------------------------------------------

daily_rotation_summary = (
    rotation_audit
    .groupby("FL_DATE")
    .agg(
        Flights=("TAIL_NUM", "size"),
        Unique_Aircraft=("TAIL_NUM", "nunique"),
        Origin_Airports=("ORIGIN", "nunique"),
        Missing_Tails=(
            "TAIL_NUM",
            lambda s: int(s.isna().sum())
        )
    )
    .reset_index()
)

print("\nDaily operational summary:")
display(daily_rotation_summary)

# ------------------------------------------------
# 3C.4 — Candidate flight identifier check
# ------------------------------------------------

flight_key = [
    "FL_DATE",
    "OP_CARRIER_FL_NUM",
    "ORIGIN",
    "DEST",
    "CRS_DEP_TIME"
]

duplicate_rows = rotation_audit.duplicated(
    subset=flight_key,
    keep=False
)

print("\nFlight identification checks:")
print("Duplicate-key records:", int(duplicate_rows.sum()))
print(
    "Missing flight numbers:",
    int(rotation_audit["OP_CARRIER_FL_NUM"].isna().sum())
)
print(
    "Missing tail numbers:",
    int(rotation_audit["TAIL_NUM"].isna().sum())
)

# ------------------------------------------------
# 3C.5 — Inspect an example aircraft rotation
# ------------------------------------------------

middle_day = pd.Timestamp("2024-01-16")

middle_day_data = rotation_audit.loc[
    rotation_audit["FL_DATE"] == middle_day
]

tail_counts = middle_day_data["TAIL_NUM"].value_counts()

if not tail_counts.empty:

    example_tail = tail_counts.index[0]

    example_rotation = (
        middle_day_data.loc[
            middle_day_data["TAIL_NUM"] == example_tail
        ]
        .sort_values("CRS_DEP_MINUTES")
    )

    print("\nExample aircraft:", example_tail)
    print("Flights on January 16:", len(example_rotation))

    display(
        example_rotation[
            [
                "FL_DATE",
                "OP_CARRIER_FL_NUM",
                "ORIGIN",
                "DEST",
                "CRS_DEP_TIME",
                "CRS_ARR_TIME",
                "PREV_DEST",
                "PREV_ARR_DELAY",
                "SCHED_GROUND_MIN",
                "ARR_DELAY_NEW"
            ]
        ]
    )

# ------------------------------------------------
# 3C.6 — Existing continuity indicators
# ------------------------------------------------

print("\nSame-day sequence values:")
display(
    rotation_audit["SAME_DAY_SEQUENCE"]
    .value_counts(dropna=False)
)

print("\nLocation continuity values:")
display(
    rotation_audit["LOCATION_CONTINUOUS"]
    .value_counts(dropna=False)
)

print("\nSTEP 3C COMPLETE")
```

```python

# ================================================================
# BUSI6029 — ASSESSMENT 2
# STEP 3D: SAME-AIRPORT AIRCRAFT SWAP OPPORTUNITY AUDIT
# ================================================================
# Purpose:
# 1. Identify aircraft with departures from the same airport.
# 2. Group scheduled departures into time windows.
# 3. Identify potential pairs for further feasibility testing.
# 4. Preserve original aircraft identities and flight schedules.
#
# IMPORTANT:
# Same-airport, similar-time departures are NOT confirmed
# feasible swaps. Aircraft availability, compatibility and
# future rotation commitments remain unverified.
# ================================================================

# ------------------------------------------------
# 3D.1 — Select an airport and operating day
# ------------------------------------------------

audit_airport = "ATL"
audit_day = pd.Timestamp("2024-01-16")

airport_flights = rotation_audit.loc[
    (rotation_audit["FL_DATE"] == audit_day)
    & (rotation_audit["ORIGIN"] == audit_airport)
].copy()

print("=" * 65)
print("STEP 3D — POTENTIAL AIRCRAFT SWAP WINDOWS")
print("=" * 65)

print("\nAirport:", audit_airport)
print("Date:", audit_day.date())
print("Departing flight records:", len(airport_flights))
print("Distinct aircraft:", airport_flights["TAIL_NUM"].nunique())

# ------------------------------------------------
# 3D.2 — Create 30-minute departure windows
# ------------------------------------------------

airport_flights["WINDOW_START_MIN"] = (
    airport_flights["CRS_DEP_MINUTES"] // 30
) * 30

airport_flights["WINDOW_LABEL"] = (
    airport_flights["WINDOW_START_MIN"].floordiv(60)
    .astype(str).str.zfill(2)
    + ":"
    + airport_flights["WINDOW_START_MIN"].mod(60)
    .astype(str).str.zfill(2)
)

# ------------------------------------------------
# 3D.3 — Summarise departure-window activity
# ------------------------------------------------

window_summary = (
    airport_flights.groupby(
        ["WINDOW_START_MIN", "WINDOW_LABEL"]
    )
    .agg(
        Flights=("TAIL_NUM", "size"),
        Distinct_Aircraft=("TAIL_NUM", "nunique"),
        Average_Incoming_Delay=("PREV_ARR_DELAY", "mean"),
        Average_Scheduled_Ground=("SCHED_GROUND_MIN", "mean"),
        Average_Downstream_Delay=("ARR_DELAY_NEW", "mean")
    )
    .reset_index()
    .sort_values("WINDOW_START_MIN")
)

window_summary["Multiple_Aircraft"] = (
    window_summary["Distinct_Aircraft"] >= 2
)

print("\n30-minute departure-window summary:")

display(
    window_summary[
        [
            "WINDOW_LABEL",
            "Flights",
            "Distinct_Aircraft",
            "Average_Incoming_Delay",
            "Average_Scheduled_Ground",
            "Average_Downstream_Delay",
            "Multiple_Aircraft"
        ]
    ].round(2)
)

# ------------------------------------------------
# 3D.4 — Count multi-aircraft investigation windows
# ------------------------------------------------

multi_windows = window_summary.loc[
    window_summary["Multiple_Aircraft"]
]

print("\nWindows containing two or more aircraft:",
      len(multi_windows))

print(
    "Flight records within those windows:",
    int(
        airport_flights["WINDOW_START_MIN"].isin(
            multi_windows["WINDOW_START_MIN"]
        ).sum()
    )
)

# ------------------------------------------------
# 3D.5 — Inspect the busiest window
# ------------------------------------------------

if not window_summary.empty:

    busiest_window = window_summary.sort_values(
        ["Distinct_Aircraft", "Flights"],
        ascending=False
    ).iloc[0]

    example_window_flights = (
        airport_flights.loc[
            airport_flights["WINDOW_START_MIN"]
            == busiest_window["WINDOW_START_MIN"]
        ]
        .sort_values("CRS_DEP_MINUTES")
    )

    print("\nBusiest 30-minute window:",
          busiest_window["WINDOW_LABEL"])

    print("\nFlights in this window:")

    display(
        example_window_flights[
            [
                "TAIL_NUM",
                "OP_CARRIER_FL_NUM",
                "ORIGIN",
                "DEST",
                "CRS_DEP_TIME",
                "CRS_ARR_TIME",
                "PREV_ARR_DELAY",
                "SCHED_GROUND_MIN",
                "ARR_DELAY_NEW"
            ]
        ].head(20)
    )

print("\nSTEP 3D COMPLETE")
```

```python

# ================================================================
# BUSI6029 — ASSESSMENT 2
# STEP 3E: AIRCRAFT READINESS AND CANDIDATE FEASIBILITY
# ================================================================
# Purpose:
# 1. Estimate aircraft ready times at ATL.
# 2. Identify flights below an assumed turnaround buffer.
# 3. Find preliminary alternative aircraft candidates.
# 4. Assess whether further optimisation is justified.
#
# Assumed minimum turnaround: 45 minutes.
# Candidate matches are NOT confirmed feasible swaps.
# ================================================================

# ------------------------------------------------
# 3E.1 — Define operational assumptions
# ------------------------------------------------

minimum_turnaround = 45

atl_flights = airport_flights.copy().reset_index(drop=True)

# Previous arrival is expressed in ATL local clock minutes.
atl_flights["EST_PREV_ARRIVAL_MIN"] = (
    atl_flights["CRS_DEP_MINUTES"]
    - atl_flights["SCHED_GROUND_MIN"]
    + atl_flights["PREV_ARR_DELAY"]
)

atl_flights["EST_READY_MIN"] = (
    atl_flights["EST_PREV_ARRIVAL_MIN"]
    + minimum_turnaround
)

atl_flights["REMAINING_GROUND_MIN"] = (
    atl_flights["CRS_DEP_MINUTES"]
    - atl_flights["EST_PREV_ARRIVAL_MIN"]
)

atl_flights["TURNAROUND_SHORTFALL"] = (
    minimum_turnaround
    - atl_flights["REMAINING_GROUND_MIN"]
).clip(lower=0)

atl_flights["CONSTRAINED"] = (
    atl_flights["TURNAROUND_SHORTFALL"] > 0
)

# ------------------------------------------------
# 3E.2 — Summarise aircraft readiness
# ------------------------------------------------

print("=" * 65)
print("STEP 3E — AIRCRAFT READINESS AUDIT")
print("=" * 65)

print("\nAssumed minimum turnaround:", minimum_turnaround)
print("ATL departures:", len(atl_flights))

print(
    "Flights below turnaround threshold:",
    int(atl_flights["CONSTRAINED"].sum())
)

print(
    "Share below threshold:",
    f"{atl_flights['CONSTRAINED'].mean():.2%}"
)

print("\nTurnaround shortfall distribution:")

display(
    atl_flights.loc[
        atl_flights["CONSTRAINED"],
        "TURNAROUND_SHORTFALL"
    ].describe().to_frame("Minutes")
)

# ------------------------------------------------
# 3E.3 — Screen alternative aircraft readiness
# ------------------------------------------------
# A different aircraft is a preliminary timing candidate
# if its estimated ready time is no later than the
# target flight's scheduled departure.
#
# This does NOT check aircraft compatibility or whether
# transferring the aircraft would disrupt its own flight.

candidate_rows = []

constrained_flights = atl_flights.loc[
    atl_flights["CONSTRAINED"]
]

for target_index, target_flight in constrained_flights.iterrows():

    candidates = atl_flights.loc[
        (atl_flights["TAIL_NUM"] != target_flight["TAIL_NUM"])
        &
        (
            atl_flights["EST_READY_MIN"]
            <= target_flight["CRS_DEP_MINUTES"]
        )
    ]

    candidate_rows.append({
        "Target Index": target_index,
        "Target Tail": target_flight["TAIL_NUM"],
        "Flight Number": target_flight["OP_CARRIER_FL_NUM"],
        "Destination": target_flight["DEST"],
        "Departure Time": target_flight["CRS_DEP_TIME"],
        "Incoming Delay": target_flight["PREV_ARR_DELAY"],
        "Turnaround Shortfall": target_flight["TURNAROUND_SHORTFALL"],
        "Preliminary Alternative Tails": candidates[
            "TAIL_NUM"
        ].nunique()
    })

candidate_summary_df = pd.DataFrame(candidate_rows)

# ------------------------------------------------
# 3E.4 — Report preliminary candidate coverage
# ------------------------------------------------

print("\nPreliminary candidate screening:")

if not candidate_summary_df.empty:

    display(
        candidate_summary_df.sort_values(
            "Turnaround Shortfall",
            ascending=False
        ).head(20)
    )

    print(
        "\nConstrained flights with at least one "
        "timing-compatible alternative:",
        int(
            (
                candidate_summary_df[
                    "Preliminary Alternative Tails"
                ] > 0
            ).sum()
        )
    )

    print(
        "Constrained flights with no such alternative:",
        int(
            (
                candidate_summary_df[
                    "Preliminary Alternative Tails"
                ] == 0
            ).sum()
        )
    )

else:
    print("No flights breach the assumed turnaround buffer.")

# ------------------------------------------------
# 3E.5 — Validate timing calculations
# ------------------------------------------------

reconstructed_ground = (
    atl_flights["CRS_DEP_MINUTES"]
    - atl_flights["EST_PREV_ARRIVAL_MIN"]
)

calculation_match = np.allclose(
    reconstructed_ground,
    atl_flights["REMAINING_GROUND_MIN"]
)

print("\nTiming arithmetic validated:", calculation_match)

print("\nSTEP 3E COMPLETE")

```

```python

# ================================================================
# BUSI6029 — ASSESSMENT 2
# STEP 3F: RECIPROCAL AIRCRAFT SWAP SCREENING
# ================================================================
# Purpose:
# 1. Examine pairs of different aircraft departing ATL.
# 2. Require departures within a 30-minute window.
# 3. Estimate turnaround shortfall before and after swapping.
# 4. Identify candidate exchanges for deeper feasibility checks.
#
# A candidate is NOT a confirmed operationally feasible swap.
# No optimisation is performed here.
# ================================================================

from itertools import combinations

# ------------------------------------------------
# 3F.1 — Define assumptions
# ------------------------------------------------

swap_window_min = 30
swap_min_turnaround = 45

swap_flights = atl_flights.copy().reset_index(drop=True)

swap_flights["BASE_SHORTFALL"] = np.maximum(
    swap_min_turnaround
    - (
        swap_flights["CRS_DEP_MINUTES"]
        - swap_flights["EST_PREV_ARRIVAL_MIN"]
    ),
    0
)

# ------------------------------------------------
# 3F.2 — Examine same-airport reciprocal exchanges
# ------------------------------------------------

swap_records = []
pairs_checked = 0

for i, j in combinations(range(len(swap_flights)), 2):

    flight_a = swap_flights.iloc[i]
    flight_b = swap_flights.iloc[j]

    # Different aircraft only
    if flight_a["TAIL_NUM"] == flight_b["TAIL_NUM"]:
        continue

    departure_a = flight_a["CRS_DEP_MINUTES"]
    departure_b = flight_b["CRS_DEP_MINUTES"]

    # Departures must be close in scheduled time
    if abs(departure_a - departure_b) > swap_window_min:
        continue

    pairs_checked += 1

    arrival_a = flight_a["EST_PREV_ARRIVAL_MIN"]
    arrival_b = flight_b["EST_PREV_ARRIVAL_MIN"]

    # Original combined turnaround shortfall
    before = (
        flight_a["BASE_SHORTFALL"]
        + flight_b["BASE_SHORTFALL"]
    )

    # Exchange incoming aircraft assignments
    after_a = max(
        swap_min_turnaround - (departure_a - arrival_b),
        0
    )

    after_b = max(
        swap_min_turnaround - (departure_b - arrival_a),
        0
    )

    after = after_a + after_b
    improvement = before - after

    if before > 0 and improvement > 0:

        swap_records.append({
            "Flight A": flight_a["OP_CARRIER_FL_NUM"],
            "Tail A": flight_a["TAIL_NUM"],
            "Destination A": flight_a["DEST"],
            "Dep A": flight_a["CRS_DEP_TIME"],
            "Flight B": flight_b["OP_CARRIER_FL_NUM"],
            "Tail B": flight_b["TAIL_NUM"],
            "Destination B": flight_b["DEST"],
            "Dep B": flight_b["CRS_DEP_TIME"],
            "Original Shortfall": before,
            "Swapped Shortfall": after,
            "Shortfall Reduction": improvement,
            "Both Meet 45-Min Buffer": after == 0
        })

# ------------------------------------------------
# 3F.3 — Summarise screening results
# ------------------------------------------------

swap_screen_df = pd.DataFrame(swap_records)

print("=" * 65)
print("STEP 3F — RECIPROCAL AIRCRAFT SWAP SCREENING")
print("=" * 65)

print("\nDepartures examined:", len(swap_flights))
print("Pairs within 30 minutes:", pairs_checked)
print("Pairs reducing combined shortfall:", len(swap_screen_df))

if not swap_screen_df.empty:

    print(
        "Pairs eliminating both turnaround shortfalls:",
        int(swap_screen_df["Both Meet 45-Min Buffer"].sum())
    )

    print("\nTop candidate exchanges:")

    display(
        swap_screen_df.sort_values(
            "Shortfall Reduction",
            ascending=False
        ).head(15)
    )

    print("\nSwap improvement distribution:")

    display(
        swap_screen_df[
            [
                "Original Shortfall",
                "Swapped Shortfall",
                "Shortfall Reduction"
            ]
        ].describe().round(2)
    )

else:
    print(
        "\nNo improving reciprocal swaps found "
        "under these assumptions."
    )

print("\nSTEP 3F COMPLETE")
```

```python

# ================================================================
# BUSI6029 — ASSESSMENT 2
# STEP 3G: DESTINATION-COMPATIBLE SWAP AUDIT
# ================================================================
# Purpose:
# 1. Filter improving swaps to the same destination.
# 2. Assess the effect of a conservative location rule.
# 3. Inspect other scheduled commitments for candidate tails.
# 4. Determine whether a restricted swap ILP is viable.
#
# Candidate swaps are NOT yet confirmed feasible.
# ================================================================

# ------------------------------------------------
# 3G.1 — Filter to matching destinations
# ------------------------------------------------

same_dest_swaps = swap_screen_df.loc[
    swap_screen_df["Destination A"]
    == swap_screen_df["Destination B"]
].copy()

print("=" * 65)
print("STEP 3G — DESTINATION-COMPATIBLE SWAP AUDIT")
print("=" * 65)

print("\nAll improving pairs:", len(swap_screen_df))
print("Same-destination improving pairs:", len(same_dest_swaps))

if len(swap_screen_df) > 0:
    print(
        "Share retaining destination compatibility:",
        f"{len(same_dest_swaps) / len(swap_screen_df):.2%}"
    )

# ------------------------------------------------
# 3G.2 — Compare screening categories
# ------------------------------------------------

comparison_rows = []

for name, df in [
    ("All improving pairs", swap_screen_df),
    ("Same-destination improving pairs", same_dest_swaps)
]:
    comparison_rows.append({
        "Category": name,
        "Candidate Pairs": len(df),
        "Both Shortfalls Eliminated": (
            int(df["Both Meet 45-Min Buffer"].sum())
            if len(df) else 0
        ),
        "Mean Shortfall Reduction": (
            round(df["Shortfall Reduction"].mean(), 2)
            if len(df) else np.nan
        ),
        "Maximum Shortfall Reduction": (
            round(df["Shortfall Reduction"].max(), 2)
            if len(df) else np.nan
        )
    })

print("\nEffect of destination compatibility:")

display(pd.DataFrame(comparison_rows))

# ------------------------------------------------
# 3G.3 — Count recorded same-day tail commitments
# ------------------------------------------------
# This counts ALL legs recorded for each tail on the day.
# It does not yet establish chronological next-flight
# feasibility or aircraft availability.

day_rotations = rotation_audit.loc[
    rotation_audit["FL_DATE"]
    == pd.Timestamp("2024-01-16")
]

tail_daily_commitments = (
    day_rotations.groupby("TAIL_NUM").size()
)

if not same_dest_swaps.empty:

    same_dest_swaps["Tail A Daily Legs"] = (
        same_dest_swaps["Tail A"]
        .map(tail_daily_commitments)
        .fillna(0)
        .astype(int)
    )

    same_dest_swaps["Tail B Daily Legs"] = (
        same_dest_swaps["Tail B"]
        .map(tail_daily_commitments)
        .fillna(0)
        .astype(int)
    )

    same_dest_swaps["Both Tails Have Multiple Legs"] = (
        (same_dest_swaps["Tail A Daily Legs"] > 1)
        & (same_dest_swaps["Tail B Daily Legs"] > 1)
    )

    print("\nSame-destination pairs involving two")
    print("aircraft with multiple recorded daily legs:")

    print(
        int(
            same_dest_swaps[
                "Both Tails Have Multiple Legs"
            ].sum()
        )
    )

    # ------------------------------------------------
    # 3G.4 — Display strongest restricted candidates
    # ------------------------------------------------

    print("\nTop same-destination swap candidates:")

    display(
        same_dest_swaps.sort_values(
            "Shortfall Reduction",
            ascending=False
        )[
            [
                "Flight A",
                "Tail A",
                "Dep A",
                "Flight B",
                "Tail B",
                "Dep B",
                "Destination A",
                "Original Shortfall",
                "Swapped Shortfall",
                "Shortfall Reduction",
                "Both Meet 45-Min Buffer",
                "Tail A Daily Legs",
                "Tail B Daily Legs"
            ]
        ].head(15)
    )

else:
    print(
        "\nNo improving same-destination swaps found."
    )

print("\nSTEP 3G COMPLETE")
```

```python

# ================================================================
# BUSI6029 — ASSESSMENT 2
# STEP 3H: ATL RETURN-TO-HUB ROTATION BLOCK AUDIT
# ================================================================
# Purpose:
# 1. Inspect consecutive flight records for each aircraft.
# 2. Identify ATL -> destination -> ATL rotation blocks.
# 3. Assess whether block-based reassignment is worth exploring.
# 4. Flag chronological and operational limitations.
#
# WARNING:
# Original CSV row order is not independently validated
# as UTC chronological order. These are candidate blocks,
# not proven time-feasible aircraft rotations.
# ================================================================

# ------------------------------------------------
# 3H.1 — Select January 16 aircraft rotations
# ------------------------------------------------

rotation_day = (
    rotation_audit.loc[
        rotation_audit["FL_DATE"]
        == pd.Timestamp("2024-01-16")
    ]
    .copy()
)

# Preserve original extracted row order for screening.
rotation_day["SOURCE_ORDER"] = np.arange(
    len(rotation_day)
)

rotation_day = rotation_day.sort_values(
    ["TAIL_NUM", "SOURCE_ORDER"]
)

# ------------------------------------------------
# 3H.2 — Identify consecutive same-tail records
# ------------------------------------------------

tail_groups = rotation_day.groupby(
    "TAIL_NUM",
    sort=False
)

rotation_day["NEXT_ORIGIN"] = (
    tail_groups["ORIGIN"].shift(-1)
)

rotation_day["NEXT_DEST"] = (
    tail_groups["DEST"].shift(-1)
)

rotation_day["NEXT_FLIGHT"] = (
    tail_groups["OP_CARRIER_FL_NUM"].shift(-1)
)

rotation_day["NEXT_DEP_TIME"] = (
    tail_groups["CRS_DEP_TIME"].shift(-1)
)

rotation_day["NEXT_ARR_TIME"] = (
    tail_groups["CRS_ARR_TIME"].shift(-1)
)

# ------------------------------------------------
# 3H.3 — Identify ATL outbound-and-return blocks
# ------------------------------------------------

return_block_mask = (
    (rotation_day["ORIGIN"] == "ATL")
    & (rotation_day["DEST"] != "ATL")
    & (rotation_day["NEXT_ORIGIN"]
       == rotation_day["DEST"])
    & (rotation_day["NEXT_DEST"] == "ATL")
)

atl_return_blocks = (
    rotation_day.loc[return_block_mask]
    .copy()
)

atl_return_blocks["BLOCK_ROUTE"] = (
    "ATL → "
    + atl_return_blocks["DEST"]
    + " → ATL"
)

# ------------------------------------------------
# 3H.4 — Summarise rotation block coverage
# ------------------------------------------------

print("=" * 65)
print("STEP 3H — ATL RETURN-TO-HUB BLOCK AUDIT")
print("=" * 65)

print("\nRecorded daily flights:", len(rotation_day))

print(
    "Unique aircraft:",
    rotation_day["TAIL_NUM"].nunique()
)

print(
    "ATL outbound-and-return candidate blocks:",
    len(atl_return_blocks)
)

print(
    "Aircraft with candidate return blocks:",
    atl_return_blocks["TAIL_NUM"].nunique()
)

print("\nMost common return-block routes:")

display(
    atl_return_blocks["BLOCK_ROUTE"]
    .value_counts()
    .head(15)
    .rename_axis("Route")
    .reset_index(name="Blocks")
)

# ------------------------------------------------
# 3H.5 — Inspect candidate blocks
# ------------------------------------------------

print("\nExample ATL return-to-hub blocks:")

display(
    atl_return_blocks[
        [
            "TAIL_NUM",
            "BLOCK_ROUTE",
            "OP_CARRIER_FL_NUM",
            "CRS_DEP_TIME",
            "CRS_ARR_TIME",
            "NEXT_FLIGHT",
            "NEXT_DEP_TIME",
            "NEXT_ARR_TIME",
            "PREV_ARR_DELAY",
            "SCHED_GROUND_MIN"
        ]
    ].head(20)
)

# ------------------------------------------------
# 3H.6 — Examine outbound departure coverage
# ------------------------------------------------

outbound_total = (
    (rotation_day["ORIGIN"] == "ATL")
    & (rotation_day["DEST"] != "ATL")
).sum()

print("\nATL outbound records:", int(outbound_total))

print(
    "Share beginning a candidate return block:",
    f"{len(atl_return_blocks) / outbound_total:.2%}"
    if outbound_total else "N/A"
)

print("\nSTEP 3H COMPLETE")
```

```python

# ================================================================
# BUSI6029 — ASSESSMENT 2
# STEP 3I: RETURN-BLOCK TIMING AND COMMITMENT VALIDATION
# ================================================================
# Purpose:
# 1. Validate same-day ATL round-trip timing.
# 2. Identify the next recorded aircraft commitment.
# 3. Exclude ambiguous or chronologically inconsistent blocks.
# 4. Prepare a conservative subset for swap-feasibility checks.
#
# No aircraft swapping or optimisation yet.
# ================================================================

# ------------------------------------------------
# 3I.1 — Recover return-leg and following-leg details
# ------------------------------------------------

ordered_rotations = rotation_day.sort_values(
    ["TAIL_NUM", "SOURCE_ORDER"]
).copy()

tail_groups = ordered_rotations.groupby(
    "TAIL_NUM", sort=False
)

ordered_rotations["RETURN_ARR_MIN"] = (
    tail_groups["CRS_ARR_MINUTES"].shift(-1)
)

ordered_rotations["RETURN_ELAPSED_MIN"] = (
    tail_groups["CRS_ELAPSED_TIME"].shift(-1)
)

ordered_rotations["FOLLOWING_ORIGIN"] = (
    tail_groups["ORIGIN"].shift(-2)
)

ordered_rotations["FOLLOWING_DEP_MIN"] = (
    tail_groups["CRS_DEP_MINUTES"].shift(-2)
)

# ------------------------------------------------
# 3I.2 — Retain identified ATL return blocks
# ------------------------------------------------

block_mask = (
    (ordered_rotations["ORIGIN"] == "ATL")
    & (ordered_rotations["DEST"] != "ATL")
    & (tail_groups["ORIGIN"].shift(-1)
       == ordered_rotations["DEST"])
    & (tail_groups["DEST"].shift(-1) == "ATL")
)

block_timing_audit = (
    ordered_rotations.loc[block_mask].copy()
)

block_timing_audit["BLOCK_START_MIN"] = (
    block_timing_audit["CRS_DEP_MINUTES"]
)

block_timing_audit["BLOCK_END_MIN"] = (
    block_timing_audit["RETURN_ARR_MIN"]
)

block_timing_audit["BLOCK_DURATION_MIN"] = (
    block_timing_audit["BLOCK_END_MIN"]
    - block_timing_audit["BLOCK_START_MIN"]
)

# ------------------------------------------------
# 3I.3 — Check same-day chronological consistency
# ------------------------------------------------

block_timing_audit["POSITIVE_DURATION"] = (
    block_timing_audit["BLOCK_DURATION_MIN"] > 0
)

block_timing_audit["FLIGHT_TIME_PLAUSIBLE"] = (
    block_timing_audit["BLOCK_DURATION_MIN"]
    >= (
        block_timing_audit["CRS_ELAPSED_TIME"]
        + block_timing_audit["RETURN_ELAPSED_MIN"]
    )
)

block_timing_audit["NEXT_COMMITMENT_RECORDED"] = (
    block_timing_audit["FOLLOWING_ORIGIN"].notna()
)

block_timing_audit["NEXT_FROM_ATL"] = (
    block_timing_audit["FOLLOWING_ORIGIN"] == "ATL"
)

block_timing_audit["NEXT_AFTER_RETURN"] = (
    block_timing_audit["FOLLOWING_DEP_MIN"]
    >= block_timing_audit["BLOCK_END_MIN"]
)

block_timing_audit["VALID_SAME_DAY_BLOCK"] = (
    block_timing_audit["POSITIVE_DURATION"]
    & block_timing_audit["FLIGHT_TIME_PLAUSIBLE"]
)

# ------------------------------------------------
# 3I.4 — Summarise chronological screening
# ------------------------------------------------

print("=" * 65)
print("STEP 3I — RETURN-BLOCK TIMING VALIDATION")
print("=" * 65)

checks = [
    ("Candidate return blocks", len(block_timing_audit)),
    ("Positive same-day duration",
     block_timing_audit["POSITIVE_DURATION"].sum()),
    ("Duration accommodates scheduled flight times",
     block_timing_audit["FLIGHT_TIME_PLAUSIBLE"].sum()),
    ("Pass both same-day checks",
     block_timing_audit["VALID_SAME_DAY_BLOCK"].sum()),
    ("Have a following recorded flight",
     block_timing_audit["NEXT_COMMITMENT_RECORDED"].sum()),
    ("Following flight departs ATL",
     block_timing_audit["NEXT_FROM_ATL"].sum()),
    ("Following flight departs after return",
     (
         block_timing_audit["NEXT_COMMITMENT_RECORDED"]
         & block_timing_audit["NEXT_AFTER_RETURN"]
     ).sum())
]

block_screen_summary = pd.DataFrame(
    checks,
    columns=["Check", "Blocks"]
)

print("\nBlock timing and commitment summary:")
display(block_screen_summary)

# ------------------------------------------------
# 3I.5 — Create conservative screening subset
# ------------------------------------------------

valid_return_blocks = block_timing_audit.loc[
    block_timing_audit["VALID_SAME_DAY_BLOCK"]
    & (
        ~block_timing_audit["NEXT_COMMITMENT_RECORDED"]
        |
        (
            block_timing_audit["NEXT_FROM_ATL"]
            & block_timing_audit["NEXT_AFTER_RETURN"]
        )
    )
].copy()

# ------------------------------------------------
# 3I.6 — Inspect validated candidates
# ------------------------------------------------

print("\nConservative candidate blocks:",
      len(valid_return_blocks))

print("\nExample candidate blocks:")

display(
    valid_return_blocks[
        [
            "TAIL_NUM",
            "DEST",
            "BLOCK_START_MIN",
            "BLOCK_END_MIN",
            "BLOCK_DURATION_MIN",
            "FOLLOWING_ORIGIN",
            "FOLLOWING_DEP_MIN",
            "PREV_ARR_DELAY",
            "SCHED_GROUND_MIN"
        ]
    ].head(15)
)

print("\nSTEP 3I COMPLETE")
```

```python

# ================================================================
# BUSI6029 — ASSESSMENT 2
# STEP 3J: ROUND-TRIP BLOCK SWAP FEASIBILITY SCREENING
# ================================================================
# Purpose:
# 1. Exchange complete ATL-return flight blocks.
# 2. Check initial aircraft readiness.
# 3. Protect recorded next departures at ATL.
# 4. Identify improving schedule-based swap candidates.
#
# Assumptions: 45-minute ATL turnaround, 120-minute
# departure screening window, scheduled block return times.
#
# NOT a confirmed operational reassignment or optimisation.
# ================================================================

# ------------------------------------------------
# 3J.1 — Restrict to blocks with recorded commitments
# ------------------------------------------------

min_turnaround = 45
block_swap_window = 120

eligible_blocks = valid_return_blocks.loc[
    valid_return_blocks["NEXT_COMMITMENT_RECORDED"]
    & valid_return_blocks["NEXT_FROM_ATL"]
    & valid_return_blocks["NEXT_AFTER_RETURN"]
].copy().reset_index(drop=True)

# Estimated arrival of each original aircraft at ATL,
# before its round-trip block.
eligible_blocks["INCOMING_ARRIVAL_MIN"] = (
    eligible_blocks["BLOCK_START_MIN"]
    - eligible_blocks["SCHED_GROUND_MIN"]
    + eligible_blocks["PREV_ARR_DELAY"]
)

eligible_blocks["ORIGINAL_START_SHORTFALL"] = np.maximum(
    eligible_blocks["INCOMING_ARRIVAL_MIN"]
    + min_turnaround
    - eligible_blocks["BLOCK_START_MIN"],
    0
)

# ------------------------------------------------
# 3J.2 — Examine reciprocal block exchanges
# ------------------------------------------------

block_swap_records = []
screened_pairs = 0
start_feasible_pairs = 0
fully_screened_pairs = 0

for i, j in combinations(range(len(eligible_blocks)), 2):

    a = eligible_blocks.iloc[i]
    b = eligible_blocks.iloc[j]

    if a["TAIL_NUM"] == b["TAIL_NUM"]:
        continue

    if abs(
        a["BLOCK_START_MIN"] - b["BLOCK_START_MIN"]
    ) > block_swap_window:
        continue

    screened_pairs += 1

    # Tail A would operate block B.
    a_ready_for_b = (
        a["INCOMING_ARRIVAL_MIN"] + min_turnaround
        <= b["BLOCK_START_MIN"]
    )

    # Tail B would operate block A.
    b_ready_for_a = (
        b["INCOMING_ARRIVAL_MIN"] + min_turnaround
        <= a["BLOCK_START_MIN"]
    )

    if not (a_ready_for_b and b_ready_for_a):
        continue

    start_feasible_pairs += 1

    # Each aircraft must return to ATL before its own
    # next recorded flight, allowing 45 minutes.
    a_next_ok = (
        b["BLOCK_END_MIN"] + min_turnaround
        <= a["FOLLOWING_DEP_MIN"]
    )

    b_next_ok = (
        a["BLOCK_END_MIN"] + min_turnaround
        <= b["FOLLOWING_DEP_MIN"]
    )

    if not (a_next_ok and b_next_ok):
        continue

    fully_screened_pairs += 1

    before = (
        a["ORIGINAL_START_SHORTFALL"]
        + b["ORIGINAL_START_SHORTFALL"]
    )

    # Both exchanged starts satisfy the assumed buffer,
    # so start shortfall after swapping is zero.
    reduction = before

    if reduction > 0:

        block_swap_records.append({
            "Block A Index": i,
            "Block B Index": j,
            "Tail A": a["TAIL_NUM"],
            "Route A": f"ATL-{a['DEST']}-ATL",
            "Start A": a["BLOCK_START_MIN"],
            "Return A": a["BLOCK_END_MIN"],
            "Next Dep A": a["FOLLOWING_DEP_MIN"],
            "Tail B": b["TAIL_NUM"],
            "Route B": f"ATL-{b['DEST']}-ATL",
            "Start B": b["BLOCK_START_MIN"],
            "Return B": b["BLOCK_END_MIN"],
            "Next Dep B": b["FOLLOWING_DEP_MIN"],
            "Original Start Shortfall": before,
            "Swapped Start Shortfall": 0.0,
            "Shortfall Reduction": reduction
        })

block_swap_screen_df = pd.DataFrame(block_swap_records)

# ------------------------------------------------
# 3J.3 — Display feasibility funnel
# ------------------------------------------------

print("=" * 65)
print("STEP 3J — ROUND-TRIP BLOCK SWAP SCREENING")
print("=" * 65)

screening_summary = pd.DataFrame([
    {
        "Stage": "Eligible blocks with recorded next ATL flight",
        "Count": len(eligible_blocks)
    },
    {
        "Stage": "Distinct-tail pairs within 120 minutes",
        "Count": screened_pairs
    },
    {
        "Stage": "Both aircraft meet exchanged start buffer",
        "Count": start_feasible_pairs
    },
    {
        "Stage": "Also protect both next ATL departures",
        "Count": fully_screened_pairs
    },
    {
        "Stage": "Pairs reducing original start shortfall",
        "Count": len(block_swap_screen_df)
    }
])

display(screening_summary)

# ------------------------------------------------
# 3J.4 — Display improving block-exchange candidates
# ------------------------------------------------

if not block_swap_screen_df.empty:

    print("\nTop improving block exchanges:")

    display(
        block_swap_screen_df.sort_values(
            "Shortfall Reduction",
            ascending=False
        ).head(15)
    )

    print("\nShortfall reduction distribution:")

    display(
        block_swap_screen_df[
            "Shortfall Reduction"
        ].describe().round(2).to_frame("Minutes")
    )

    print(
        "\nDistinct aircraft involved:",
        len(
            set(block_swap_screen_df["Tail A"])
            | set(block_swap_screen_df["Tail B"])
        )
    )

else:
    print(
        "\nNo improving block exchanges satisfy "
        "these screening assumptions."
    )

print("\nSTEP 3J COMPLETE")
```

```python

# ================================================================
# BUSI6029 — ASSESSMENT 2
# STEP 3K: BINARY INTEGER PROGRAMMING — AIRCRAFT BLOCK SWAPS
# ================================================================
# Purpose:
# 1. Maximise modelled turnaround-shortfall reduction.
# 2. Select a combination of non-conflicting block swaps.
# 3. Prevent each aircraft from participating more than once.
# 4. Compare the selected swaps against the no-swap baseline.
#
# Source: 47 preliminary candidates from Step 3J.
# This is a restricted schedule-based optimisation scenario.
# NOT a validated operational or monetary savings estimate.
# ================================================================

from scipy.optimize import milp, Bounds, LinearConstraint
from scipy.sparse import lil_matrix

# ------------------------------------------------
# 3K.1 — Prepare optimisation candidates
# ------------------------------------------------

ilp_candidates = block_swap_screen_df.copy().reset_index(drop=True)

assert not ilp_candidates.empty, "No candidate swaps available."

assert (
    ilp_candidates["Shortfall Reduction"] > 0
).all(), "All selected candidates must offer positive benefit."

n_swaps = len(ilp_candidates)

aircraft_ids = sorted(
    set(ilp_candidates["Tail A"])
    | set(ilp_candidates["Tail B"])
)

aircraft_index = {
    tail: idx
    for idx, tail in enumerate(aircraft_ids)
}

n_aircraft = len(aircraft_ids)

print("=" * 65)
print("STEP 3K — AIRCRAFT BLOCK SWAP OPTIMISATION")
print("=" * 65)

print("\nCandidate swaps:", n_swaps)
print("Aircraft represented:", n_aircraft)

# ------------------------------------------------
# 3K.2 — Build aircraft conflict constraints
# ------------------------------------------------
# Each column represents a possible block swap.
# Each row represents one aircraft.
# An aircraft can appear in at most one selected swap.

constraint_matrix = lil_matrix(
    (n_aircraft, n_swaps),
    dtype=float
)

for swap_id, swap in ilp_candidates.iterrows():

    a_idx = aircraft_index[swap["Tail A"]]
    b_idx = aircraft_index[swap["Tail B"]]

    constraint_matrix[a_idx, swap_id] = 1
    constraint_matrix[b_idx, swap_id] = 1

aircraft_constraint = LinearConstraint(
    constraint_matrix.tocsr(),
    lb=np.zeros(n_aircraft),
    ub=np.ones(n_aircraft)
)

# ------------------------------------------------
# 3K.3 — Define objective and binary variables
# ------------------------------------------------
# SciPy minimises objectives, so negate the benefit.

benefits = ilp_candidates[
    "Shortfall Reduction"
].to_numpy(dtype=float)

objective = -benefits

variable_bounds = Bounds(
    lb=np.zeros(n_swaps),
    ub=np.ones(n_swaps)
)

# Every variable must be an integer (0 or 1).
integrality = np.ones(n_swaps, dtype=int)

# ------------------------------------------------
# 3K.4 — Solve the integer programme
# ------------------------------------------------

ilp_result = milp(
    c=objective,
    integrality=integrality,
    bounds=variable_bounds,
    constraints=aircraft_constraint,
    options={
        "time_limit": 60,
        "mip_rel_gap": 0.0
    }
)

print("\nSolver status:", ilp_result.status)
print("Solver message:", ilp_result.message)

if ilp_result.status != 0:
    raise RuntimeError(
        "Optimality was not confirmed. "
        "Inspect the solver result before reporting savings."
    )

# ------------------------------------------------
# 3K.5 — Extract selected aircraft exchanges
# ------------------------------------------------

selected_mask = ilp_result.x > 0.5

selected_block_swaps = (
    ilp_candidates.loc[selected_mask]
    .copy()
    .sort_values(
        "Shortfall Reduction",
        ascending=False
    )
)

# Verify no tail appears in multiple selected swaps.
selected_tails = (
    selected_block_swaps["Tail A"].tolist()
    + selected_block_swaps["Tail B"].tolist()
)

assert len(selected_tails) == len(set(selected_tails))

# ------------------------------------------------
# 3K.6 — Calculate optimisation outcomes
# ------------------------------------------------

total_reduction = selected_block_swaps[
    "Shortfall Reduction"
].sum()

affected_original_shortfall = selected_block_swaps[
    "Original Start Shortfall"
].sum()

affected_swapped_shortfall = selected_block_swaps[
    "Swapped Start Shortfall"
].sum()

optimisation_summary_df = pd.DataFrame([
    {
        "Metric": "Candidate exchange pairs",
        "Value": n_swaps
    },
    {
        "Metric": "Optimised exchanges selected",
        "Value": len(selected_block_swaps)
    },
    {
        "Metric": "Distinct aircraft reassigned",
        "Value": len(selected_tails)
    },
    {
        "Metric": "Original shortfall for selected pairs (min)",
        "Value": affected_original_shortfall
    },
    {
        "Metric": "Swapped shortfall for selected pairs (min)",
        "Value": affected_swapped_shortfall
    },
    {
        "Metric": "Total modelled shortfall reduction (min)",
        "Value": total_reduction
    }
])

print("\nOptimisation summary:")
display(optimisation_summary_df)

# ------------------------------------------------
# 3K.7 — Display selected block exchanges
# ------------------------------------------------

print("\nOptimally selected aircraft exchanges:")

display(
    selected_block_swaps[
        [
            "Tail A",
            "Route A",
            "Start A",
            "Return A",
            "Tail B",
            "Route B",
            "Start B",
            "Return B",
            "Original Start Shortfall",
            "Swapped Start Shortfall",
            "Shortfall Reduction"
        ]
    ]
)

print("\nSTEP 3K COMPLETE")
```

```python

# ================================================================
# BUSI6029 — ASSESSMENT 2
# STEP 3L: OPTIMISATION BASELINE AND SOLUTION VALIDATION
# ================================================================
# Purpose:
# 1. Calculate original shortfall across ALL eligible blocks.
# 2. Apply selected block swaps to the same population.
# 3. Validate that the 17 swaps do not conflict.
# 4. Report the full-population shortfall reduction.
#
# This measures modelled turnaround shortfall only.
# NOT actual flight-delay savings or financial savings.
# ================================================================

# ------------------------------------------------
# 3L.1 — Calculate full eligible-block baseline
# ------------------------------------------------

original_shortfalls = (
    eligible_blocks["ORIGINAL_START_SHORTFALL"]
    .to_numpy(dtype=float)
)

baseline_shortfall = original_shortfalls.sum()

baseline_affected = int(
    (original_shortfalls > 0).sum()
)

# ------------------------------------------------
# 3L.2 — Apply the selected exchanges
# ------------------------------------------------

optimised_shortfalls = original_shortfalls.copy()

used_block_indices = []

for _, swap in selected_block_swaps.iterrows():

    i = int(swap["Block A Index"])
    j = int(swap["Block B Index"])

    used_block_indices.extend([i, j])

    # Recalculate each aircraft's start-buffer shortfall
    # after exchanging complete round-trip blocks.

    aircraft_a = eligible_blocks.iloc[i]
    aircraft_b = eligible_blocks.iloc[j]

    new_shortfall_a = max(
        aircraft_b["INCOMING_ARRIVAL_MIN"]
        + min_turnaround
        - aircraft_a["BLOCK_START_MIN"],
        0
    )

    new_shortfall_b = max(
        aircraft_a["INCOMING_ARRIVAL_MIN"]
        + min_turnaround
        - aircraft_b["BLOCK_START_MIN"],
        0
    )

    optimised_shortfalls[i] = new_shortfall_a
    optimised_shortfalls[j] = new_shortfall_b

# Ensure that no block was exchanged more than once.
assert len(used_block_indices) == len(set(used_block_indices))

# ------------------------------------------------
# 3L.3 — Validate total modelled improvement
# ------------------------------------------------

optimised_total = optimised_shortfalls.sum()

recalculated_reduction = (
    baseline_shortfall - optimised_total
)

assert np.isclose(
    recalculated_reduction,
    total_reduction
), "Recalculated reduction differs from solver result."

# ------------------------------------------------
# 3L.4 — Calculate full-population comparison
# ------------------------------------------------

baseline_comparison_df = pd.DataFrame([
    {
        "Scenario": "Original schedule",
        "Eligible Blocks": len(eligible_blocks),
        "Blocks with Shortfall": baseline_affected,
        "Total Shortfall (min)": baseline_shortfall
    },
    {
        "Scenario": "Optimised block exchanges",
        "Eligible Blocks": len(eligible_blocks),
        "Blocks with Shortfall": int(
            (optimised_shortfalls > 0).sum()
        ),
        "Total Shortfall (min)": optimised_total
    }
])

improvement_pct = (
    100 * recalculated_reduction / baseline_shortfall
    if baseline_shortfall > 0 else 0
)

# ------------------------------------------------
# 3L.5 — Report validated outcomes
# ------------------------------------------------

print("=" * 65)
print("STEP 3L — OPTIMISATION BASELINE VALIDATION")
print("=" * 65)

print("\nFull eligible-block baseline comparison:")
display(baseline_comparison_df)

print("\nSelected swaps:", len(selected_block_swaps))
print("Distinct exchanged blocks:", len(set(used_block_indices)))
print("Total modelled reduction:", round(recalculated_reduction, 2))
print("Improvement over full eligible baseline:",
      f"{improvement_pct:.2f}%")

print("\nSolver-result consistency check: PASSED")
print("Block-conflict check: PASSED")

print("\nSTEP 3L COMPLETE")
```

```python

# ================================================================
# BUSI6029 — ASSESSMENT 2
# STEP 3M: AIRCRAFT-SWAP CAPACITY SENSITIVITY
# ================================================================
# Purpose:
# 1. Limit how many block exchanges may be selected.
# 2. Re-solve the binary integer programme for each limit.
# 3. Measure changes in total turnaround shortfall.
# 4. Identify diminishing returns from additional swaps.
#
# Fixed: 47 screened candidates and 45-minute buffer.
# Outcome: modelled shortfall, NOT actual delay savings.
# ================================================================

from scipy.sparse import vstack, csr_matrix

# ------------------------------------------------
# 3M.1 — Define capacity scenarios
# ------------------------------------------------

swap_capacity_scenarios = [0, 2, 4, 6, 8, 10, 12, 15, 17]

# Reuse aircraft-conflict constraints from Step 3K.
base_matrix = constraint_matrix.tocsr()

capacity_row = csr_matrix(
    np.ones((1, n_swaps))
)

combined_matrix = vstack(
    [base_matrix, capacity_row],
    format="csr"
)

# ------------------------------------------------
# 3M.2 — Solve each capacity-constrained scenario
# ------------------------------------------------

capacity_results = []

print("=" * 65)
print("STEP 3M — SWAP CAPACITY SENSITIVITY")
print("=" * 65)

for capacity in swap_capacity_scenarios:

    scenario_constraint = LinearConstraint(
        combined_matrix,
        lb=np.zeros(n_aircraft + 1),
        ub=np.concatenate([
            np.ones(n_aircraft),
            [capacity]
        ])
    )

    scenario_result = milp(
        c=objective,
        integrality=integrality,
        bounds=variable_bounds,
        constraints=scenario_constraint,
        options={
            "time_limit": 60,
            "mip_rel_gap": 0.0
        }
    )

    if scenario_result.status != 0:
        raise RuntimeError(
            f"Optimal solution not confirmed for "
            f"capacity {capacity}: {scenario_result.message}"
        )

    selected = scenario_result.x > 0.5

    selected_count = int(selected.sum())
    reduction = float(benefits[selected].sum())

    remaining_shortfall = baseline_shortfall - reduction

    capacity_results.append({
        "Maximum Swaps": capacity,
        "Selected Swaps": selected_count,
        "Shortfall Reduction (min)": round(reduction, 2),
        "Remaining Shortfall (min)": round(
            remaining_shortfall, 2
        ),
        "Improvement %": round(
            100 * reduction / baseline_shortfall, 2
        )
    })

# ------------------------------------------------
# 3M.3 — Display sensitivity results
# ------------------------------------------------

capacity_sensitivity_df = pd.DataFrame(
    capacity_results
)

print("\nOptimal results by swap-capacity limit:")

display(capacity_sensitivity_df)

# ------------------------------------------------
# 3M.4 — Calculate marginal benefits
# ------------------------------------------------

capacity_sensitivity_df["Additional Swaps"] = (
    capacity_sensitivity_df["Selected Swaps"].diff()
)

capacity_sensitivity_df["Additional Reduction (min)"] = (
    capacity_sensitivity_df[
        "Shortfall Reduction (min)"
    ].diff()
)

capacity_sensitivity_df["Reduction per Additional Swap"] = (
    capacity_sensitivity_df["Additional Reduction (min)"]
    / capacity_sensitivity_df["Additional Swaps"]
)

print("\nMarginal improvement between scenarios:")

display(
    capacity_sensitivity_df[
        [
            "Maximum Swaps",
            "Additional Swaps",
            "Additional Reduction (min)",
            "Reduction per Additional Swap"
        ]
    ].round(2)
)

# ------------------------------------------------
# 3M.5 — Check unrestricted optimum consistency
# ------------------------------------------------

unrestricted_reduction = (
    capacity_sensitivity_df.iloc[-1][
        "Shortfall Reduction (min)"
    ]
)

assert np.isclose(
    unrestricted_reduction,
    total_reduction
), "Capacity-17 result differs from the original optimum."

print("\nConsistency with Step 3K: PASSED")
print("\nSTEP 3M COMPLETE")
```

```python

# ================================================================
# BUSI6029 — ASSESSMENT 2
# STEP 3N: AIRCRAFT REASSIGNMENT PENALTY SENSITIVITY
# ================================================================
# Purpose:
# 1. Introduce a hypothetical penalty per aircraft exchange.
# 2. Re-optimise the number of swaps under each penalty.
# 3. Separate gross shortfall reduction from net objective.
# 4. Identify when further exchanges become unattractive.
#
# Fixed: 47 screened candidate exchanges.
# Baseline: 1,816 modelled shortfall minutes.
# Penalties are NOT actual airline costs.
# ================================================================

# ------------------------------------------------
# 3N.1 — Define illustrative penalty scenarios
# ------------------------------------------------

swap_penalties = [0, 5, 10, 15, 20, 25, 30, 40, 50, 60, 80]

penalty_results = []

print("=" * 65)
print("STEP 3N — REASSIGNMENT PENALTY SENSITIVITY")
print("=" * 65)

# ------------------------------------------------
# 3N.2 — Re-solve the ILP for each penalty
# ------------------------------------------------

for penalty in swap_penalties:

    # Positive benefit minus penalty, converted
    # to minimisation for scipy.optimize.milp.
    penalised_objective = -(benefits - penalty)

    scenario_result = milp(
        c=penalised_objective,
        integrality=integrality,
        bounds=variable_bounds,
        constraints=aircraft_constraint,
        options={
            "time_limit": 60,
            "mip_rel_gap": 0.0
        }
    )

    if scenario_result.status != 0:
        raise RuntimeError(
            f"Optimality not confirmed for penalty {penalty}: "
            f"{scenario_result.message}"
        )

    selected = scenario_result.x > 0.5

    swaps_selected = int(selected.sum())
    gross_reduction = float(benefits[selected].sum())
    penalty_total = penalty * swaps_selected
    net_benefit = gross_reduction - penalty_total

    penalty_results.append({
        "Penalty per Swap": penalty,
        "Selected Swaps": swaps_selected,
        "Gross Reduction (min)": round(gross_reduction, 2),
        "Total Penalty": round(penalty_total, 2),
        "Net Objective": round(net_benefit, 2),
        "Remaining Shortfall (min)": round(
            baseline_shortfall - gross_reduction, 2
        ),
        "Gross Improvement %": round(
            100 * gross_reduction / baseline_shortfall, 2
        )
    })

# ------------------------------------------------
# 3N.3 — Display sensitivity results
# ------------------------------------------------

penalty_sensitivity_df = pd.DataFrame(penalty_results)

print("\nOptimal solutions across penalty scenarios:")
display(penalty_sensitivity_df)

# ------------------------------------------------
# 3N.4 — Check zero-penalty benchmark
# ------------------------------------------------

zero_penalty_result = penalty_sensitivity_df.iloc[0]

assert np.isclose(
    zero_penalty_result["Gross Reduction (min)"],
    total_reduction
), "Zero-penalty solution differs from Step 3K."

print("\nZero-penalty consistency check: PASSED")

# ------------------------------------------------
# 3N.5 — Check for decreasing intervention uptake
# ------------------------------------------------

swap_counts = penalty_sensitivity_df[
    "Selected Swaps"
].to_numpy()

assert np.all(np.diff(swap_counts) <= 0), (
    "Selected swap counts increased as penalties rose."
)

print("Non-increasing intervention count check: PASSED")

print("\nSTEP 3N COMPLETE")
```

```python

# ================================================================
# BUSI6029 — ASSESSMENT 2
# STEP 3O: INTERMEDIATE-AIRPORT TURNAROUND AUDIT
# ================================================================
# Purpose:
# 1. Check the connecting turnaround inside each ATL round trip.
# 2. Flag ambiguous overnight intermediate-airport connections.
# 3. Evaluate how many proposed swaps pass this extra screen.
# 4. Prepare more conservative candidates for optimisation.
#
# Assumed intermediate turnaround: 45 minutes.
# Aircraft-type compatibility remains unverified.
# No new optimisation is performed in this step.
# ================================================================

# ------------------------------------------------
# 3O.1 — Retrieve return-leg scheduled departure times
# ------------------------------------------------

remote_dep_by_first_leg = (
    ordered_rotations.groupby("TAIL_NUM")[
        "CRS_DEP_MINUTES"
    ].shift(-1)
)

# Align using the preserved source row indices.
eligible_blocks_checked = eligible_blocks.copy()

eligible_blocks_checked["REMOTE_DEP_MIN"] = (
    remote_dep_by_first_leg.reindex(
        eligible_blocks_checked.index
    ).to_numpy()
)

# The outbound arrival clock time is at the
# intermediate airport.
eligible_blocks_checked["REMOTE_ARR_MIN"] = (
    eligible_blocks_checked["CRS_ARR_MINUTES"]
)

# ------------------------------------------------
# 3O.2 — Calculate intermediate-airport turnaround
# ------------------------------------------------

eligible_blocks_checked["REMOTE_GROUND_MIN"] = (
    eligible_blocks_checked["REMOTE_DEP_MIN"]
    - eligible_blocks_checked["REMOTE_ARR_MIN"]
)

remote_min_turnaround = 45

eligible_blocks_checked["REMOTE_TIMING_VALID"] = (
    eligible_blocks_checked["REMOTE_GROUND_MIN"].notna()
    & (eligible_blocks_checked["REMOTE_GROUND_MIN"] >= 0)
)

eligible_blocks_checked["REMOTE_BUFFER_OK"] = (
    eligible_blocks_checked["REMOTE_GROUND_MIN"]
    >= remote_min_turnaround
)

print("=" * 65)
print("STEP 3O — INTERMEDIATE TURNAROUND FEASIBILITY")
print("=" * 65)

print("\nEligible blocks:", len(eligible_blocks_checked))

print(
    "Blocks with unambiguous non-negative remote ground time:",
    int(eligible_blocks_checked["REMOTE_TIMING_VALID"].sum())
)

print(
    "Blocks meeting assumed 45-minute remote buffer:",
    int(eligible_blocks_checked["REMOTE_BUFFER_OK"].sum())
)

print("\nIntermediate turnaround distribution:")

display(
    eligible_blocks_checked[
        "REMOTE_GROUND_MIN"
    ].describe().to_frame("Minutes")
)

# ------------------------------------------------
# 3O.3 — Audit originally selected swap pairs
# ------------------------------------------------

selected_swap_remote_checks = []

for _, swap in selected_block_swaps.iterrows():

    a = eligible_blocks_checked.iloc[
        int(swap["Block A Index"])
    ]

    b = eligible_blocks_checked.iloc[
        int(swap["Block B Index"])
    ]

    selected_swap_remote_checks.append({
        "Tail A": swap["Tail A"],
        "Tail B": swap["Tail B"],
        "Route A": swap["Route A"],
        "Route B": swap["Route B"],
        "Remote Ground A": a["REMOTE_GROUND_MIN"],
        "Remote Ground B": b["REMOTE_GROUND_MIN"],
        "Both Pass 45-Min Remote Buffer": bool(
            a["REMOTE_BUFFER_OK"]
            and b["REMOTE_BUFFER_OK"]
        ),
        "Modelled Reduction": swap["Shortfall Reduction"]
    })

selected_remote_df = pd.DataFrame(
    selected_swap_remote_checks
)

print("\nSelected-swap intermediate turnaround checks:")

display(selected_remote_df)

print(
    "\nSelected swaps passing both remote buffers:",
    int(
        selected_remote_df[
            "Both Pass 45-Min Remote Buffer"
        ].sum()
    ),
    "of",
    len(selected_remote_df)
)

# ------------------------------------------------
# 3O.4 — Check all 47 optimisation candidates
# ------------------------------------------------

candidate_remote_ok = []

for _, swap in ilp_candidates.iterrows():

    a = eligible_blocks_checked.iloc[
        int(swap["Block A Index"])
    ]

    b = eligible_blocks_checked.iloc[
        int(swap["Block B Index"])
    ]

    candidate_remote_ok.append(
        bool(
            a["REMOTE_BUFFER_OK"]
            and b["REMOTE_BUFFER_OK"]
        )
    )

ilp_candidates_remote_checked = ilp_candidates.copy()

ilp_candidates_remote_checked["REMOTE_BUFFER_OK"] = (
    candidate_remote_ok
)

print(
    "\nAll original candidates passing remote-buffer screen:",
    int(
        ilp_candidates_remote_checked["REMOTE_BUFFER_OK"].sum()
    ),
    "of",
    len(ilp_candidates_remote_checked)
)

print("\nSTEP 3O COMPLETE")
```

```python

# ================================================================
# BUSI6029 — ASSESSMENT 2
# STEP 3P: CORRECT INTERMEDIATE TURNAROUND ALIGNMENT
# ================================================================
# Purpose:
# 1. Correct the row-index mismatch identified in Step 3O.
# 2. Match outbound legs to their actual next recorded leg.
# 3. Recalculate intermediate-airport ground times.
# 4. Recheck all 47 candidate swaps and 17 selected swaps.
#
# No model retraining or optimisation yet.
# ================================================================

# ------------------------------------------------
# 3P.1 — Build reliable same-tail flight sequence
# ------------------------------------------------

rotation_sequence = (
    rotation_day
    .sort_values(["TAIL_NUM", "SOURCE_ORDER"])
    .copy()
)

sequence_group = rotation_sequence.groupby(
    "TAIL_NUM", sort=False
)

rotation_sequence["NEXT_DEP_MIN_CORRECT"] = (
    sequence_group["CRS_DEP_MINUTES"].shift(-1)
)

rotation_sequence["NEXT_ORIGIN_CORRECT"] = (
    sequence_group["ORIGIN"].shift(-1)
)

rotation_sequence["NEXT_DEST_CORRECT"] = (
    sequence_group["DEST"].shift(-1)
)

# ------------------------------------------------
# 3P.2 — Map by stable SOURCE_ORDER, not reset index
# ------------------------------------------------

return_leg_lookup = (
    rotation_sequence
    .set_index("SOURCE_ORDER")
    [
        [
            "NEXT_DEP_MIN_CORRECT",
            "NEXT_ORIGIN_CORRECT",
            "NEXT_DEST_CORRECT"
        ]
    ]
)

eligible_blocks_corrected = eligible_blocks.copy()

assert eligible_blocks_corrected["SOURCE_ORDER"].is_unique

for col in return_leg_lookup.columns:
    eligible_blocks_corrected[col] = (
        eligible_blocks_corrected["SOURCE_ORDER"]
        .map(return_leg_lookup[col])
    )

# ------------------------------------------------
# 3P.3 — Verify outbound / return location continuity
# ------------------------------------------------

location_matches = (
    (
        eligible_blocks_corrected["DEST"]
        == eligible_blocks_corrected["NEXT_ORIGIN_CORRECT"]
    )
    &
    (
        eligible_blocks_corrected["NEXT_DEST_CORRECT"]
        == "ATL"
    )
)

assert location_matches.all(), (
    "Return-leg location mismatch. "
    "Do not continue until investigated."
)

# ------------------------------------------------
# 3P.4 — Calculate corrected remote turnaround
# ------------------------------------------------

eligible_blocks_corrected["REMOTE_GROUND_MIN"] = (
    eligible_blocks_corrected["NEXT_DEP_MIN_CORRECT"]
    - eligible_blocks_corrected["CRS_ARR_MINUTES"]
)

remote_min_turnaround = 45

eligible_blocks_corrected["REMOTE_BUFFER_OK"] = (
    eligible_blocks_corrected["REMOTE_GROUND_MIN"]
    >= remote_min_turnaround
)

# ------------------------------------------------
# 3P.5 — Recheck selected and candidate exchanges
# ------------------------------------------------

selected_checks = []

for _, swap in selected_block_swaps.iterrows():

    a = eligible_blocks_corrected.iloc[
        int(swap["Block A Index"])
    ]

    b = eligible_blocks_corrected.iloc[
        int(swap["Block B Index"])
    ]

    selected_checks.append({
        "Tail A": swap["Tail A"],
        "Tail B": swap["Tail B"],
        "Remote Ground A": a["REMOTE_GROUND_MIN"],
        "Remote Ground B": b["REMOTE_GROUND_MIN"],
        "Both Pass 45 Min": bool(
            a["REMOTE_BUFFER_OK"]
            and b["REMOTE_BUFFER_OK"]
        ),
        "Modelled Reduction": swap["Shortfall Reduction"]
    })

selected_remote_corrected_df = pd.DataFrame(
    selected_checks
)

candidate_checks = []

for _, swap in ilp_candidates.iterrows():

    a = eligible_blocks_corrected.iloc[
        int(swap["Block A Index"])
    ]

    b = eligible_blocks_corrected.iloc[
        int(swap["Block B Index"])
    ]

    candidate_checks.append(
        bool(
            a["REMOTE_BUFFER_OK"]
            and b["REMOTE_BUFFER_OK"]
        )
    )

# ------------------------------------------------
# 3P.6 — Display corrected audit
# ------------------------------------------------

print("=" * 65)
print("STEP 3P — CORRECTED INTERMEDIATE TURNAROUND")
print("=" * 65)

print("\nEligible blocks:", len(eligible_blocks_corrected))

print(
    "Non-missing remote turnaround times:",
    int(
        eligible_blocks_corrected[
            "REMOTE_GROUND_MIN"
        ].notna().sum()
    )
)

print(
    "Location matching checks:",
    int(location_matches.sum()),
    "of",
    len(location_matches)
)

print("\nCorrected remote turnaround distribution:")

display(
    eligible_blocks_corrected[
        "REMOTE_GROUND_MIN"
    ].describe().to_frame("Minutes")
)

print("\nCorrected selected-swap audit:")

display(selected_remote_corrected_df)

print(
    "\nSelected swaps passing remote buffer:",
    int(
        selected_remote_corrected_df[
            "Both Pass 45 Min"
        ].sum()
    ),
    "of",
    len(selected_remote_corrected_df)
)

print(
    "All candidates passing remote buffer:",
    sum(candidate_checks),
    "of",
    len(candidate_checks)
)

print("\nSTEP 3P COMPLETE")
```

```python

# ================================================================
# BUSI6029 — ASSESSMENT 2
# STEP 3Q: SELECTED SWAP COMMITMENT CONFLICT AUDIT
# ================================================================
# Purpose:
# 1. Revalidate all 17 selected block exchanges.
# 2. Check exchanged aircraft readiness and next departures.
# 3. Detect overlapping recorded ATL departure commitments.
# 4. Identify any conflicts needing ILP correction.
#
# Timing comparisons use ATL-local event times only.
# Still a partial, filtered-schedule feasibility audit.
# ================================================================

# ------------------------------------------------
# 3Q.1 — Prepare recorded ATL departure commitments
# ------------------------------------------------

atl_commitments = rotation_day.loc[
    rotation_day["ORIGIN"] == "ATL",
    ["SOURCE_ORDER", "TAIL_NUM",
     "CRS_DEP_MINUTES", "OP_CARRIER_FL_NUM"]
].copy()

audit_records = []

# ------------------------------------------------
# 3Q.2 — Validate each selected reciprocal block swap
# ------------------------------------------------

for _, swap in selected_block_swaps.iterrows():

    i = int(swap["Block A Index"])
    j = int(swap["Block B Index"])

    a = eligible_blocks_corrected.iloc[i]
    b = eligible_blocks_corrected.iloc[j]

    for original, assigned in [(a, b), (b, a)]:

        tail = original["TAIL_NUM"]

        new_start = float(assigned["BLOCK_START_MIN"])
        new_end = float(assigned["BLOCK_END_MIN"])

        ready_time = (
            float(original["INCOMING_ARRIVAL_MIN"])
            + min_turnaround
        )

        next_departure = float(
            original["FOLLOWING_DEP_MIN"]
        )

        # The aircraft gives up its own original block.
        # Exclude that original outbound departure.
        other_departures = atl_commitments.loc[
            (atl_commitments["TAIL_NUM"] == tail)
            & (
                atl_commitments["SOURCE_ORDER"]
                != original["SOURCE_ORDER"]
            )
        ]

        overlap = other_departures.loc[
            (other_departures["CRS_DEP_MINUTES"] > new_start)
            & (other_departures["CRS_DEP_MINUTES"] < new_end)
        ]

        audit_records.append({
            "Aircraft": tail,
            "Original Route": f"ATL-{original['DEST']}-ATL",
            "New Route": f"ATL-{assigned['DEST']}-ATL",
            "New Start": new_start,
            "New Return": new_end,
            "Ready Before Start": bool(
                ready_time <= new_start
            ),
            "Next Departure Protected": bool(
                new_end + min_turnaround <= next_departure
            ),
            "Other ATL Departures During Block": len(overlap),
            "Overlap Flight Numbers": (
                ", ".join(
                    overlap["OP_CARRIER_FL_NUM"]
                    .astype(str).tolist()
                )
                if len(overlap) else ""
            )
        })

# ------------------------------------------------
# 3Q.3 — Summarise feasibility checks
# ------------------------------------------------

commitment_audit_df = pd.DataFrame(audit_records)

commitment_audit_df["Pass Recorded Commitment Checks"] = (
    commitment_audit_df["Ready Before Start"]
    & commitment_audit_df["Next Departure Protected"]
    & (
        commitment_audit_df[
            "Other ATL Departures During Block"
        ] == 0
    )
)

print("=" * 65)
print("STEP 3Q — SELECTED SWAP COMMITMENT AUDIT")
print("=" * 65)

print("\nAircraft reassignment records:",
      len(commitment_audit_df))

print(
    "Ready before exchanged block:",
    int(commitment_audit_df["Ready Before Start"].sum())
)

print(
    "Next recorded departure protected:",
    int(commitment_audit_df["Next Departure Protected"].sum())
)

print(
    "No overlapping recorded ATL departures:",
    int((
        commitment_audit_df[
            "Other ATL Departures During Block"
        ] == 0
    ).sum())
)

print(
    "Pass all recorded commitment checks:",
    int(
        commitment_audit_df[
            "Pass Recorded Commitment Checks"
        ].sum()
    )
)

# ------------------------------------------------
# 3Q.4 — Display violations, if any
# ------------------------------------------------

commitment_violations_df = commitment_audit_df.loc[
    ~commitment_audit_df["Pass Recorded Commitment Checks"]
]

print("\nCommitment violations:")
display(commitment_violations_df)

# ------------------------------------------------
# 3Q.5 — Verify unique aircraft assignments
# ------------------------------------------------

assert commitment_audit_df["Aircraft"].is_unique, (
    "An aircraft appears more than once in selected swaps."
)

print("\nUnique aircraft assignment check: PASSED")
print("\nSTEP 3Q COMPLETE")
```

```python

# ================================================================
# BUSI6029 — ASSESSMENT 2
# STEP 3R: INTERMEDIATE-AIRPORT TURNAROUND ROBUSTNESS
# ================================================================
# Purpose:
# 1. Test stricter hypothetical remote-airport turnarounds.
# 2. Filter candidate exchanges at each threshold.
# 3. Re-solve the non-overlapping aircraft swap ILP.
# 4. Compare optimal outcomes with a common baseline.
#
# ATL turnaround assumption remains 45 minutes.
# Remote-airport thresholds are illustrative, not
# verified Delta operational requirements.
# ================================================================

# ------------------------------------------------
# 3R.1 — Define remote turnaround scenarios
# ------------------------------------------------

remote_buffer_scenarios = [45, 50, 55, 60, 65, 70]

remote_robustness_results = []

print("=" * 65)
print("STEP 3R — REMOTE TURNAROUND ROBUSTNESS")
print("=" * 65)

# ------------------------------------------------
# 3R.2 — Rebuild candidate eligibility by scenario
# ------------------------------------------------

for remote_buffer in remote_buffer_scenarios:

    eligible_candidate_ids = []

    for candidate_id, swap in ilp_candidates.iterrows():

        a = eligible_blocks_corrected.iloc[
            int(swap["Block A Index"])
        ]

        b = eligible_blocks_corrected.iloc[
            int(swap["Block B Index"])
        ]

        if (
            a["REMOTE_GROUND_MIN"] >= remote_buffer
            and b["REMOTE_GROUND_MIN"] >= remote_buffer
        ):
            eligible_candidate_ids.append(candidate_id)

    # ------------------------------------------------
    # 3R.3 — Solve restricted binary optimisation
    # ------------------------------------------------

    if eligible_candidate_ids:

        scenario_objective = objective[
            eligible_candidate_ids
        ]

        scenario_constraints = LinearConstraint(
            constraint_matrix.tocsr()[
                :, eligible_candidate_ids
            ],
            lb=np.zeros(n_aircraft),
            ub=np.ones(n_aircraft)
        )

        scenario_result = milp(
            c=scenario_objective,
            integrality=np.ones(
                len(eligible_candidate_ids),
                dtype=int
            ),
            bounds=Bounds(
                np.zeros(len(eligible_candidate_ids)),
                np.ones(len(eligible_candidate_ids))
            ),
            constraints=scenario_constraints,
            options={
                "time_limit": 60,
                "mip_rel_gap": 0.0
            }
        )

        if scenario_result.status != 0:
            raise RuntimeError(
                f"Optimality not confirmed for "
                f"remote buffer {remote_buffer}: "
                f"{scenario_result.message}"
            )

        selected_ids = np.array(
            eligible_candidate_ids
        )[scenario_result.x > 0.5]

        selected_count = len(selected_ids)

        reduction = float(
            benefits[selected_ids].sum()
        )

    else:
        selected_count = 0
        reduction = 0.0

    # ------------------------------------------------
    # 3R.4 — Store consistent scenario outcomes
    # ------------------------------------------------

    remote_robustness_results.append({
        "Remote Minimum (min)": remote_buffer,
        "Eligible Candidate Swaps": len(
            eligible_candidate_ids
        ),
        "Optimal Swaps": selected_count,
        "Shortfall Reduction (min)": round(
            reduction, 2
        ),
        "Remaining Shortfall (min)": round(
            baseline_shortfall - reduction, 2
        ),
        "Improvement %": round(
            100 * reduction / baseline_shortfall, 2
        )
    })

# ------------------------------------------------
# 3R.5 — Display scenario sensitivity table
# ------------------------------------------------

remote_robustness_df = pd.DataFrame(
    remote_robustness_results
)

print("\nOptimisation under stricter remote buffers:")
display(remote_robustness_df)

# ------------------------------------------------
# 3R.6 — Validate the baseline scenario
# ------------------------------------------------

baseline_row = remote_robustness_df.loc[
    remote_robustness_df[
        "Remote Minimum (min)"
    ] == 45
].iloc[0]

assert baseline_row["Eligible Candidate Swaps"] == len(
    ilp_candidates
)

assert np.isclose(
    baseline_row["Shortfall Reduction (min)"],
    total_reduction
)

assert (
    remote_robustness_df[
        "Shortfall Reduction (min)"
    ].diff().dropna() <= 0
).all(), (
    "Optimal reduction increased after tightening "
    "the remote turnaround requirement."
)

print("\n45-minute baseline consistency: PASSED")
print("Non-increasing benefit check: PASSED")
print("\nSTEP 3R COMPLETE")
```

```python

# ================================================================
# BUSI6029 — ASSESSMENT 2
# STEP 3S: BASELINE FEASIBILITY UNDER REMOTE TURNAROUND RULES
# ================================================================
# Purpose:
# 1. Audit all 171 original ATL return blocks.
# 2. Identify historical blocks below each remote minimum.
# 3. Measure the original ATL start shortfall in each group.
# 4. Distinguish baseline infeasibility from swap sensitivity.
#
# No optimisation is changed or rerun in this step.
# Thresholds are illustrative, not verified Delta rules.
# ================================================================

# ------------------------------------------------
# 3S.1 — Establish the full original baseline
# ------------------------------------------------

baseline_blocks_audit = eligible_blocks_corrected.copy()

assert len(baseline_blocks_audit) == 171

assert np.isclose(
    baseline_blocks_audit[
        "ORIGINAL_START_SHORTFALL"
    ].sum(),
    baseline_shortfall
)

# ------------------------------------------------
# 3S.2 — Compare baseline feasibility by threshold
# ------------------------------------------------

baseline_feasibility_records = []

for remote_minimum in remote_buffer_scenarios:

    passes = (
        baseline_blocks_audit["REMOTE_GROUND_MIN"]
        >= remote_minimum
    )

    fails = ~passes

    baseline_feasibility_records.append({
        "Remote Minimum (min)": remote_minimum,
        "Original Blocks Meeting Minimum": int(passes.sum()),
        "Original Blocks Below Minimum": int(fails.sum()),
        "Baseline Pass Rate %": round(
            100 * passes.mean(), 2
        ),
        "ATL Start Shortfall in Passing Blocks": round(
            baseline_blocks_audit.loc[
                passes, "ORIGINAL_START_SHORTFALL"
            ].sum(), 2
        ),
        "ATL Start Shortfall in Failing Blocks": round(
            baseline_blocks_audit.loc[
                fails, "ORIGINAL_START_SHORTFALL"
            ].sum(), 2
        )
    })

baseline_feasibility_df = pd.DataFrame(
    baseline_feasibility_records
)

# ------------------------------------------------
# 3S.3 — Combine with previous optimisation results
# ------------------------------------------------

turnaround_interpretation_df = (
    baseline_feasibility_df.merge(
        remote_robustness_df[
            [
                "Remote Minimum (min)",
                "Eligible Candidate Swaps",
                "Optimal Swaps",
                "Shortfall Reduction (min)",
                "Improvement %"
            ]
        ],
        on="Remote Minimum (min)",
        validate="one_to_one"
    )
)

# ------------------------------------------------
# 3S.4 — Display main audit
# ------------------------------------------------

print("=" * 65)
print("STEP 3S — BASELINE TURNAROUND FEASIBILITY")
print("=" * 65)

print("\nOriginal blocks:", len(baseline_blocks_audit))
print("Original ATL start shortfall:", baseline_shortfall)

print("\nCombined baseline and optimisation audit:")

display(turnaround_interpretation_df)

# ------------------------------------------------
# 3S.5 — Inspect blocks below the 70-minute threshold
# ------------------------------------------------

strict_threshold = 70

below_70 = baseline_blocks_audit.loc[
    baseline_blocks_audit["REMOTE_GROUND_MIN"]
    < strict_threshold
].copy()

print("\nHistorical blocks below 70-minute remote minimum:",
      len(below_70))

print("\nExamples of original blocks below 70 minutes:")

display(
    below_70[
        [
            "TAIL_NUM",
            "DEST",
            "REMOTE_GROUND_MIN",
            "BLOCK_START_MIN",
            "BLOCK_END_MIN",
            "ORIGINAL_START_SHORTFALL"
        ]
    ].sort_values(
        "REMOTE_GROUND_MIN"
    ).head(15)
)

# ------------------------------------------------
# 3S.6 — Confirm consistency
# ------------------------------------------------

assert (
    baseline_feasibility_df[
        "Original Blocks Meeting Minimum"
    ]
    + baseline_feasibility_df[
        "Original Blocks Below Minimum"
    ]
    == len(baseline_blocks_audit)
).all()

assert np.allclose(
    baseline_feasibility_df[
        "ATL Start Shortfall in Passing Blocks"
    ]
    + baseline_feasibility_df[
        "ATL Start Shortfall in Failing Blocks"
    ],
    baseline_shortfall
)

print("\nBaseline accounting checks: PASSED")
print("\nSTEP 3S COMPLETE")
```

```python

# ================================================================
# BUSI6029 — ASSESSMENT 2
# STEP 3T: RECOVERY SCOPE AND BASELINE COMPARABILITY
# ================================================================
# Purpose:
# 1. Separate feasible-subset and full-population outcomes.
# 2. Quantify shortfall excluded under stricter assumptions.
# 3. Verify optimisation savings stay within eligible scope.
# 4. Produce a defensible sensitivity reporting table.
#
# IMPORTANT:
# Eligible-subset percentages use changing populations.
# They must NOT be compared as if they measure the same
# 171-block population across all scenarios.
# ================================================================

# ------------------------------------------------
# 3T.1 — Build combined interpretation table
# ------------------------------------------------

recovery_scope_records = []

for _, scenario in turnaround_interpretation_df.iterrows():

    threshold = int(scenario["Remote Minimum (min)"])

    passing_blocks = int(
        scenario["Original Blocks Meeting Minimum"]
    )

    failing_blocks = int(
        scenario["Original Blocks Below Minimum"]
    )

    eligible_shortfall = float(
        scenario["ATL Start Shortfall in Passing Blocks"]
    )

    excluded_shortfall = float(
        scenario["ATL Start Shortfall in Failing Blocks"]
    )

    reduction = float(
        scenario["Shortfall Reduction (min)"]
    )

    # The optimisation selects pairs in which BOTH
    # round-trip blocks meet the remote threshold.
    assert reduction <= eligible_shortfall + 1e-8, (
        f"Reduction exceeds eligible shortfall "
        f"at threshold {threshold}."
    )

    recovery_scope_records.append({
        "Remote Minimum (min)": threshold,
        "Passing Blocks": passing_blocks,
        "Failing Blocks": failing_blocks,
        "Eligible Baseline Shortfall": eligible_shortfall,
        "Excluded Baseline Shortfall": excluded_shortfall,
        "Optimal Swaps": int(scenario["Optimal Swaps"]),
        "Reduction (min)": reduction,
        "Full Baseline Improvement %": round(
            100 * reduction / baseline_shortfall, 2
        ),
        "Eligible Subset Improvement %": (
            round(100 * reduction / eligible_shortfall, 2)
            if eligible_shortfall > 0 else np.nan
        ),
        "Excluded Share of Baseline Shortfall %": round(
            100 * excluded_shortfall / baseline_shortfall, 2
        )
    })

recovery_scope_df = pd.DataFrame(
    recovery_scope_records
)

# ------------------------------------------------
# 3T.2 — Calculate residual shortfall decomposition
# ------------------------------------------------

recovery_scope_df["Remaining Eligible Shortfall"] = (
    recovery_scope_df["Eligible Baseline Shortfall"]
    - recovery_scope_df["Reduction (min)"]
)

recovery_scope_df["Total Remaining Shortfall"] = (
    recovery_scope_df["Remaining Eligible Shortfall"]
    + recovery_scope_df["Excluded Baseline Shortfall"]
)

# ------------------------------------------------
# 3T.3 — Validate accounting identities
# ------------------------------------------------

assert np.allclose(
    recovery_scope_df["Eligible Baseline Shortfall"]
    + recovery_scope_df["Excluded Baseline Shortfall"],
    baseline_shortfall
)

assert np.allclose(
    recovery_scope_df["Total Remaining Shortfall"],
    baseline_shortfall - recovery_scope_df["Reduction (min)"]
)

assert (
    recovery_scope_df["Remaining Eligible Shortfall"] >= -1e-8
).all()

# ------------------------------------------------
# 3T.4 — Display scope-aware sensitivity findings
# ------------------------------------------------

print("=" * 65)
print("STEP 3T — RECOVERY SCOPE AND BASELINE AUDIT")
print("=" * 65)

display(
    recovery_scope_df[
        [
            "Remote Minimum (min)",
            "Passing Blocks",
            "Failing Blocks",
            "Eligible Baseline Shortfall",
            "Excluded Baseline Shortfall",
            "Reduction (min)",
            "Full Baseline Improvement %",
            "Eligible Subset Improvement %",
            "Excluded Share of Baseline Shortfall %"
        ]
    ]
)

print("\nResidual shortfall decomposition:")

display(
    recovery_scope_df[
        [
            "Remote Minimum (min)",
            "Remaining Eligible Shortfall",
            "Excluded Baseline Shortfall",
            "Total Remaining Shortfall"
        ]
    ]
)

print("\nBaseline accounting checks: PASSED")
print("Subset-benefit consistency checks: PASSED")
print("\nSTEP 3T COMPLETE")
```

```python

# ================================================================
# BUSI6029 — ASSESSMENT 2
# STEP 3U: SWAP SAFETY MARGINS AND COMPATIBILITY AUDIT
# ================================================================
# Purpose:
# 1. Measure remaining time margins for selected exchanges.
# 2. Identify swaps sensitive to small schedule disruptions.
# 3. Distinguish scheduling checks from aircraft compatibility.
# 4. Prepare operational risk evidence for the final report.
#
# All margins use scheduled or estimated flight events.
# Actual spare aircraft, equipment and maintenance
# compatibility cannot be verified in the current dataset.
# ================================================================

margin_records = []

for _, swap in selected_block_swaps.iterrows():

    a = eligible_blocks_corrected.iloc[
        int(swap["Block A Index"])
    ]

    b = eligible_blocks_corrected.iloc[
        int(swap["Block B Index"])
    ]

    # Aircraft A operates B's round trip.
    start_margin_a = (
        b["BLOCK_START_MIN"]
        - a["INCOMING_ARRIVAL_MIN"]
        - min_turnaround
    )

    next_margin_a = (
        a["FOLLOWING_DEP_MIN"]
        - b["BLOCK_END_MIN"]
        - min_turnaround
    )

    # Aircraft B operates A's round trip.
    start_margin_b = (
        a["BLOCK_START_MIN"]
        - b["INCOMING_ARRIVAL_MIN"]
        - min_turnaround
    )

    next_margin_b = (
        b["FOLLOWING_DEP_MIN"]
        - a["BLOCK_END_MIN"]
        - min_turnaround
    )

    # Remote turnarounds belong to the assigned
    # flight blocks. Common minimum = 45 minutes.
    remote_margin_a = (
        b["REMOTE_GROUND_MIN"] - remote_min_turnaround
    )

    remote_margin_b = (
        a["REMOTE_GROUND_MIN"] - remote_min_turnaround
    )

    margins = [
        start_margin_a,
        start_margin_b,
        next_margin_a,
        next_margin_b,
        remote_margin_a,
        remote_margin_b
    ]

    margin_records.append({
        "Tail A": swap["Tail A"],
        "Tail B": swap["Tail B"],
        "Route A": swap["Route A"],
        "Route B": swap["Route B"],
        "Modelled Reduction": swap["Shortfall Reduction"],
        "ATL Start Margin A": start_margin_a,
        "ATL Start Margin B": start_margin_b,
        "Next Flight Margin A": next_margin_a,
        "Next Flight Margin B": next_margin_b,
        "Remote Margin A": remote_margin_a,
        "Remote Margin B": remote_margin_b,
        "Minimum Margin": min(margins),
        "Pass Extra 15-Min Margin": min(margins) >= 15,
        "Pass Extra 30-Min Margin": min(margins) >= 30,
        "Equipment Compatibility": "UNVERIFIED"
    })

swap_margin_audit_df = pd.DataFrame(margin_records)

# ------------------------------------------------
# 3U.2 — Validate and report scheduling margins
# ------------------------------------------------

print("=" * 65)
print("STEP 3U — SWAP SAFETY MARGIN AUDIT")
print("=" * 65)

assert (
    swap_margin_audit_df["Minimum Margin"] >= -1e-8
).all(), "A selected swap has a negative scheduling margin."

print("\nSelected exchanges:", len(swap_margin_audit_df))

print(
    "Minimum margin across selected swaps:",
    round(swap_margin_audit_df["Minimum Margin"].min(), 2)
)

print(
    "Median minimum margin:",
    round(swap_margin_audit_df["Minimum Margin"].median(), 2)
)

print(
    "Swaps with less than 15 minutes extra margin:",
    int((swap_margin_audit_df["Minimum Margin"] < 15).sum())
)

print(
    "Swaps with less than 30 minutes extra margin:",
    int((swap_margin_audit_df["Minimum Margin"] < 30).sum())
)

print(
    "Swaps with verified equipment compatibility:",
    0,
    "— equipment attributes not available"
)

# ------------------------------------------------
# 3U.3 — Inspect most scheduling-sensitive exchanges
# ------------------------------------------------

print("\nSelected exchanges ordered by minimum margin:")

display(
    swap_margin_audit_df[
        [
            "Tail A",
            "Tail B",
            "Modelled Reduction",
            "ATL Start Margin A",
            "ATL Start Margin B",
            "Next Flight Margin A",
            "Next Flight Margin B",
            "Remote Margin A",
            "Remote Margin B",
            "Minimum Margin",
            "Equipment Compatibility"
        ]
    ].sort_values("Minimum Margin").round(2)
)

print("\nSTEP 3U COMPLETE")
```

```python

# ================================================================
# BUSI6029 — ASSESSMENT 2
# STEP 3V: ROBUST AIRCRAFT SWAP OPTIMISATION
# ================================================================
# Purpose:
# 1. Recalculate scheduling margins for ALL 47 candidate swaps.
# 2. Require extra safety margins of 0, 5, 10, 15, 20 min.
# 3. Re-solve the binary ILP at each margin.
# 4. Compare robust solutions with the same baseline.
#
# Equipment compatibility remains UNVERIFIED.
# Historical inbound delays make this a retrospective scenario.
# ================================================================

# ------------------------------------------------
# 3V.1 — Calculate six margins for every candidate
# ------------------------------------------------

all_candidate_margin_records = []

for candidate_id, swap in ilp_candidates.iterrows():

    a = eligible_blocks_corrected.iloc[
        int(swap["Block A Index"])
    ]

    b = eligible_blocks_corrected.iloc[
        int(swap["Block B Index"])
    ]

    margins = {
        "ATL Start A": (
            b["BLOCK_START_MIN"]
            - a["INCOMING_ARRIVAL_MIN"]
            - min_turnaround
        ),
        "ATL Start B": (
            a["BLOCK_START_MIN"]
            - b["INCOMING_ARRIVAL_MIN"]
            - min_turnaround
        ),
        "Next Flight A": (
            a["FOLLOWING_DEP_MIN"]
            - b["BLOCK_END_MIN"]
            - min_turnaround
        ),
        "Next Flight B": (
            b["FOLLOWING_DEP_MIN"]
            - a["BLOCK_END_MIN"]
            - min_turnaround
        ),
        "Remote A": (
            b["REMOTE_GROUND_MIN"]
            - remote_min_turnaround
        ),
        "Remote B": (
            a["REMOTE_GROUND_MIN"]
            - remote_min_turnaround
        )
    }

    all_candidate_margin_records.append({
        "Candidate ID": candidate_id,
        "Tail A": swap["Tail A"],
        "Tail B": swap["Tail B"],
        "Benefit": swap["Shortfall Reduction"],
        "Minimum Margin": min(margins.values()),
        "Binding Constraint": min(
            margins, key=margins.get
        )
    })

candidate_margin_df = pd.DataFrame(
    all_candidate_margin_records
)

assert len(candidate_margin_df) == len(ilp_candidates)

assert (
    candidate_margin_df["Minimum Margin"] >= -1e-8
).all(), "Original candidates contain negative margins."

# ------------------------------------------------
# 3V.2 — Re-solve at different minimum safety margins
# ------------------------------------------------

extra_margin_scenarios = [0, 5, 10, 15, 20]

robust_results = []
robust_selected_ids = {}

print("=" * 65)
print("STEP 3V — SAFETY-MARGIN OPTIMISATION")
print("=" * 65)

for extra_margin in extra_margin_scenarios:

    candidate_ids = candidate_margin_df.loc[
        candidate_margin_df["Minimum Margin"]
        >= extra_margin - 1e-8,
        "Candidate ID"
    ].astype(int).to_numpy()

    if len(candidate_ids) == 0:
        chosen_ids = np.array([], dtype=int)
        reduction = 0.0

    else:
        scenario_result = milp(
            c=objective[candidate_ids],
            integrality=np.ones(
                len(candidate_ids), dtype=int
            ),
            bounds=Bounds(
                np.zeros(len(candidate_ids)),
                np.ones(len(candidate_ids))
            ),
            constraints=LinearConstraint(
                constraint_matrix.tocsr()[:, candidate_ids],
                lb=np.zeros(n_aircraft),
                ub=np.ones(n_aircraft)
            ),
            options={
                "time_limit": 60,
                "mip_rel_gap": 0.0
            }
        )

        if scenario_result.status != 0:
            raise RuntimeError(
                f"Optimality not confirmed at "
                f"{extra_margin}-minute extra margin: "
                f"{scenario_result.message}"
            )

        chosen_ids = candidate_ids[
            scenario_result.x > 0.5
        ]

        reduction = float(
            benefits[chosen_ids].sum()
        )

    robust_selected_ids[extra_margin] = chosen_ids

    robust_results.append({
        "Extra Margin (min)": extra_margin,
        "Eligible Candidate Pairs": len(candidate_ids),
        "Optimal Swaps": len(chosen_ids),
        "Reduction (min)": reduction,
        "Remaining Shortfall (min)": (
            baseline_shortfall - reduction
        ),
        "Full Baseline Improvement %": round(
            100 * reduction / baseline_shortfall, 2
        )
    })

# ------------------------------------------------
# 3V.3 — Display robust optimisation results
# ------------------------------------------------

safety_margin_sensitivity_df = pd.DataFrame(
    robust_results
)

print("\nOptimal results by additional safety margin:")
display(safety_margin_sensitivity_df)

# ------------------------------------------------
# 3V.4 — Validate monotonicity and original solution
# ------------------------------------------------

assert np.isclose(
    safety_margin_sensitivity_df.iloc[0][
        "Reduction (min)"
    ],
    total_reduction
)

assert (
    safety_margin_sensitivity_df[
        "Reduction (min)"
    ].diff().dropna() <= 1e-8
).all()

print("\nZero-margin benchmark: PASSED")
print("Non-increasing objective check: PASSED")

# ------------------------------------------------
# 3V.5 — Inspect the 10- and 15-minute solutions
# ------------------------------------------------

for extra_margin in [10, 15]:

    chosen_ids = robust_selected_ids[extra_margin]

    print(
        f"\nSelected swaps with {extra_margin}-minute "
        f"additional margin:"
    )

    if len(chosen_ids):

        result_table = (
            ilp_candidates.iloc[chosen_ids][
                [
                    "Tail A",
                    "Route A",
                    "Tail B",
                    "Route B",
                    "Shortfall Reduction"
                ]
            ]
            .copy()
        )

        result_table["Minimum Margin"] = (
            candidate_margin_df.set_index("Candidate ID")
            .loc[chosen_ids, "Minimum Margin"]
            .to_numpy()
        )

        display(
            result_table.sort_values(
                "Shortfall Reduction",
                ascending=False
            )
        )

    else:
        print("No qualifying exchanges.")

print("\nSTEP 3V COMPLETE")
```

```python

# ================================================================
# BUSI6029 — ASSESSMENT 2
# STEP 3W: INDEPENDENT ROBUST-SOLUTION VALIDATION
# ================================================================
# Purpose:
# 1. Validate the 5-, 10- and 15-minute ILP solutions.
# 2. Check unique aircraft and block assignments.
# 3. Check readiness, next flight and other ATL departures.
# 4. Recalculate full-baseline shortfall reductions.
#
# This is a retrospective, filtered-schedule audit.
# Equipment, crews and maintenance remain unverified.
# ================================================================

print("=" * 65)
print("STEP 3W — ROBUST RECOVERY SOLUTION VALIDATION")
print("=" * 65)

validation_records = []
robust_audit_violations = []

for extra_margin in [5, 10, 15]:

    chosen_ids = robust_selected_ids[extra_margin]
    chosen_swaps = ilp_candidates.iloc[chosen_ids]

    used_tails = []
    used_blocks = []
    new_shortfalls = original_shortfalls.copy()

    assignment_count = 0
    overlap_count = 0
    readiness_failures = 0
    following_failures = 0
    margin_failures = 0

    for _, swap in chosen_swaps.iterrows():

        i = int(swap["Block A Index"])
        j = int(swap["Block B Index"])

        a = eligible_blocks_corrected.iloc[i]
        b = eligible_blocks_corrected.iloc[j]

        used_blocks.extend([i, j])
        used_tails.extend([a["TAIL_NUM"], b["TAIL_NUM"]])

        for original, assigned, original_idx in [
            (a, b, i),
            (b, a, j)
        ]:

            assignment_count += 1

            start = float(assigned["BLOCK_START_MIN"])
            finish = float(assigned["BLOCK_END_MIN"])

            start_margin = (
                start
                - float(original["INCOMING_ARRIVAL_MIN"])
                - min_turnaround
            )

            next_margin = (
                float(original["FOLLOWING_DEP_MIN"])
                - finish
                - min_turnaround
            )

            remote_margin = (
                float(assigned["REMOTE_GROUND_MIN"])
                - remote_min_turnaround
            )

            other_departures = atl_commitments.loc[
                (atl_commitments["TAIL_NUM"]
                 == original["TAIL_NUM"])
                & (atl_commitments["SOURCE_ORDER"]
                   != original["SOURCE_ORDER"])
                & (atl_commitments["CRS_DEP_MINUTES"] > start)
                & (atl_commitments["CRS_DEP_MINUTES"] < finish)
            ]

            failed = []

            if start_margin < extra_margin - 1e-8:
                readiness_failures += 1
                failed.append("Start margin")

            if next_margin < extra_margin - 1e-8:
                following_failures += 1
                failed.append("Next-flight margin")

            if remote_margin < extra_margin - 1e-8:
                margin_failures += 1
                failed.append("Remote margin")

            if len(other_departures):
                overlap_count += 1
                failed.append("Overlapping ATL departure")

            if failed:
                robust_audit_violations.append({
                    "Extra Margin": extra_margin,
                    "Aircraft": original["TAIL_NUM"],
                    "Failure": ", ".join(failed)
                })

            # Independently recalculate new start shortfall.
            new_shortfalls[original_idx] = max(
                float(original["INCOMING_ARRIVAL_MIN"])
                + min_turnaround
                - start,
                0.0
            )

    unique_tails = len(used_tails) == len(set(used_tails))
    unique_blocks = len(used_blocks) == len(set(used_blocks))

    actual_reduction = (
        baseline_shortfall - new_shortfalls.sum()
    )

    solver_reduction = float(
        benefits[chosen_ids].sum()
    )

    objective_matches = np.isclose(
        actual_reduction,
        solver_reduction
    )

    scenario_passed = (
        unique_tails
        and unique_blocks
        and overlap_count == 0
        and readiness_failures == 0
        and following_failures == 0
        and margin_failures == 0
        and objective_matches
    )

    validation_records.append({
        "Extra Margin (min)": extra_margin,
        "Selected Swaps": len(chosen_swaps),
        "Aircraft Checked": assignment_count,
        "Unique Aircraft": unique_tails,
        "Unique Blocks": unique_blocks,
        "Overlaps": overlap_count,
        "Timing Failures": (
            readiness_failures
            + following_failures
            + margin_failures
        ),
        "Recalculated Reduction": round(
            actual_reduction, 2
        ),
        "Improvement %": round(
            100 * actual_reduction / baseline_shortfall, 2
        ),
        "All Checks Passed": scenario_passed
    })

robust_validation_df = pd.DataFrame(validation_records)
robust_violations_df = pd.DataFrame(
    robust_audit_violations,
    columns=["Extra Margin", "Aircraft", "Failure"]
)

print("\nIndependent scenario validation:")
display(robust_validation_df)

print("\nRecorded violations:")
display(robust_violations_df)

assert robust_validation_df["All Checks Passed"].all(), (
    "At least one scenario failed independent validation."
)

print("\nAll tested scenario checks: PASSED")
print("\nSTEP 3W COMPLETE")
```

```python

# ================================================================
# BUSI6029 — ASSESSMENT 2
# STEP 3X: DECISION-TIME INFORMATION AVAILABILITY AUDIT
# ================================================================
# Purpose:
# 1. Identify when a reciprocal swap must be decided.
# 2. Compare that deadline with both estimated inbound arrivals.
# 3. Flag swaps potentially dependent on future information.
# 4. Test 0-, 15-, and 30-minute decision lead times.
#
# The inbound arrival estimates use realised PREV_ARR_DELAY.
# This audit does NOT establish historical information timestamps.
# ================================================================

decision_lead_scenarios = [0, 15, 30]
decision_audit_records = []

for scenario_margin in [0, 5, 10, 15]:

    chosen_ids = robust_selected_ids[scenario_margin]

    for candidate_id in chosen_ids:

        swap = ilp_candidates.iloc[int(candidate_id)]

        a = eligible_blocks_corrected.iloc[
            int(swap["Block A Index"])
        ]
        b = eligible_blocks_corrected.iloc[
            int(swap["Block B Index"])
        ]

        first_departure = min(
            float(a["BLOCK_START_MIN"]),
            float(b["BLOCK_START_MIN"])
        )

        latest_inbound_arrival = max(
            float(a["INCOMING_ARRIVAL_MIN"]),
            float(b["INCOMING_ARRIVAL_MIN"])
        )

        for lead in decision_lead_scenarios:

            decision_deadline = first_departure - lead

            arrivals_known_by_deadline = (
                latest_inbound_arrival <= decision_deadline
            )

            decision_audit_records.append({
                "Safety Margin": scenario_margin,
                "Candidate ID": int(candidate_id),
                "Tail A": swap["Tail A"],
                "Tail B": swap["Tail B"],
                "Decision Lead (min)": lead,
                "First Block Departure": first_departure,
                "Decision Deadline": decision_deadline,
                "Latest Estimated Inbound Arrival": (
                    latest_inbound_arrival
                ),
                "Both Inbound Arrivals Before Deadline": bool(
                    arrivals_known_by_deadline
                ),
                "Modelled Reduction": float(
                    swap["Shortfall Reduction"]
                )
            })

decision_timing_audit_df = pd.DataFrame(
    decision_audit_records
)

# ------------------------------------------------
# 3X.2 — Summarise results for each solution
# ------------------------------------------------

decision_summary_df = (
    decision_timing_audit_df.groupby(
        ["Safety Margin", "Decision Lead (min)"],
        as_index=False
    )
    .agg(
        Selected_Swaps=("Candidate ID", "count"),
        Swaps_With_Both_Inbounds_Arrived=(
            "Both Inbound Arrivals Before Deadline",
            "sum"
        ),
        Total_Selected_Benefit=(
            "Modelled Reduction", "sum"
        )
    )
)

decision_summary_df["Arrival_Confirmed_Benefit"] = (
    decision_timing_audit_df.loc[
        decision_timing_audit_df[
            "Both Inbound Arrivals Before Deadline"
        ]
    ]
    .groupby(
        ["Safety Margin", "Decision Lead (min)"]
    )["Modelled Reduction"]
    .sum()
    .reindex(
        pd.MultiIndex.from_frame(
            decision_summary_df[
                ["Safety Margin", "Decision Lead (min)"]
            ]
        ),
        fill_value=0
    )
    .to_numpy()
)

# ------------------------------------------------
# 3X.3 — Print audit results
# ------------------------------------------------

print("=" * 65)
print("STEP 3X — DECISION-TIME INFORMATION AUDIT")
print("=" * 65)

print("\nDecision-time feasibility by scenario:")

display(decision_summary_df)

print("\nPotential hindsight-dependent swaps (10-minute safety case,")
print("15-minute decision lead):")

display(
    decision_timing_audit_df.loc[
        (decision_timing_audit_df["Safety Margin"] == 10)
        & (decision_timing_audit_df["Decision Lead (min)"] == 15)
        & ~decision_timing_audit_df[
            "Both Inbound Arrivals Before Deadline"
        ],
        [
            "Tail A",
            "Tail B",
            "First Block Departure",
            "Decision Deadline",
            "Latest Estimated Inbound Arrival",
            "Modelled Reduction"
        ]
    ]
)

print("\nSTEP 3X COMPLETE")
```

```python

# ================================================================
# BUSI6029 — ASSESSMENT 2
# STEP 3Y: DECISION-TIME-CONSTRAINED RECOVERY OPTIMISATION
# ================================================================
# Purpose:
# 1. Screen ALL 47 candidate exchanges by decision deadline.
# 2. Combine decision lead times with scheduling safety margins.
# 3. Re-solve the binary ILP for every scenario.
# 4. Compare benefits with the common 1,816-minute baseline.
#
# IMPORTANT:
# Estimated historical inbound arrivals are used as a
# necessary information-timing screen only.
# They are NOT verified information-availability timestamps.
# Benefits are retrospective shortfall-proxy improvements.
# ================================================================

# ------------------------------------------------
# 3Y.1 — Calculate decision timing for ALL candidates
# ------------------------------------------------

decision_candidate_records = []

for candidate_id, swap in ilp_candidates.iterrows():

    a = eligible_blocks_corrected.iloc[
        int(swap["Block A Index"])
    ]

    b = eligible_blocks_corrected.iloc[
        int(swap["Block B Index"])
    ]

    first_departure = min(
        float(a["BLOCK_START_MIN"]),
        float(b["BLOCK_START_MIN"])
    )

    latest_inbound = max(
        float(a["INCOMING_ARRIVAL_MIN"]),
        float(b["INCOMING_ARRIVAL_MIN"])
    )

    decision_candidate_records.append({
        "Candidate ID": int(candidate_id),
        "First Departure": first_departure,
        "Latest Estimated Inbound": latest_inbound,
        "Available Decision Lead (min)": (
            first_departure - latest_inbound
        )
    })

decision_candidate_df = pd.DataFrame(
    decision_candidate_records
)

candidate_margin_lookup = (
    candidate_margin_df
    .set_index("Candidate ID")["Minimum Margin"]
)

decision_candidate_df["Minimum Safety Margin"] = (
    decision_candidate_df["Candidate ID"]
    .map(candidate_margin_lookup)
)

assert decision_candidate_df[
    "Minimum Safety Margin"
].notna().all()

# ------------------------------------------------
# 3Y.2 — Define joint sensitivity scenarios
# ------------------------------------------------

decision_leads = [0, 15, 30]
safety_margins = [0, 5, 10]

joint_results = []
joint_selected_ids = {}

print("=" * 65)
print("STEP 3Y — DECISION-TIME-CONSTRAINED OPTIMISATION")
print("=" * 65)

# ------------------------------------------------
# 3Y.3 — Filter and re-optimise every scenario
# ------------------------------------------------

for safety_margin in safety_margins:

    for decision_lead in decision_leads:

        qualifying = decision_candidate_df.loc[
            (
                decision_candidate_df["Minimum Safety Margin"]
                >= safety_margin - 1e-8
            )
            &
            (
                decision_candidate_df["Available Decision Lead (min)"]
                >= decision_lead - 1e-8
            )
        ]

        candidate_ids = (
            qualifying["Candidate ID"]
            .astype(int)
            .to_numpy()
        )

        if len(candidate_ids) == 0:

            chosen_ids = np.array([], dtype=int)
            reduction = 0.0

        else:

            scenario_result = milp(
                c=objective[candidate_ids],
                integrality=np.ones(
                    len(candidate_ids), dtype=int
                ),
                bounds=Bounds(
                    np.zeros(len(candidate_ids)),
                    np.ones(len(candidate_ids))
                ),
                constraints=LinearConstraint(
                    constraint_matrix.tocsr()[:, candidate_ids],
                    lb=np.zeros(n_aircraft),
                    ub=np.ones(n_aircraft)
                ),
                options={
                    "time_limit": 60,
                    "mip_rel_gap": 0.0
                }
            )

            if scenario_result.status != 0:
                raise RuntimeError(
                    f"Optimality not confirmed: "
                    f"safety={safety_margin}, "
                    f"lead={decision_lead}. "
                    f"{scenario_result.message}"
                )

            chosen_ids = candidate_ids[
                scenario_result.x > 0.5
            ]

            reduction = float(
                benefits[chosen_ids].sum()
            )

        joint_selected_ids[
            (safety_margin, decision_lead)
        ] = chosen_ids

        joint_results.append({
            "Extra Safety Margin": safety_margin,
            "Decision Lead": decision_lead,
            "Eligible Candidate Pairs": len(candidate_ids),
            "Optimal Swaps": len(chosen_ids),
            "Reduction (min)": round(reduction, 2),
            "Remaining Shortfall (min)": round(
                baseline_shortfall - reduction, 2
            ),
            "Full Baseline Improvement %": round(
                100 * reduction / baseline_shortfall, 2
            )
        })

# ------------------------------------------------
# 3Y.4 — Display joint sensitivity results
# ------------------------------------------------

decision_robustness_df = pd.DataFrame(joint_results)

print("\nJoint safety-margin and decision-time sensitivity:")

display(decision_robustness_df)

print("\nImprovement percentage comparison:")

display(
    decision_robustness_df.pivot(
        index="Extra Safety Margin",
        columns="Decision Lead",
        values="Full Baseline Improvement %"
    )
)

# ------------------------------------------------
# 3Y.5 — Validate optimisation against prior results
# ------------------------------------------------

# Decision-time filtering cannot improve an optimum
# relative to the same safety scenario without it.

for safety_margin in safety_margins:

    original_benefit = float(
        safety_margin_sensitivity_df.loc[
            safety_margin_sensitivity_df["Extra Margin (min)"]
            == safety_margin,
            "Reduction (min)"
        ].iloc[0]
    )

    scenario_rows = decision_robustness_df.loc[
        decision_robustness_df["Extra Safety Margin"]
        == safety_margin
    ].sort_values("Decision Lead")

    assert (
        scenario_rows["Reduction (min)"]
        <= original_benefit + 1e-8
    ).all()

    assert (
        scenario_rows["Reduction (min)"]
        .diff().dropna()
        <= 1e-8
    ).all()

print("\nSafety-baseline upper-bound checks: PASSED")
print("Decision-lead monotonicity checks: PASSED")

# ------------------------------------------------
# 3Y.6 — Inspect 10-minute safety, 15-minute lead
# ------------------------------------------------

focus_key = (10, 15)
focus_ids = joint_selected_ids[focus_key]

print("\nSelected exchanges: 10-minute safety margin,")
print("15-minute decision lead:")

if len(focus_ids):

    focus_table = ilp_candidates.iloc[focus_ids][
        [
            "Tail A",
            "Route A",
            "Tail B",
            "Route B",
            "Shortfall Reduction"
        ]
    ].copy()

    focus_table["Minimum Safety Margin"] = (
        decision_candidate_df.set_index("Candidate ID")
        .loc[focus_ids, "Minimum Safety Margin"]
        .to_numpy()
    )

    focus_table["Available Decision Lead"] = (
        decision_candidate_df.set_index("Candidate ID")
        .loc[focus_ids, "Available Decision Lead (min)"]
        .to_numpy()
    )

    display(
        focus_table.sort_values(
            "Shortfall Reduction",
            ascending=False
        )
    )

else:
    print("No exchanges selected.")

print("\nSTEP 3Y COMPLETE")
```

```python

# ================================================================
# BUSI6029 — ASSESSMENT 2
# STEP 3Z: INBOUND FLIGHT PROVENANCE AUDIT
# ================================================================
# Purpose:
# 1. Examine inbound histories for the 7-swap scenario.
# 2. Check preceding-flight location and date fields.
# 3. Assess whether preceding legs exist in the audit sample.
# 4. Identify limitations for real-time delay forecasting.
#
# No optimisation or predictive model is retrained.
# Historical PREV_ARR_DELAY is NOT a prospective forecast.
# ================================================================

# ------------------------------------------------
# 3Z.1 — Retrieve the conservative selected blocks
# ------------------------------------------------

focus_ids = joint_selected_ids[(10, 15)]
focus_swaps = ilp_candidates.iloc[focus_ids].copy()

focus_block_ids = sorted(set(
    focus_swaps["Block A Index"].astype(int).tolist()
    + focus_swaps["Block B Index"].astype(int).tolist()
))

inbound_audit = (
    eligible_blocks_corrected.iloc[focus_block_ids]
    .copy()
)

# ------------------------------------------------
# 3Z.2 — Audit preceding-leg information
# ------------------------------------------------

inbound_audit["PREVIOUS_DATE_PARSED"] = pd.to_datetime(
    inbound_audit["PREV_DATE"], errors="coerce"
).dt.normalize()

inbound_audit["CURRENT_DATE_PARSED"] = pd.to_datetime(
    inbound_audit["FL_DATE"], errors="coerce"
).dt.normalize()

inbound_audit["PREVIOUS_SAME_DAY"] = (
    inbound_audit["PREVIOUS_DATE_PARSED"]
    == inbound_audit["CURRENT_DATE_PARSED"]
)

inbound_audit["PREVIOUS_LOCATION_ATL"] = (
    inbound_audit["PREV_DEST"] == "ATL"
)

inbound_audit["INBOUND_DELAY_AVAILABLE"] = (
    inbound_audit["PREV_ARR_DELAY"].notna()
)

inbound_audit["GROUND_INTERVAL_AVAILABLE"] = (
    inbound_audit["SCHED_GROUND_MIN"].notna()
)

# ------------------------------------------------
# 3Z.3 — Check for a recorded previous leg
# ------------------------------------------------
# This only checks the 3-day filtered rotation sample.
# An unmatched leg may exist in the full original schedule.

rotation_lookup = (
    rotation_audit[
        ["FL_DATE", "TAIL_NUM", "DEST"]
    ]
    .copy()
)

rotation_lookup["FL_DATE"] = pd.to_datetime(
    rotation_lookup["FL_DATE"], errors="coerce"
).dt.normalize()

previous_leg_keys = set(
    zip(
        rotation_lookup["FL_DATE"],
        rotation_lookup["TAIL_NUM"],
        rotation_lookup["DEST"]
    )
)

inbound_audit["PREVIOUS_LEG_IN_AUDIT_SAMPLE"] = [
    (
        date, tail, previous_dest
    ) in previous_leg_keys
    if pd.notna(date) else False
    for date, tail, previous_dest in zip(
        inbound_audit["PREVIOUS_DATE_PARSED"],
        inbound_audit["TAIL_NUM"],
        inbound_audit["PREV_DEST"]
    )
]

# ------------------------------------------------
# 3Z.4 — Summarise provenance coverage
# ------------------------------------------------

print("=" * 65)
print("STEP 3Z — INBOUND FLIGHT PROVENANCE AUDIT")
print("=" * 65)

print("\nSelected swaps:", len(focus_swaps))
print("Aircraft / original blocks inspected:", len(inbound_audit))

checks = {
    "Preceding date recorded":
        inbound_audit["PREVIOUS_DATE_PARSED"].notna().sum(),
    "Preceding flight is on same date":
        inbound_audit["PREVIOUS_SAME_DAY"].sum(),
    "Preceding destination is ATL":
        inbound_audit["PREVIOUS_LOCATION_ATL"].sum(),
    "Historical inbound delay recorded":
        inbound_audit["INBOUND_DELAY_AVAILABLE"].sum(),
    "Scheduled ground interval recorded":
        inbound_audit["GROUND_INTERVAL_AVAILABLE"].sum(),
    "Potential preceding leg found in 3-day sample":
        inbound_audit["PREVIOUS_LEG_IN_AUDIT_SAMPLE"].sum()
}

display(
    pd.DataFrame(
        checks.items(),
        columns=["Provenance Check", "Aircraft / Blocks"]
    )
)

# ------------------------------------------------
# 3Z.5 — Inspect individual inbound histories
# ------------------------------------------------

print("\nSelected aircraft inbound provenance:")

display(
    inbound_audit[
        [
            "TAIL_NUM",
            "DEST",
            "PREV_DATE",
            "PREV_DEST",
            "PREV_ARR_DELAY",
            "SCHED_GROUND_MIN",
            "INCOMING_ARRIVAL_MIN",
            "PREVIOUS_SAME_DAY",
            "PREVIOUS_LEG_IN_AUDIT_SAMPLE"
        ]
    ].sort_values("TAIL_NUM")
)

# ------------------------------------------------
# 3Z.6 — Display the existing predictive model inputs
# ------------------------------------------------

print("\nExisting Random Forest input features:")
print(candidate_features)

print("\nSTEP 3Z COMPLETE")
```

```python

# ================================================================
# BUSI6029 — ASSESSMENT 2
# STEP 3AA: EXACT INCOMING FLIGHT LINKAGE VALIDATION
# ================================================================
# Purpose:
# 1. Identify the immediately preceding recorded flight
#    for each selected aircraft.
# 2. Compare preceding arrival airport, date and delay.
# 3. Verify actual flight-to-flight sequence consistency.
# 4. Prepare reliable incoming-flight references for
#    predictive-prescriptive integration.
#
# ARR_DELAY_NEW is non-negative, so negative historical
# arrival delays cannot be verified exactly here.
# No model training or optimisation in this step.
# ================================================================

# ------------------------------------------------
# 3AA.1 — Reconstruct preceding flight references
# ------------------------------------------------

preceding_sequence = (
    rotation_day
    .sort_values(["TAIL_NUM", "SOURCE_ORDER"])
    .copy()
)

tail_sequence = preceding_sequence.groupby(
    "TAIL_NUM", sort=False
)

previous_fields = {
    "FL_DATE": "LINK_PREV_DATE",
    "OP_CARRIER_FL_NUM": "LINK_PREV_FLIGHT",
    "ORIGIN": "LINK_PREV_ORIGIN",
    "DEST": "LINK_PREV_DEST",
    "CRS_DEP_MINUTES": "LINK_PREV_DEP_MIN",
    "CRS_ARR_MINUTES": "LINK_PREV_ARR_MIN",
    "ARR_DELAY_NEW": "LINK_PREV_DELAY_NEW"
}

for original_col, linked_col in previous_fields.items():
    preceding_sequence[linked_col] = (
        tail_sequence[original_col].shift(1)
    )

# ------------------------------------------------
# 3AA.2 — Match the 14 selected blocks by SOURCE_ORDER
# ------------------------------------------------

previous_lookup = preceding_sequence.set_index(
    "SOURCE_ORDER"
)[list(previous_fields.values())]

exact_inbound_audit = inbound_audit.copy()

for linked_col in previous_fields.values():
    exact_inbound_audit[linked_col] = (
        exact_inbound_audit["SOURCE_ORDER"]
        .map(previous_lookup[linked_col])
    )

# ------------------------------------------------
# 3AA.3 — Verify location, date and arrival delay
# ------------------------------------------------

exact_inbound_audit["LINK_EXISTS"] = (
    exact_inbound_audit["LINK_PREV_FLIGHT"].notna()
)

exact_inbound_audit["LINK_DEST_MATCH"] = (
    exact_inbound_audit["LINK_PREV_DEST"]
    == exact_inbound_audit["PREV_DEST"]
)

exact_inbound_audit["LINK_DATE_MATCH"] = (
    pd.to_datetime(
        exact_inbound_audit["LINK_PREV_DATE"],
        errors="coerce"
    ).dt.normalize()
    ==
    pd.to_datetime(
        exact_inbound_audit["PREV_DATE"],
        errors="coerce"
    ).dt.normalize()
)

exact_inbound_audit["EXPECTED_PREV_DELAY_NEW"] = (
    exact_inbound_audit["PREV_ARR_DELAY"]
    .clip(lower=0)
)

exact_inbound_audit["LINK_DELAY_MATCH"] = np.isclose(
    exact_inbound_audit["LINK_PREV_DELAY_NEW"],
    exact_inbound_audit["EXPECTED_PREV_DELAY_NEW"],
    atol=1e-6,
    equal_nan=False
)

exact_inbound_audit["ALL_LINK_CHECKS_PASS"] = (
    exact_inbound_audit["LINK_EXISTS"]
    & exact_inbound_audit["LINK_DEST_MATCH"]
    & exact_inbound_audit["LINK_DATE_MATCH"]
    & exact_inbound_audit["LINK_DELAY_MATCH"]
)

# ------------------------------------------------
# 3AA.4 — Summarise exact record linkage
# ------------------------------------------------

print("=" * 65)
print("STEP 3AA — EXACT INCOMING FLIGHT LINKAGE")
print("=" * 65)

linkage_summary = pd.DataFrame({
    "Check": [
        "Selected aircraft blocks",
        "Preceding record exists",
        "Preceding destination matches",
        "Preceding flight date matches",
        "Non-negative arrival delay matches",
        "All linkage checks pass"
    ],
    "Count": [
        len(exact_inbound_audit),
        int(exact_inbound_audit["LINK_EXISTS"].sum()),
        int(exact_inbound_audit["LINK_DEST_MATCH"].sum()),
        int(exact_inbound_audit["LINK_DATE_MATCH"].sum()),
        int(exact_inbound_audit["LINK_DELAY_MATCH"].sum()),
        int(exact_inbound_audit["ALL_LINK_CHECKS_PASS"].sum())
    ]
})

print("\nIncoming-flight linkage summary:")
display(linkage_summary)

# ------------------------------------------------
# 3AA.5 — Inspect exact preceding flights
# ------------------------------------------------

print("\nDetailed incoming-flight linkage:")

display(
    exact_inbound_audit[
        [
            "TAIL_NUM",
            "DEST",
            "LINK_PREV_FLIGHT",
            "LINK_PREV_ORIGIN",
            "LINK_PREV_DEST",
            "LINK_PREV_ARR_MIN",
            "PREV_ARR_DELAY",
            "LINK_PREV_DELAY_NEW",
            "LINK_DATE_MATCH",
            "LINK_DELAY_MATCH",
            "ALL_LINK_CHECKS_PASS"
        ]
    ].sort_values("TAIL_NUM")
)

# ------------------------------------------------
# 3AA.6 — Show any mismatches
# ------------------------------------------------

print("\nRecords requiring further investigation:")

display(
    exact_inbound_audit.loc[
        ~exact_inbound_audit["ALL_LINK_CHECKS_PASS"],
        [
            "TAIL_NUM",
            "SOURCE_ORDER",
            "LINK_PREV_FLIGHT",
            "PREV_DATE",
            "LINK_PREV_DATE",
            "PREV_DEST",
            "LINK_PREV_DEST",
            "PREV_ARR_DELAY",
            "LINK_PREV_DELAY_NEW"
        ]
    ]
)

print("\nSTEP 3AA COMPLETE")
```

```python

# ================================================================
# BUSI6029 — ASSESSMENT 2
# STEP 3AB: UNRESOLVED INCOMING FLIGHT COVERAGE AUDIT
# ================================================================
# Purpose:
# 1. Investigate the six unresolved preceding-flight links.
# 2. Search for candidate ATL arrivals in the existing sample.
# 3. Compare aircraft, date, scheduled arrival and delay.
# 4. Distinguish filtered-data gaps from mismatched records.
#
# A candidate match is NOT automatically an exact flight link.
# No predictive retraining or optimisation is performed.
# ================================================================

# ------------------------------------------------
# 3AB.1 — Isolate unresolved aircraft
# ------------------------------------------------

unresolved_inbounds = exact_inbound_audit.loc[
    ~exact_inbound_audit["ALL_LINK_CHECKS_PASS"]
].copy()

print("=" * 65)
print("STEP 3AB — UNRESOLVED INCOMING FLIGHT AUDIT")
print("=" * 65)

print("\nUnresolved aircraft:", len(unresolved_inbounds))

# ------------------------------------------------
# 3AB.2 — Build candidate incoming-ATL flight lookup
# ------------------------------------------------

atl_arrivals_sample = rotation_audit.loc[
    rotation_audit["DEST"] == "ATL"
].copy()

atl_arrivals_sample["FL_DATE"] = pd.to_datetime(
    atl_arrivals_sample["FL_DATE"]
).dt.normalize()

audit_records = []
possible_matches = []

# ------------------------------------------------
# 3AB.3 — Check each unresolved aircraft separately
# ------------------------------------------------

for _, target in unresolved_inbounds.iterrows():

    tail = target["TAIL_NUM"]

    previous_date = pd.Timestamp(
        target["PREV_DATE"]
    ).normalize()

    # Reconstruct previous SCHEDULED arrival at ATL.
    # All times in this comparison refer to ATL local time.
    expected_sched_arrival = (
        float(target["BLOCK_START_MIN"])
        - float(target["SCHED_GROUND_MIN"])
    )

    expected_delay_new = max(
        float(target["PREV_ARR_DELAY"]), 0.0
    )

    candidates = atl_arrivals_sample.loc[
        (atl_arrivals_sample["TAIL_NUM"] == tail)
        & (atl_arrivals_sample["FL_DATE"] == previous_date)
    ].copy()

    candidates["SCHEDULE_TIME_MATCH"] = np.isclose(
        candidates["CRS_ARR_MINUTES"],
        expected_sched_arrival,
        atol=1.0
    )

    candidates["DELAY_MATCH"] = np.isclose(
        candidates["ARR_DELAY_NEW"],
        expected_delay_new,
        atol=1e-6
    )

    candidates["BOTH_MATCH"] = (
        candidates["SCHEDULE_TIME_MATCH"]
        & candidates["DELAY_MATCH"]
    )

    audit_records.append({
        "Aircraft": tail,
        "Target Route": f"ATL-{target['DEST']}-ATL",
        "Target Source Order": int(target["SOURCE_ORDER"]),
        "Expected Previous ATL Arrival": expected_sched_arrival,
        "Expected Previous Delay New": expected_delay_new,
        "ATL Arrivals in Sample": len(candidates),
        "Scheduled-Time Matches": int(
            candidates["SCHEDULE_TIME_MATCH"].sum()
        ),
        "Time-and-Delay Matches": int(
            candidates["BOTH_MATCH"].sum()
        )
    })

    for _, flight in candidates.iterrows():

        possible_matches.append({
            "Aircraft": tail,
            "Candidate Flight": flight["OP_CARRIER_FL_NUM"],
            "Incoming Origin": flight["ORIGIN"],
            "Scheduled ATL Arrival": flight["CRS_ARR_MINUTES"],
            "Observed Delay New": flight["ARR_DELAY_NEW"],
            "Expected ATL Arrival": expected_sched_arrival,
            "Time Match": bool(flight["SCHEDULE_TIME_MATCH"]),
            "Delay Match": bool(flight["DELAY_MATCH"]),
            "Both Match": bool(flight["BOTH_MATCH"])
        })

# ------------------------------------------------
# 3AB.4 — Display coverage summary
# ------------------------------------------------

unresolved_coverage_df = pd.DataFrame(audit_records)

print("\nUnresolved incoming-flight coverage:")
display(unresolved_coverage_df)

print("\nCandidate preceding ATL arrivals:")

candidate_inbound_df = pd.DataFrame(
    possible_matches,
    columns=[
        "Aircraft",
        "Candidate Flight",
        "Incoming Origin",
        "Scheduled ATL Arrival",
        "Observed Delay New",
        "Expected ATL Arrival",
        "Time Match",
        "Delay Match",
        "Both Match"
    ]
)

display(candidate_inbound_df)

print("\nSTEP 3AB COMPLETE")
```

```python

# ================================================================
# BUSI6029 — ASSESSMENT 2
# STEP 3AC: SOURCE DATASET INVENTORY
# ================================================================
# Purpose:
# 1. Identify available CSV, ZIP and Parquet sources.
# 2. Inspect CSV column names without loading full datasets.
# 3. Locate possible less-filtered flight records.
# 4. Avoid making assumptions about missing inbound flights.
#
# No datasets, models or results are modified.
# ================================================================

# ------------------------------------------------
# 3AC.1 — Inventory likely data files
# ------------------------------------------------

source_files = sorted(
    [
        p for p in DATA_DIR.rglob("*")
        if p.is_file()
        and p.suffix.lower() in {
            ".csv", ".zip", ".parquet", ".xlsx"
        }
    ],
    key=lambda p: str(p).lower()
)

inventory_records = []

for path in source_files:

    record = {
        "Filename": path.name,
        "Type": path.suffix.lower(),
        "Size (MB)": round(
            path.stat().st_size / (1024 ** 2), 2
        ),
        "Parent Folder": str(
            path.parent.relative_to(DATA_DIR)
        )
    }

    if path.suffix.lower() == ".csv":

        try:
            header = pd.read_csv(
                path, nrows=0, low_memory=False
            ).columns.tolist()

            record["Column Count"] = len(header)

            record["Has Flight Date"] = (
                "FL_DATE" in header
            )

            record["Has Aircraft Tail"] = (
                "TAIL_NUM" in header
            )

            record["Has Arrival Delay"] = (
                "ARR_DELAY" in header
                or "ARR_DELAY_NEW" in header
            )

            record["Has ATL Arrival Fields"] = all(
                col in header
                for col in ["DEST", "CRS_ARR_TIME"]
            )

        except Exception as exc:
            record["Header Error"] = str(exc)[:120]

    inventory_records.append(record)

source_inventory_df = pd.DataFrame(
    inventory_records
)

# ------------------------------------------------
# 3AC.2 — Display results
# ------------------------------------------------

print("=" * 65)
print("STEP 3AC — SOURCE DATASET INVENTORY")
print("=" * 65)

print("\nProject directory:", DATA_DIR)
print("Potential source files found:", len(source_files))

display(source_inventory_df)

print("\nSTEP 3AC COMPLETE")
```

```python

# ================================================================
# BUSI6029 — ASSESSMENT 2
# STEP 3AD: RECOVER MISSING INBOUND FLIGHTS FROM RAW DATA
# ================================================================
# Purpose:
# 1. Search raw flight records for 6 unresolved predecessors.
# 2. Match aircraft, date, ATL arrival time and signed delay.
# 3. Detect unique matches versus ambiguous/missing records.
# 4. Preserve existing predictive and optimisation results.
#
# Reads only required columns and processes the CSV in chunks.
# ================================================================

# ------------------------------------------------
# 3AD.1 — Locate and inspect the raw source
# ------------------------------------------------

raw_file = next(
    p for p in source_files
    if p.name == "Delta_2023_2025_Raw.csv"
)

raw_columns = pd.read_csv(raw_file, nrows=0).columns.tolist()

required = [
    "FL_DATE", "TAIL_NUM", "OP_CARRIER_FL_NUM",
    "ORIGIN", "DEST", "CRS_ARR_TIME",
    "ARR_DELAY"
]

missing = [c for c in required if c not in raw_columns]
assert not missing, f"Raw columns missing: {missing}"

# ------------------------------------------------
# 3AD.2 — Build expected incoming flight references
# ------------------------------------------------

targets = unresolved_inbounds.copy()

targets["EXPECTED_ARR_MIN"] = (
    targets["BLOCK_START_MIN"]
    - targets["SCHED_GROUND_MIN"]
)

targets["TARGET_PREV_DATE"] = pd.to_datetime(
    targets["PREV_DATE"]
).dt.normalize()

target_tails = set(targets["TAIL_NUM"])
target_dates = set(targets["TARGET_PREV_DATE"])

# ------------------------------------------------
# 3AD.3 — Extract relevant raw records in chunks
# ------------------------------------------------

raw_matches = []
rows_scanned = 0

for chunk in pd.read_csv(
    raw_file,
    usecols=required,
    chunksize=150_000,
    low_memory=False
):
    rows_scanned += len(chunk)

    chunk["FL_DATE"] = pd.to_datetime(
        chunk["FL_DATE"], errors="coerce"
    ).dt.normalize()

    relevant = chunk.loc[
        chunk["TAIL_NUM"].isin(target_tails)
        & chunk["FL_DATE"].isin(target_dates)
        & (chunk["DEST"] == "ATL")
    ].copy()

    if not relevant.empty:
        raw_matches.append(relevant)

raw_inbound_candidates = (
    pd.concat(raw_matches, ignore_index=True)
    if raw_matches
    else pd.DataFrame(columns=required)
)

# ------------------------------------------------
# 3AD.4 — Convert HHMM scheduled arrival to minutes
# ------------------------------------------------

arr_hhmm = pd.to_numeric(
    raw_inbound_candidates["CRS_ARR_TIME"],
    errors="coerce"
)

raw_inbound_candidates["RAW_ARR_MIN"] = (
    np.floor(arr_hhmm / 100) * 60
    + (arr_hhmm % 100)
)

# ------------------------------------------------
# 3AD.5 — Test exact predecessor matches
# ------------------------------------------------

recovered_records = []
recovery_summary = []

for _, target in targets.iterrows():

    candidates = raw_inbound_candidates.loc[
        (
            raw_inbound_candidates["TAIL_NUM"]
            == target["TAIL_NUM"]
        )
        &
        (
            raw_inbound_candidates["FL_DATE"]
            == target["TARGET_PREV_DATE"]
        )
    ].copy()

    candidates["TIME_MATCH"] = np.isclose(
        candidates["RAW_ARR_MIN"],
        target["EXPECTED_ARR_MIN"],
        atol=1.0
    )

    candidates["SIGNED_DELAY_MATCH"] = np.isclose(
        pd.to_numeric(
            candidates["ARR_DELAY"], errors="coerce"
        ),
        float(target["PREV_ARR_DELAY"]),
        atol=1.0
    )

    candidates["EXACT_MATCH"] = (
        candidates["TIME_MATCH"]
        & candidates["SIGNED_DELAY_MATCH"]
    )

    matched = candidates.loc[candidates["EXACT_MATCH"]]

    recovery_summary.append({
        "Aircraft": target["TAIL_NUM"],
        "Expected Arrival Minute": target["EXPECTED_ARR_MIN"],
        "Expected Signed Delay": target["PREV_ARR_DELAY"],
        "Raw ATL Arrivals": len(candidates),
        "Time Matches": int(candidates["TIME_MATCH"].sum()),
        "Exact Matches": len(matched),
        "Match Status": (
            "UNIQUE" if len(matched) == 1
            else "MISSING" if len(matched) == 0
            else "AMBIGUOUS"
        )
    })

    for _, flight in matched.iterrows():
        recovered_records.append({
            "Aircraft": target["TAIL_NUM"],
            "Incoming Flight": flight["OP_CARRIER_FL_NUM"],
            "Incoming Origin": flight["ORIGIN"],
            "Incoming Destination": flight["DEST"],
            "Flight Date": flight["FL_DATE"],
            "Scheduled ATL Arrival": flight["CRS_ARR_TIME"],
            "Arrival Delay": flight["ARR_DELAY"]
        })

raw_recovery_summary_df = pd.DataFrame(recovery_summary)
recovered_inbound_df = pd.DataFrame(recovered_records)

# ------------------------------------------------
# 3AD.6 — Display results
# ------------------------------------------------

print("=" * 65)
print("STEP 3AD — RAW PREDECESSOR FLIGHT RECOVERY")
print("=" * 65)

print("\nRaw records scanned:", f"{rows_scanned:,}")
print("Relevant ATL arrivals found:",
      len(raw_inbound_candidates))

print("\nMatch summary:")
display(raw_recovery_summary_df)

print("\nRecovered exact matching flights:")
display(recovered_inbound_df)

print(
    "\nUnique incoming flights recovered:",
    int(
        (raw_recovery_summary_df["Match Status"]
         == "UNIQUE").sum()
    ),
    "of",
    len(raw_recovery_summary_df)
)

print("\nSTEP 3AD COMPLETE")
```

```python

# ================================================================
# BUSI6029 — ASSESSMENT 2
# STEP 3AE: INCOMING FLIGHT FEATURE AVAILABILITY AUDIT
# ================================================================
# Purpose:
# 1. Locate all 14 verified incoming flights in engineered data.
# 2. Check availability of the 11 RF input features.
# 3. Identify missing information or ambiguous matches.
# 4. Assess readiness for forecast-assisted recovery.
#
# No predictive retraining, forecasting or ILP changes.
# ================================================================

# ------------------------------------------------
# 3AE.1 — Locate engineered source and columns
# ------------------------------------------------

engineered_file = next(
    p for p in source_files
    if p.name == "Delta_2023_2025_Engineered.csv"
)

engineered_columns = pd.read_csv(
    engineered_file, nrows=0
).columns.tolist()

match_columns = [
    "FL_DATE", "TAIL_NUM", "DEST",
    "CRS_ARR_TIME", "ARR_DELAY"
]

available_features = [
    c for c in candidate_features
    if c in engineered_columns
]

missing_features = [
    c for c in candidate_features
    if c not in engineered_columns
]

print("=" * 65)
print("STEP 3AE — INCOMING FEATURE AVAILABILITY")
print("=" * 65)

print("\nRF features present:", len(available_features),
      "of", len(candidate_features))
print("Missing RF columns:", missing_features)

missing_match = [
    c for c in match_columns
    if c not in engineered_columns
]

assert not missing_match, (
    f"Engineered matching columns absent: {missing_match}"
)

# ------------------------------------------------
# 3AE.2 — Build expected references for all 14 flights
# ------------------------------------------------

inbound_targets = inbound_audit.copy()

inbound_targets["EXPECTED_ARR_MIN"] = (
    inbound_targets["BLOCK_START_MIN"]
    - inbound_targets["SCHED_GROUND_MIN"]
)

inbound_targets["TARGET_DATE"] = pd.to_datetime(
    inbound_targets["PREV_DATE"],
    errors="coerce"
).dt.normalize()

focus_tails = set(inbound_targets["TAIL_NUM"])

# ------------------------------------------------
# 3AE.3 — Read relevant engineered records in chunks
# ------------------------------------------------

columns_to_read = list(dict.fromkeys(
    match_columns + available_features
))

engineered_parts = []

for chunk in pd.read_csv(
    engineered_file,
    usecols=columns_to_read,
    chunksize=150_000,
    low_memory=False
):
    # Parse only relevant date rows, avoiding the
    # repeated auto-format warning from Step 3AD.
    relevant = chunk.loc[
        chunk["TAIL_NUM"].isin(focus_tails)
        & (chunk["DEST"] == "ATL")
    ].copy()

    if relevant.empty:
        continue

    relevant["FL_DATE"] = pd.to_datetime(
        relevant["FL_DATE"],
        format="mixed",
        errors="coerce"
    ).dt.normalize()

    relevant = relevant.loc[
        relevant["FL_DATE"].eq(
            pd.Timestamp("2024-01-16")
        )
    ]

    if not relevant.empty:
        engineered_parts.append(relevant)

engineered_inbound_pool = (
    pd.concat(engineered_parts, ignore_index=True)
    if engineered_parts
    else pd.DataFrame(columns=columns_to_read)
)

# ------------------------------------------------
# 3AE.4 — Match by ATL arrival schedule and delay
# ------------------------------------------------

hhmm = pd.to_numeric(
    engineered_inbound_pool["CRS_ARR_TIME"],
    errors="coerce"
)

engineered_inbound_pool["ARR_MIN"] = (
    np.floor(hhmm / 100) * 60
    + hhmm % 100
)

feature_match_records = []
matched_inbound_rows = []

for _, target in inbound_targets.iterrows():

    possibilities = engineered_inbound_pool.loc[
        (engineered_inbound_pool["TAIL_NUM"]
         == target["TAIL_NUM"])
        & (engineered_inbound_pool["FL_DATE"]
           == target["TARGET_DATE"])
    ].copy()

    matches = possibilities.loc[
        np.isclose(
            possibilities["ARR_MIN"],
            target["EXPECTED_ARR_MIN"],
            atol=1
        )
        & np.isclose(
            pd.to_numeric(
                possibilities["ARR_DELAY"],
                errors="coerce"
            ),
            target["PREV_ARR_DELAY"],
            atol=1
        )
    ]

    feature_match_records.append({
        "Aircraft": target["TAIL_NUM"],
        "Exact Engineered Matches": len(matches),
        "Feature Columns Present": len(available_features),
        "Complete Feature Rows": int(
            matches[available_features]
            .notna().all(axis=1).sum()
        ) if available_features else 0
    })

    if len(matches) == 1:
        matched_inbound_rows.append(matches.iloc[0])

engineered_match_summary_df = pd.DataFrame(
    feature_match_records
)

engineered_inbound_matched_df = (
    pd.DataFrame(matched_inbound_rows)
    .reset_index(drop=True)
)

# ------------------------------------------------
# 3AE.5 — Report feature coverage
# ------------------------------------------------

print("\nIncoming-flight engineered record coverage:")
display(engineered_match_summary_df)

print(
    "\nAircraft with exactly one engineered match:",
    int(
        (engineered_match_summary_df[
            "Exact Engineered Matches"
        ] == 1).sum()
    ),
    "of",
    len(engineered_match_summary_df)
)

if not engineered_inbound_matched_df.empty:

    print("\nMissing RF feature values across matched flights:")

    display(
        engineered_inbound_matched_df[
            available_features
        ].isna().sum()
        .rename("Missing Values")
        .to_frame()
    )

    print("\nIncoming-flight recovery feature examples:")

    display(
        engineered_inbound_matched_df[
            [
                c for c in [
                    "TAIL_NUM", "ORIGIN", "DEST",
                    "PREV_ARR_DELAY", "SCHED_GROUND_MIN"
                ]
                if c in engineered_inbound_matched_df.columns
            ]
        ]
    )

print("\nSTEP 3AE COMPLETE")
```

```python

# ================================================================
# BUSI6029 — ASSESSMENT 2
# STEP 3AF: CORRECTED FITTED MODEL INVENTORY
# ================================================================
# Purpose:
# 1. Identify existing trained scikit-learn models.
# 2. Exclude estimator classes and unrelated objects.
# 3. Safely inspect fitted status and feature requirements.
# 4. Locate the 9-feature schedule-only Random Forest.
#
# No model training, predictions or optimisation changes.
# ================================================================

from sklearn.base import BaseEstimator
from sklearn.utils.validation import check_is_fitted

print("=" * 65)
print("STEP 3AF — CORRECTED EXISTING MODEL INVENTORY")
print("=" * 65)

model_records = []

# ------------------------------------------------
# 3AF.1 — Search existing notebook variables
# ------------------------------------------------

for variable_name, obj in list(globals().items()):

    if variable_name.startswith("_"):
        continue

    # Search stored models inside dictionaries.
    if isinstance(obj, dict):

        entries = [
            (f"{variable_name}[{key!r}]", value)
            for key, value in obj.items()
            if isinstance(value, BaseEstimator)
            and hasattr(value, "predict")
        ]

    # Search individual estimator instances.
    elif isinstance(obj, BaseEstimator) and hasattr(obj, "predict"):

        entries = [(variable_name, obj)]

    else:
        continue

    # ------------------------------------------------
    # 3AF.2 — Inspect each estimator safely
    # ------------------------------------------------

    for model_name, model in entries:

        try:
            check_is_fitted(model)
            fitted = True
        except Exception:
            fitted = False

        try:
            n_features = getattr(
                model, "n_features_in_", None
            )
        except Exception:
            n_features = None

        try:
            feature_names = getattr(
                model, "feature_names_in_", None
            )

            if feature_names is not None:
                feature_names = list(feature_names)

        except (TypeError, AttributeError):
            feature_names = None

        model_records.append({
            "Model Variable": model_name,
            "Model Type": type(model).__name__,
            "Fitted": fitted,
            "Input Features": n_features,
            "Recorded Feature Names": (
                str(feature_names)[:250]
                if feature_names is not None
                else "Not directly available"
            )
        })

# ------------------------------------------------
# 3AF.3 — Display model inventory
# ------------------------------------------------

model_inventory_df = pd.DataFrame(
    model_records,
    columns=[
        "Model Variable",
        "Model Type",
        "Fitted",
        "Input Features",
        "Recorded Feature Names"
    ]
)

print("\nExisting scikit-learn prediction models:")

display(model_inventory_df)

# ------------------------------------------------
# 3AF.4 — Identify possible schedule-only models
# ------------------------------------------------

print("\nPossible 9-feature schedule-only models:")

schedule_model_candidates = model_inventory_df.loc[
    model_inventory_df["Fitted"]
    & (
        model_inventory_df["Input Features"] == len(schedule_features)
    )
]

display(schedule_model_candidates)

print("\nExpected schedule-only features:")
print(schedule_features)

print("\nSTEP 3AF COMPLETE")
```

```python

# ================================================================
# BUSI6029 — ASSESSMENT 2
# STEP 3AG: SCHEDULE-ONLY INCOMING AIRCRAFT DELAY FORECASTS
# ================================================================
# Purpose:
# 1. Use the EXISTING fitted schedule-only Random Forest.
# 2. Forecast arrival delays for 14 linked inbound flights.
# 3. Compare predictions with historical observed delays.
# 4. Establish inputs for a later forecast-assisted scenario.
#
# This is a retrospective 2024 diagnostic, NOT a new
# independent validation or causal delay-saving estimate.
# No models are trained or modified.
# ================================================================

# ------------------------------------------------
# 3AG.1 — Validate the fitted model and feature schema
# ------------------------------------------------

assert len(engineered_inbound_matched_df) == 14, (
    "Expected exactly 14 linked incoming flights."
)

assert list(schedule_rf_pipeline.feature_names_in_) == (
    schedule_features
), "Schedule-only model feature order does not match."

assert all(
    feature in engineered_inbound_matched_df.columns
    for feature in schedule_features
), "Required schedule features are missing."

# ------------------------------------------------
# 3AG.2 — Prepare the incoming flight feature matrix
# ------------------------------------------------

incoming_features = (
    engineered_inbound_matched_df[
        schedule_features
    ].copy()
)

assert incoming_features.notna().all().all(), (
    "Missing values found in incoming flight features."
)

# ------------------------------------------------
# 3AG.3 — Predict using the EXISTING schedule-only model
# ------------------------------------------------

incoming_predictions = (
    schedule_rf_pipeline.predict(incoming_features)
)

# Target is non-negative arrival delay.
incoming_predictions = np.maximum(
    np.asarray(incoming_predictions, dtype=float),
    0.0
)

# ------------------------------------------------
# 3AG.4 — Build the flight-level diagnostic table
# ------------------------------------------------

incoming_forecast_df = (
    engineered_inbound_matched_df[
        [
            "TAIL_NUM",
            "ORIGIN",
            "DEST",
            "FL_DATE",
            "CRS_ARR_TIME",
            "ARR_DELAY"
        ]
    ]
    .copy()
    .reset_index(drop=True)
)

incoming_forecast_df["Predicted Delay (min)"] = (
    incoming_predictions
)

incoming_forecast_df["Actual Delay New (min)"] = (
    pd.to_numeric(
        incoming_forecast_df["ARR_DELAY"],
        errors="coerce"
    ).clip(lower=0)
)

incoming_forecast_df["Absolute Error (min)"] = (
    incoming_forecast_df["Predicted Delay (min)"]
    - incoming_forecast_df["Actual Delay New (min)"]
).abs()

# ------------------------------------------------
# 3AG.5 — Calculate descriptive diagnostic metrics
# ------------------------------------------------

diagnostic_mae = (
    incoming_forecast_df[
        "Absolute Error (min)"
    ].mean()
)

diagnostic_rmse = np.sqrt(
    np.mean(
        (
            incoming_forecast_df["Predicted Delay (min)"]
            - incoming_forecast_df["Actual Delay New (min)"]
        ) ** 2
    )
)

# ------------------------------------------------
# 3AG.6 — Display results
# ------------------------------------------------

print("=" * 65)
print("STEP 3AG — INCOMING AIRCRAFT DELAY FORECASTS")
print("=" * 65)

print("\nIncoming flights:", len(incoming_forecast_df))

print(
    "Descriptive MAE (14 flights):",
    round(diagnostic_mae, 2),
    "minutes"
)

print(
    "Descriptive RMSE (14 flights):",
    round(diagnostic_rmse, 2),
    "minutes"
)

print("\nIncoming-flight forecast comparison:")

display(
    incoming_forecast_df[
        [
            "TAIL_NUM",
            "ORIGIN",
            "DEST",
            "ARR_DELAY",
            "Predicted Delay (min)",
            "Actual Delay New (min)",
            "Absolute Error (min)"
        ]
    ].sort_values(
        "Absolute Error (min)",
        ascending=False
    ).round(2)
)

print("\nSTEP 3AG COMPLETE")
```

```python

# ================================================================
# BUSI6029 — ASSESSMENT 2
# STEP 3AH: FORECAST ERROR AND AIRCRAFT READINESS RISK
# ================================================================
# Purpose:
# 1. Measure signed forecast errors for 14 incoming flights.
# 2. Quantify overoptimistic aircraft-readiness estimates.
# 3. Examine readiness assumptions for 7 selected exchanges.
# 4. Identify risks before forecast-assisted optimisation.
#
# Retrospective diagnostic only.
# NO model fitting, uncertainty calibration or new ILP yet.
# ================================================================

# ------------------------------------------------
# 3AH.1 — Join forecasts to their ATL outbound blocks
# ------------------------------------------------

forecast_readiness_df = inbound_audit[
    [
        "TAIL_NUM",
        "DEST",
        "BLOCK_START_MIN",
        "SCHED_GROUND_MIN",
        "PREV_ARR_DELAY",
        "INCOMING_ARRIVAL_MIN"
    ]
].copy()

forecast_readiness_df = forecast_readiness_df.merge(
    incoming_forecast_df[
        [
            "TAIL_NUM",
            "Predicted Delay (min)",
            "Actual Delay New (min)"
        ]
    ],
    on="TAIL_NUM",
    how="left",
    validate="one_to_one"
)

assert len(forecast_readiness_df) == 14
assert forecast_readiness_df[
    "Predicted Delay (min)"
].notna().all()

# ------------------------------------------------
# 3AH.2 — Compare signed historical delays with forecasts
# ------------------------------------------------
# The model predicts a NON-NEGATIVE delay target.
# For readiness, use the signed actual arrival delay
# to preserve historical early arrivals.

forecast_readiness_df["Signed Error (min)"] = (
    forecast_readiness_df["PREV_ARR_DELAY"]
    - forecast_readiness_df["Predicted Delay (min)"]
)

forecast_readiness_df["Underpredicted (min)"] = (
    forecast_readiness_df["Signed Error (min)"]
    .clip(lower=0)
)

forecast_readiness_df["Forecast Inbound Arrival"] = (
    forecast_readiness_df["BLOCK_START_MIN"]
    - forecast_readiness_df["SCHED_GROUND_MIN"]
    + forecast_readiness_df["Predicted Delay (min)"]
)

forecast_readiness_df["Actual Inbound Arrival"] = (
    forecast_readiness_df["INCOMING_ARRIVAL_MIN"]
)

forecast_readiness_df["Forecast Arrival Error"] = (
    forecast_readiness_df["Actual Inbound Arrival"]
    - forecast_readiness_df["Forecast Inbound Arrival"]
)

assert np.allclose(
    forecast_readiness_df["Signed Error (min)"],
    forecast_readiness_df["Forecast Arrival Error"]
)

# ------------------------------------------------
# 3AH.3 — Audit forecast versus actual swap-start margins
# ------------------------------------------------

readiness_lookup = forecast_readiness_df.set_index("TAIL_NUM")

focus_swap_risks = []

for _, swap in focus_swaps.iterrows():

    a = eligible_blocks_corrected.iloc[
        int(swap["Block A Index"])
    ]
    b = eligible_blocks_corrected.iloc[
        int(swap["Block B Index"])
    ]

    forecast_a = readiness_lookup.loc[a["TAIL_NUM"]]
    forecast_b = readiness_lookup.loc[b["TAIL_NUM"]]

    # Tail A would operate B's block, and vice versa.
    predicted_margins = [
        b["BLOCK_START_MIN"]
        - forecast_a["Forecast Inbound Arrival"]
        - min_turnaround,
        a["BLOCK_START_MIN"]
        - forecast_b["Forecast Inbound Arrival"]
        - min_turnaround
    ]

    actual_margins = [
        b["BLOCK_START_MIN"]
        - a["INCOMING_ARRIVAL_MIN"]
        - min_turnaround,
        a["BLOCK_START_MIN"]
        - b["INCOMING_ARRIVAL_MIN"]
        - min_turnaround
    ]

    focus_swap_risks.append({
        "Tail A": a["TAIL_NUM"],
        "Tail B": b["TAIL_NUM"],
        "Modelled Benefit (min)": swap["Shortfall Reduction"],
        "Minimum Forecast Start Margin": min(predicted_margins),
        "Minimum Actual Start Margin": min(actual_margins),
        "Forecast Passes 10-Min Start Margin": (
            min(predicted_margins) >= 10
        ),
        "Actual Passes 10-Min Start Margin": (
            min(actual_margins) >= 10
        )
    })

forecast_swap_risk_df = pd.DataFrame(focus_swap_risks)

# ------------------------------------------------
# 3AH.4 — Display diagnostic summaries
# ------------------------------------------------

print("=" * 65)
print("STEP 3AH — FORECAST READINESS RISK")
print("=" * 65)

print("\nIncoming flights:", len(forecast_readiness_df))

print(
    "Flights with forecasts earlier than actual arrival:",
    int((forecast_readiness_df["Signed Error (min)"] > 0).sum())
)

print(
    "Mean signed arrival-time error:",
    round(forecast_readiness_df["Signed Error (min)"].mean(), 2),
    "minutes"
)

print(
    "Largest underprediction:",
    round(forecast_readiness_df["Underpredicted (min)"].max(), 2),
    "minutes"
)

print("\nFlight-level readiness errors:")

display(
    forecast_readiness_df[
        [
            "TAIL_NUM",
            "PREV_ARR_DELAY",
            "Predicted Delay (min)",
            "Signed Error (min)",
            "Actual Inbound Arrival",
            "Forecast Inbound Arrival"
        ]
    ].sort_values(
        "Signed Error (min)",
        ascending=False
    ).round(2)
)

print("\nSeven selected swaps — predicted versus actual readiness:")

display(forecast_swap_risk_df.round(2))

print(
    "\nSwaps meeting 10-minute start margin using forecasts:",
    int(
        forecast_swap_risk_df[
            "Forecast Passes 10-Min Start Margin"
        ].sum()
    ),
    "of 7"
)

print(
    "Swaps meeting 10-minute start margin using actual arrivals:",
    int(
        forecast_swap_risk_df[
            "Actual Passes 10-Min Start Margin"
        ].sum()
    ),
    "of 7"
)

print("\nSTEP 3AH COMPLETE")
```

```python

# ================================================================
# BUSI6029 — ASSESSMENT 2
# STEP 3AK: FULL-POPULATION AIRCRAFT READINESS FORECASTS
# ================================================================
# Purpose:
# 1. Forecast incoming delays for all 171 eligible blocks.
# 2. Preserve exact block-to-incoming-flight linkage.
# 3. Calculate forecast versus observed ATL readiness.
# 4. Audit prediction errors before candidate generation.
#
# Existing fitted schedule_rf_pipeline only.
# No retraining, optimisation or parameter tuning.
# 2024 results are retrospective diagnostics.
# ================================================================

# ------------------------------------------------
# 3AK.1 — Validate exact block and feature alignment
# ------------------------------------------------

assert len(all_forecast_blocks) == 171
assert len(full_incoming_aj_df) == 171

assert full_incoming_aj_df["BLOCK_ID"].is_unique
assert all_forecast_blocks["BLOCK_ID"].is_unique

assert list(schedule_rf_pipeline.feature_names_in_) == (
    schedule_features
)

# Align incoming flights by their unique block identifier.
incoming_ak = (
    full_incoming_aj_df
    .set_index("BLOCK_ID")
    .loc[all_forecast_blocks["BLOCK_ID"]]
    .reset_index()
)

assert np.array_equal(
    incoming_ak["BLOCK_ID"].to_numpy(),
    all_forecast_blocks["BLOCK_ID"].to_numpy()
)

assert np.array_equal(
    incoming_ak["TAIL_NUM"].to_numpy(),
    all_forecast_blocks["TAIL_NUM"].to_numpy()
)

assert incoming_ak[schedule_features].notna().all().all()

# ------------------------------------------------
# 3AK.2 — Generate existing-model arrival predictions
# ------------------------------------------------

forecast_delay_ak = np.maximum(
    schedule_rf_pipeline.predict(
        incoming_ak[schedule_features]
    ).astype(float),
    0.0
)

# ------------------------------------------------
# 3AK.3 — Calculate predicted and observed readiness
# ------------------------------------------------

forecast_blocks_ak = (
    all_forecast_blocks
    .copy()
    .reset_index(drop=True)
)

forecast_blocks_ak["PRED_INBOUND_DELAY"] = (
    forecast_delay_ak
)

forecast_blocks_ak["OBS_INBOUND_DELAY"] = (
    forecast_blocks_ak["PREV_ARR_DELAY"].astype(float)
)

# Both arrival estimates refer to ATL-local clock minutes.
forecast_blocks_ak["PRED_INBOUND_ARR_MIN"] = (
    forecast_blocks_ak["BLOCK_START_MIN"]
    - forecast_blocks_ak["SCHED_GROUND_MIN"]
    + forecast_blocks_ak["PRED_INBOUND_DELAY"]
)

forecast_blocks_ak["OBS_INBOUND_ARR_MIN"] = (
    forecast_blocks_ak["INCOMING_ARRIVAL_MIN"]
)

forecast_blocks_ak["PRED_READY_MIN"] = (
    forecast_blocks_ak["PRED_INBOUND_ARR_MIN"]
    + min_turnaround
)

forecast_blocks_ak["OBS_READY_MIN"] = (
    forecast_blocks_ak["OBS_INBOUND_ARR_MIN"]
    + min_turnaround
)

forecast_blocks_ak["READINESS_ERROR_MIN"] = (
    forecast_blocks_ak["OBS_READY_MIN"]
    - forecast_blocks_ak["PRED_READY_MIN"]
)

# Positive error = aircraft was ready later than predicted.
forecast_blocks_ak["OPTIMISTIC_ERROR_MIN"] = (
    forecast_blocks_ak["READINESS_ERROR_MIN"]
    .clip(lower=0)
)

# ------------------------------------------------
# 3AK.4 — Check reconstruction against historical data
# ------------------------------------------------

assert np.allclose(
    forecast_blocks_ak["OBS_INBOUND_ARR_MIN"],
    forecast_blocks_ak["BLOCK_START_MIN"]
    - forecast_blocks_ak["SCHED_GROUND_MIN"]
    + forecast_blocks_ak["OBS_INBOUND_DELAY"]
)

assert forecast_blocks_ak[
    ["PRED_READY_MIN", "OBS_READY_MIN"]
].notna().all().all()

# ------------------------------------------------
# 3AK.5 — Summarise readiness-forecast uncertainty
# ------------------------------------------------

error_ak = forecast_blocks_ak["READINESS_ERROR_MIN"]

readiness_summary_ak = pd.DataFrame([
    {
        "Metric": "Eligible blocks forecast",
        "Value": len(forecast_blocks_ak)
    },
    {
        "Metric": "Distinct aircraft",
        "Value": forecast_blocks_ak["TAIL_NUM"].nunique()
    },
    {
        "Metric": "Mean signed readiness error (min)",
        "Value": error_ak.mean()
    },
    {
        "Metric": "Readiness MAE (min)",
        "Value": error_ak.abs().mean()
    },
    {
        "Metric": "Readiness RMSE (min)",
        "Value": np.sqrt(np.mean(error_ak ** 2))
    },
    {
        "Metric": "Blocks ready later than forecast",
        "Value": int((error_ak > 0).sum())
    },
    {
        "Metric": "Blocks ready earlier than forecast",
        "Value": int((error_ak < 0).sum())
    },
    {
        "Metric": "Largest optimistic error (min)",
        "Value": error_ak.max()
    }
])

print("=" * 65)
print("STEP 3AK — FULL-POPULATION READINESS FORECASTS")
print("=" * 65)

print("\nForecast and readiness diagnostic:")
display(readiness_summary_ak.round(2))

print("\nReadiness-error distribution:")
display(
    error_ak.describe(
        percentiles=[0.10, 0.25, 0.50, 0.75, 0.90, 0.95]
    ).round(2).to_frame("Minutes")
)

# ------------------------------------------------
# 3AK.6 — Inspect the largest optimistic errors
# ------------------------------------------------

print("\nAircraft most at risk of optimistic readiness forecasts:")

display(
    forecast_blocks_ak[
        [
            "BLOCK_ID",
            "TAIL_NUM",
            "DEST",
            "BLOCK_START_MIN",
            "OBS_INBOUND_DELAY",
            "PRED_INBOUND_DELAY",
            "PRED_READY_MIN",
            "OBS_READY_MIN",
            "READINESS_ERROR_MIN"
        ]
    ].sort_values(
        "READINESS_ERROR_MIN",
        ascending=False
    ).head(15).round(2)
)

print("\nSTEP 3AK COMPLETE")
```

```python

# ================================================================
# BUSI6029 — ASSESSMENT 2
# STEP 3AL: FORECAST-BASED SWAP CANDIDATE GENERATION
# ================================================================
# Purpose:
# 1. Generate reciprocal block exchanges from forecast readiness.
# 2. Use all 171 eligible ATL return blocks.
# 3. Calculate benefits using predicted delays ONLY.
# 4. Preserve schedule constraints and an extra 10-min margin.
# 5. Audit whether meaningful forecast-based candidates exist.
#
# OBSERVED arrival delays are NOT used to screen candidates.
# No ILP optimisation or realised-benefit evaluation yet.
# ================================================================

from itertools import combinations

# ------------------------------------------------
# 3AL.1 — Specify the illustrative recovery assumptions
# ------------------------------------------------

forecast_extra_margin = 10
forecast_swap_window = 120

forecast_population = (
    forecast_blocks_ak
    .sort_values("BLOCK_ID")
    .reset_index(drop=True)
)

assert len(forecast_population) == 171
assert np.array_equal(
    forecast_population["BLOCK_ID"].to_numpy(),
    np.arange(171)
)

# ------------------------------------------------
# 3AL.2 — Calculate the forecast-only no-swap baseline
# ------------------------------------------------

forecast_population["PRED_BASE_SHORTFALL"] = np.maximum(
    forecast_population["PRED_READY_MIN"]
    - forecast_population["BLOCK_START_MIN"],
    0.0
)

predicted_baseline_shortfall = float(
    forecast_population["PRED_BASE_SHORTFALL"].sum()
)

# Historical baseline retained for later evaluation ONLY.
observed_baseline_reference = float(baseline_shortfall)

# ------------------------------------------------
# 3AL.3 — Screen all distinct-tail block pairs
# ------------------------------------------------

forecast_candidate_records = []

screening_counts = {
    "Distinct-tail pairs within departure window": 0,
    "Both forecasted swap starts meet safety margin": 0,
    "Next-flight and remote buffers also pass": 0,
    "Positive forecasted shortfall benefit": 0
}

for i, j in combinations(range(len(forecast_population)), 2):

    a = forecast_population.iloc[i]
    b = forecast_population.iloc[j]

    if a["TAIL_NUM"] == b["TAIL_NUM"]:
        continue

    if abs(
        a["BLOCK_START_MIN"] - b["BLOCK_START_MIN"]
    ) > forecast_swap_window:
        continue

    screening_counts[
        "Distinct-tail pairs within departure window"
    ] += 1

    # Tail A would operate flight block B.
    start_margin_a = (
        b["BLOCK_START_MIN"]
        - a["PRED_READY_MIN"]
    )

    # Tail B would operate flight block A.
    start_margin_b = (
        a["BLOCK_START_MIN"]
        - b["PRED_READY_MIN"]
    )

    if min(start_margin_a, start_margin_b) < (
        forecast_extra_margin
    ):
        continue

    screening_counts[
        "Both forecasted swap starts meet safety margin"
    ] += 1

    # Each aircraft must return before its own
    # next recorded ATL departure.
    next_margin_a = (
        a["FOLLOWING_DEP_MIN"]
        - b["BLOCK_END_MIN"]
        - min_turnaround
    )

    next_margin_b = (
        b["FOLLOWING_DEP_MIN"]
        - a["BLOCK_END_MIN"]
        - min_turnaround
    )

    # Remote-airport margins belong to flight blocks.
    remote_margin_a = (
        b["REMOTE_GROUND_MIN"] - remote_min_turnaround
    )

    remote_margin_b = (
        a["REMOTE_GROUND_MIN"] - remote_min_turnaround
    )

    all_margins = [
        start_margin_a,
        start_margin_b,
        next_margin_a,
        next_margin_b,
        remote_margin_a,
        remote_margin_b
    ]

    if min(all_margins) < forecast_extra_margin:
        continue

    screening_counts[
        "Next-flight and remote buffers also pass"
    ] += 1

    # Crucially, use the FORECASTED original shortfall,
    # not the observed baseline from earlier steps.
    predicted_before = (
        a["PRED_BASE_SHORTFALL"]
        + b["PRED_BASE_SHORTFALL"]
    )

    predicted_after = (
        max(-start_margin_a, 0.0)
        + max(-start_margin_b, 0.0)
    )

    predicted_benefit = predicted_before - predicted_after

    if predicted_benefit <= 1e-8:
        continue

    screening_counts[
        "Positive forecasted shortfall benefit"
    ] += 1

    forecast_candidate_records.append({
        "Block A Index": int(a["BLOCK_ID"]),
        "Block B Index": int(b["BLOCK_ID"]),
        "Tail A": a["TAIL_NUM"],
        "Tail B": b["TAIL_NUM"],
        "Route A": f"ATL-{a['DEST']}-ATL",
        "Route B": f"ATL-{b['DEST']}-ATL",
        "Predicted Original Shortfall": predicted_before,
        "Predicted Swapped Shortfall": predicted_after,
        "Predicted Benefit": predicted_benefit,
        "Minimum Forecast Scheduling Margin": min(all_margins)
    })

forecast_candidate_df = pd.DataFrame(
    forecast_candidate_records
)

# ------------------------------------------------
# 3AL.4 — Summarise forecast-only opportunities
# ------------------------------------------------

print("=" * 65)
print("STEP 3AL — FORECAST-BASED RECOVERY CANDIDATES")
print("=" * 65)

print("\nEligible blocks:", len(forecast_population))

print(
    "Forecasted no-swap shortfall:",
    round(predicted_baseline_shortfall, 2),
    "minutes"
)

print(
    "Observed no-swap shortfall (evaluation reference):",
    round(observed_baseline_reference, 2),
    "minutes"
)

print("\nForecast-only candidate screening funnel:")

display(
    pd.DataFrame(
        screening_counts.items(),
        columns=["Screening Stage", "Pairs"]
    )
)

print(
    "\nPositive-benefit forecast candidates:",
    len(forecast_candidate_df)
)

# ------------------------------------------------
# 3AL.5 — Inspect forecast-generated candidates
# ------------------------------------------------

if not forecast_candidate_df.empty:

    print("\nHighest forecasted-benefit exchanges:")

    display(
        forecast_candidate_df.sort_values(
            "Predicted Benefit",
            ascending=False
        ).head(15).round(2)
    )

else:

    print(
        "\nNo positive-benefit aircraft block exchanges "
        "identified under the forecast-only assumptions."
    )

print("\nSTEP 3AL COMPLETE")
```

```python

# ================================================================
# BUSI6029 — ASSESSMENT 2
# STEP 3AM: FORECAST-BASED ILP AND HISTORICAL EVALUATION
# ================================================================
# Purpose:
# 1. Optimise the 19 forecast-generated exchange candidates.
# 2. Prevent any aircraft from participating more than once.
# 3. Evaluate selected exchanges using observed inbound times.
# 4. Separate predicted benefit from historical feasibility.
#
# Observed delays are used ONLY after model selection.
# Counterfactual shortfall is a mathematical proxy,
# NOT observed post-swap delays or realised savings.
# ================================================================

# ------------------------------------------------
# 3AM.1 — Build forecast-only integer programme
# ------------------------------------------------

forecast_ilp_df = forecast_candidate_df.copy().reset_index(drop=True)

assert len(forecast_ilp_df) > 0
assert forecast_ilp_df["Predicted Benefit"].gt(0).all()

forecast_aircraft = sorted(
    set(forecast_ilp_df["Tail A"])
    | set(forecast_ilp_df["Tail B"])
)

forecast_aircraft_index = {
    tail: k for k, tail in enumerate(forecast_aircraft)
}

forecast_conflicts = lil_matrix(
    (len(forecast_aircraft), len(forecast_ilp_df)),
    dtype=float
)

for k, swap in forecast_ilp_df.iterrows():
    forecast_conflicts[
        forecast_aircraft_index[swap["Tail A"]], k
    ] = 1

    forecast_conflicts[
        forecast_aircraft_index[swap["Tail B"]], k
    ] = 1

forecast_benefits = forecast_ilp_df[
    "Predicted Benefit"
].to_numpy(dtype=float)

forecast_ilp_result = milp(
    c=-forecast_benefits,
    integrality=np.ones(len(forecast_ilp_df), dtype=int),
    bounds=Bounds(
        np.zeros(len(forecast_ilp_df)),
        np.ones(len(forecast_ilp_df))
    ),
    constraints=LinearConstraint(
        forecast_conflicts.tocsr(),
        lb=np.zeros(len(forecast_aircraft)),
        ub=np.ones(len(forecast_aircraft))
    ),
    options={
        "time_limit": 60,
        "mip_rel_gap": 0.0
    }
)

print("=" * 65)
print("STEP 3AM — FORECAST-BASED ILP EVALUATION")
print("=" * 65)

print("\nSolver status:", forecast_ilp_result.status)
print("Solver message:", forecast_ilp_result.message)

if forecast_ilp_result.status != 0:
    raise RuntimeError("Forecast-based ILP optimum not confirmed.")

forecast_selected_swaps = (
    forecast_ilp_df.loc[forecast_ilp_result.x > 0.5]
    .copy()
    .sort_values("Predicted Benefit", ascending=False)
)

# ------------------------------------------------
# 3AM.2 — Audit selected swaps using observed arrivals
# ------------------------------------------------

forecast_evaluation_records = []

for _, swap in forecast_selected_swaps.iterrows():

    a = forecast_population.iloc[int(swap["Block A Index"])]
    b = forecast_population.iloc[int(swap["Block B Index"])]

    # Tail A operates block B; tail B operates block A.
    actual_margin_a = (
        b["BLOCK_START_MIN"] - a["OBS_READY_MIN"]
    )
    actual_margin_b = (
        a["BLOCK_START_MIN"] - b["OBS_READY_MIN"]
    )

    observed_before = (
        max(a["OBS_READY_MIN"] - a["BLOCK_START_MIN"], 0.0)
        + max(b["OBS_READY_MIN"] - b["BLOCK_START_MIN"], 0.0)
    )

    observed_after_proxy = (
        max(-actual_margin_a, 0.0)
        + max(-actual_margin_b, 0.0)
    )

    observed_proxy_change = (
        observed_before - observed_after_proxy
    )

    # Recheck recorded ATL commitments that were not part
    # of the exchanged original outbound flight.
    extra_overlaps = 0

    for original, assigned in [(a, b), (b, a)]:
        extra_overlaps += int(
            (
                (atl_commitments["TAIL_NUM"] == original["TAIL_NUM"])
                & (
                    atl_commitments["SOURCE_ORDER"]
                    != original["SOURCE_ORDER"]
                )
                & (
                    atl_commitments["CRS_DEP_MINUTES"]
                    > assigned["BLOCK_START_MIN"]
                )
                & (
                    atl_commitments["CRS_DEP_MINUTES"]
                    < assigned["BLOCK_END_MIN"]
                )
            ).sum()
        )

    historical_start_pass = bool(
        min(actual_margin_a, actual_margin_b)
        >= forecast_extra_margin - 1e-8
    )

    forecast_evaluation_records.append({
        "Tail A": swap["Tail A"],
        "Tail B": swap["Tail B"],
        "Predicted Benefit": float(swap["Predicted Benefit"]),
        "Minimum Forecast Margin": float(
            swap["Minimum Forecast Scheduling Margin"]
        ),
        "Minimum Observed Start Margin": float(
            min(actual_margin_a, actual_margin_b)
        ),
        "Observed Start Passes 10 Min": historical_start_pass,
        "Other ATL Departure Conflicts": extra_overlaps,
        "Historically Screened Feasible": bool(
            historical_start_pass and extra_overlaps == 0
        ),
        "Observed Before Shortfall": float(observed_before),
        "Observed After Shortfall Proxy": float(
            observed_after_proxy
        ),
        "Observed Shortfall Proxy Change": float(
            observed_proxy_change
        )
    })

forecast_evaluation_df = pd.DataFrame(
    forecast_evaluation_records
)

# ------------------------------------------------
# 3AM.3 — Recalculate full-population proxy totals
# ------------------------------------------------

predicted_selected_benefit = float(
    forecast_selected_swaps["Predicted Benefit"].sum()
)

observed_selected_proxy_change = float(
    forecast_evaluation_df[
        "Observed Shortfall Proxy Change"
    ].sum()
)

historically_feasible_count = int(
    forecast_evaluation_df[
        "Historically Screened Feasible"
    ].sum()
)

historically_feasible_predicted_benefit = float(
    forecast_evaluation_df.loc[
        forecast_evaluation_df["Historically Screened Feasible"],
        "Predicted Benefit"
    ].sum()
)

# Check no aircraft was assigned to multiple exchanges.
used_forecast_tails = (
    forecast_selected_swaps["Tail A"].tolist()
    + forecast_selected_swaps["Tail B"].tolist()
)

assert len(used_forecast_tails) == len(set(used_forecast_tails))

# ------------------------------------------------
# 3AM.4 — Display the forecast versus history comparison
# ------------------------------------------------

forecast_optimisation_summary_df = pd.DataFrame([
    {
        "Metric": "Forecast-generated candidate pairs",
        "Value": len(forecast_ilp_df)
    },
    {
        "Metric": "Optimal forecast-selected swaps",
        "Value": len(forecast_selected_swaps)
    },
    {
        "Metric": "Predicted no-swap shortfall (min)",
        "Value": predicted_baseline_shortfall
    },
    {
        "Metric": "Predicted selected benefit (min)",
        "Value": predicted_selected_benefit
    },
    {
        "Metric": "Observed no-swap shortfall (min)",
        "Value": baseline_shortfall
    },
    {
        "Metric": "Selected swaps passing observed scheduling checks",
        "Value": historically_feasible_count
    },
    {
        "Metric": "Predicted benefit of historically feasible swaps",
        "Value": historically_feasible_predicted_benefit
    },
    {
        "Metric": "Observed-arrival shortfall proxy change (min)",
        "Value": observed_selected_proxy_change
    }
])

print("\nForecast-based optimisation summary:")
display(forecast_optimisation_summary_df.round(2))

print("\nSelected exchanges and historical evaluation:")

display(
    forecast_evaluation_df.sort_values(
        "Predicted Benefit", ascending=False
    ).round(2)
)

print("\nUnique aircraft assignment check: PASSED")
print("\nSTEP 3AM COMPLETE")
```

```python

# ================================================================
# BUSI6029 — ASSESSMENT 2
# STEP 3AN: FEASIBILITY-ADJUSTED OUTCOME AUDIT
# ================================================================
# Purpose:
# 1. Separate historically feasible and infeasible proposals.
# 2. Quantify forecast benefit that failed feasibility checks.
# 3. Calculate observed proxy change for feasible swaps only.
# 4. Produce report-ready evaluation figures.
#
# No retraining, optimisation or new candidate selection.
# This is retrospective evaluation, NOT realised savings.
# ================================================================

import numpy as np
import pandas as pd

audit_an = forecast_evaluation_df.copy()

# ------------------------------------------------
# 3AN.1 — Verify consistency of historical screening
# ------------------------------------------------

audit_an["Rechecked Feasible"] = (
    audit_an["Minimum Observed Start Margin"].ge(
        forecast_extra_margin - 1e-8
    )
    & audit_an["Other ATL Departure Conflicts"].eq(0)
)

assert (
    audit_an["Rechecked Feasible"].to_numpy()
    == audit_an["Historically Screened Feasible"].to_numpy()
).all()

feasible_an = audit_an.loc[
    audit_an["Rechecked Feasible"]
].copy()

infeasible_an = audit_an.loc[
    ~audit_an["Rechecked Feasible"]
].copy()

# ------------------------------------------------
# 3AN.2 — Separate expected and observed proxy changes
# ------------------------------------------------

expected_total_an = float(
    audit_an["Predicted Benefit"].sum()
)

expected_feasible_an = float(
    feasible_an["Predicted Benefit"].sum()
)

expected_infeasible_an = float(
    infeasible_an["Predicted Benefit"].sum()
)

observed_feasible_change_an = float(
    feasible_an["Observed Shortfall Proxy Change"].sum()
)

observed_all_change_an = float(
    audit_an["Observed Shortfall Proxy Change"].sum()
)

# Calculate proportions using the appropriate baselines.
predicted_pct_an = (
    100 * expected_total_an / predicted_baseline_shortfall
    if predicted_baseline_shortfall > 0 else np.nan
)

observed_feasible_pct_an = (
    100 * observed_feasible_change_an / baseline_shortfall
    if baseline_shortfall > 0 else np.nan
)

# ------------------------------------------------
# 3AN.3 — Build report-ready summary
# ------------------------------------------------

final_comparison_an = pd.DataFrame([
    {
        "Metric": "Forecast-selected swaps",
        "Value": len(audit_an)
    },
    {
        "Metric": "Historical screening passes",
        "Value": len(feasible_an)
    },
    {
        "Metric": "Historical screening failures",
        "Value": len(infeasible_an)
    },
    {
        "Metric": "Forecasted baseline shortfall (min)",
        "Value": predicted_baseline_shortfall
    },
    {
        "Metric": "Forecasted improvement (min)",
        "Value": expected_total_an
    },
    {
        "Metric": "Forecasted baseline improvement (%)",
        "Value": predicted_pct_an
    },
    {
        "Metric": "Predicted benefit allocated to passing swaps (min)",
        "Value": expected_feasible_an
    },
    {
        "Metric": "Predicted benefit allocated to failing swaps (min)",
        "Value": expected_infeasible_an
    },
    {
        "Metric": "Historical baseline shortfall (min)",
        "Value": baseline_shortfall
    },
    {
        "Metric": "Observed proxy improvement from passing swaps (min)",
        "Value": observed_feasible_change_an
    },
    {
        "Metric": "Observed baseline improvement from passing swaps (%)",
        "Value": observed_feasible_pct_an
    },
    {
        "Metric": "Observed proxy change for all proposals (min)",
        "Value": observed_all_change_an
    }
])

# ------------------------------------------------
# 3AN.4 — Check numerical reconciliation
# ------------------------------------------------

assert np.isclose(
    expected_total_an,
    expected_feasible_an + expected_infeasible_an
)

assert np.isclose(
    observed_all_change_an,
    audit_an["Observed Shortfall Proxy Change"].sum()
)

# ------------------------------------------------
# 3AN.5 — Display results
# ------------------------------------------------

print("=" * 65)
print("STEP 3AN — FEASIBILITY-ADJUSTED FINAL AUDIT")
print("=" * 65)

print("\nForecast versus historical evaluation:")
display(final_comparison_an.round(2))

print("\nHistorically passing exchanges:")
display(
    feasible_an[
        [
            "Tail A",
            "Tail B",
            "Predicted Benefit",
            "Minimum Observed Start Margin",
            "Observed Shortfall Proxy Change"
        ]
    ].round(2)
)

print("\nHistorically failing exchanges:")
display(
    infeasible_an[
        [
            "Tail A",
            "Tail B",
            "Predicted Benefit",
            "Minimum Observed Start Margin",
            "Observed Shortfall Proxy Change"
        ]
    ].round(2)
)

print("\nSTEP 3AN COMPLETE")
```

```python

# ================================================================
# BUSI6029 — ASSESSMENT 2
# STEP 3AO: FORECAST SAFETY-MARGIN SENSITIVITY
# ================================================================
# Purpose:
# 1. Reoptimise forecast candidate swaps at larger margins.
# 2. Keep forecast inputs separate from historical outcomes.
# 3. Measure forecast benefit versus observed feasibility.
# 4. Quantify the reliability–benefit trade-off.
#
# This is illustrative sensitivity analysis, not
# statistical calibration or realised delay savings.
# ================================================================

import numpy as np
import pandas as pd
from scipy.optimize import milp, Bounds, LinearConstraint
from scipy.sparse import lil_matrix

buffers_ao = [10, 15, 20, 30, 45, 60]

assert len(forecast_candidate_df) == 19
assert len(forecast_population) == 171

# ------------------------------------------------
# 3AO.1 — Audit every candidate using observed arrivals
# ------------------------------------------------

candidate_audit_ao = forecast_candidate_df.copy().reset_index(drop=True)

observed_start_margins = []
observed_proxy_changes = []
observed_other_margins = []
observed_atl_conflicts = []

for _, swap in candidate_audit_ao.iterrows():

    a = forecast_population.iloc[int(swap["Block A Index"])]
    b = forecast_population.iloc[int(swap["Block B Index"])]

    # Historical start margins for exchanged aircraft.
    start_a = b["BLOCK_START_MIN"] - a["OBS_READY_MIN"]
    start_b = a["BLOCK_START_MIN"] - b["OBS_READY_MIN"]

    # Schedule-only margins unaffected by incoming forecasts.
    remaining_margins = [
        a["FOLLOWING_DEP_MIN"] - b["BLOCK_END_MIN"]
        - min_turnaround,

        b["FOLLOWING_DEP_MIN"] - a["BLOCK_END_MIN"]
        - min_turnaround,

        b["REMOTE_GROUND_MIN"] - remote_min_turnaround,

        a["REMOTE_GROUND_MIN"] - remote_min_turnaround
    ]

    # Historical shortfall proxy.
    before = (
        max(a["OBS_READY_MIN"] - a["BLOCK_START_MIN"], 0)
        + max(b["OBS_READY_MIN"] - b["BLOCK_START_MIN"], 0)
    )

    after = max(-start_a, 0) + max(-start_b, 0)

    observed_start_margins.append(min(start_a, start_b))
    observed_other_margins.append(min(remaining_margins))
    observed_proxy_changes.append(before - after)

    # Preserve recorded ATL flight commitment check.
    conflicts = 0

    for original, assigned in [(a, b), (b, a)]:

        conflicts += int((
            (
                atl_commitments["TAIL_NUM"]
                == original["TAIL_NUM"]
            )
            & (
                atl_commitments["SOURCE_ORDER"]
                != original["SOURCE_ORDER"]
            )
            & (
                atl_commitments["CRS_DEP_MINUTES"]
                > assigned["BLOCK_START_MIN"]
            )
            & (
                atl_commitments["CRS_DEP_MINUTES"]
                < assigned["BLOCK_END_MIN"]
            )
        ).sum())

    observed_atl_conflicts.append(conflicts)

candidate_audit_ao["Observed Start Margin"] = (
    observed_start_margins
)

candidate_audit_ao["Observed Other Margin"] = (
    observed_other_margins
)

candidate_audit_ao["Observed Proxy Change"] = (
    observed_proxy_changes
)

candidate_audit_ao["ATL Conflicts"] = (
    observed_atl_conflicts
)

# ------------------------------------------------
# 3AO.2 — Reoptimise separately at each margin
# ------------------------------------------------

sensitivity_rows_ao = []
selected_by_buffer_ao = {}

for buffer in buffers_ao:

    # The original 19 candidates all satisfied 10 minutes.
    # Higher margins only REMOVE candidates; they cannot
    # introduce new positive-benefit candidate pairs.
    candidates = candidate_audit_ao.loc[
        candidate_audit_ao[
            "Minimum Forecast Scheduling Margin"
        ] >= buffer - 1e-8
    ].copy().reset_index(drop=True)

    if candidates.empty:
        selected = candidates.copy()

    else:
        aircraft = sorted(
            set(candidates["Tail A"])
            | set(candidates["Tail B"])
        )

        aircraft_map = {
            tail: i for i, tail in enumerate(aircraft)
        }

        conflicts = lil_matrix(
            (len(aircraft), len(candidates)),
            dtype=float
        )

        for j, swap in candidates.iterrows():
            conflicts[aircraft_map[swap["Tail A"]], j] = 1
            conflicts[aircraft_map[swap["Tail B"]], j] = 1

        benefits = candidates[
            "Predicted Benefit"
        ].to_numpy(dtype=float)

        result = milp(
            c=-benefits,
            integrality=np.ones(len(candidates)),
            bounds=Bounds(
                np.zeros(len(candidates)),
                np.ones(len(candidates))
            ),
            constraints=LinearConstraint(
                conflicts.tocsr(),
                lb=np.zeros(len(aircraft)),
                ub=np.ones(len(aircraft))
            ),
            options={"mip_rel_gap": 0.0}
        )

        assert result.status == 0, (
            f"ILP did not confirm optimum at margin {buffer}"
        )

        selected = candidates.loc[
            result.x > 0.5
        ].copy()

    selected["Historical Pass"] = (
        selected["Observed Start Margin"].ge(buffer - 1e-8)
        & selected["Observed Other Margin"].ge(buffer - 1e-8)
        & selected["ATL Conflicts"].eq(0)
    )

    selected_by_buffer_ao[buffer] = selected.copy()

    passing = selected.loc[selected["Historical Pass"]]

    predicted_benefit = float(
        selected["Predicted Benefit"].sum()
    )

    observed_passing_change = float(
        passing["Observed Proxy Change"].sum()
    )

    sensitivity_rows_ao.append({
        "Extra Margin (min)": buffer,
        "Forecast Candidates": len(candidates),
        "Optimal Swaps": len(selected),
        "Predicted Benefit (min)": predicted_benefit,
        "Predicted Baseline Reduction (%)": (
            100 * predicted_benefit
            / predicted_baseline_shortfall
        ),
        "Historically Passing Swaps": len(passing),
        "Historical Pass Rate (%)": (
            100 * len(passing) / len(selected)
            if len(selected) else np.nan
        ),
        "Observed Proxy Change, Passing Only (min)":
            observed_passing_change
    })

forecast_sensitivity_ao_df = pd.DataFrame(
    sensitivity_rows_ao
)

# ------------------------------------------------
# 3AO.3 — Verify the original 10-minute scenario
# ------------------------------------------------

original_ao = forecast_sensitivity_ao_df.iloc[0]

assert original_ao["Optimal Swaps"] == 8
assert np.isclose(
    original_ao["Predicted Benefit (min)"],
    81.82,
    atol=0.02
)
assert original_ao["Historically Passing Swaps"] == 4

# ------------------------------------------------
# 3AO.4 — Display sensitivity results
# ------------------------------------------------

print("=" * 65)
print("STEP 3AO — FORECAST SAFETY-MARGIN SENSITIVITY")
print("=" * 65)

print("\nReoptimised forecast-driven solutions:")
display(forecast_sensitivity_ao_df.round(2))

print("\nOriginal 10-minute scenario: RECONCILED")

print(
    "\nImportant: zero swaps is not evidence "
    "of a successful recovery intervention."
)

print("\nSTEP 3AO COMPLETE")
```

### Step 3AQ — Integrated Predictive and Prescriptive Findings

The study developed and evaluated an integrated predictive–prescriptive framework for analysing Delta Air Lines' arrival delays and illustrative aircraft recovery opportunities. Predictive modelling used historical 2023–2025 flight records and a temporally separated evaluation strategy. The selected recovery-enhanced Random Forest achieved a mean absolute error (MAE) of 12.76 minutes on the 2025 holdout data.

For the prescriptive component, 171 eligible aircraft rotation blocks from 16 January 2024 were evaluated using binary integer programming. The retrospective hindsight optimisation selected 17 aircraft exchanges, producing a modelled turnaround-shortfall reduction of 361 minutes (19.88%) against a historical baseline of 1,816 minutes. To investigate predictive–prescriptive integration, an existing schedule-only Random Forest estimated incoming-aircraft arrival delays for all 171 eligible blocks. The forecast-based readiness estimates showed a mean absolute error of 32.58 minutes and a maximum optimistic error of 248.82 minutes within this retrospectively selected sample. The model substantially underestimated disruption severity, producing a forecast-derived baseline shortfall of 232.78 minutes compared with the observed 1,816 minutes.

Forecast-based candidate screening identified 19 positive-benefit aircraft exchanges. Binary integer programming selected eight non-conflicting exchanges, with a predicted improvement of 81.82 minutes (35.15% of the forecast-derived baseline). However, only four exchanges passed the subsequent historical scheduling checks, and these four produced no observed turnaround-shortfall proxy improvement. Safety-margin sensitivity analysis further showed that increasing the additional scheduling margin reduced the number of forecast-driven opportunities without consistently improving historical feasibility; no exchanges were selected at margins of 30 minutes or greater.

These findings indicate that deterministic optimisation based on schedule-only delay forecasts can generate attractive mathematical solutions while remaining vulnerable to severe forecast errors. The retrospective benefits represent modelled scheduling potential rather than demonstrated operational savings. Practical implementation would require more timely aircraft-arrival information, uncertainty-aware optimisation and verification of aircraft compatibility, crew, maintenance and wider network constraints.
